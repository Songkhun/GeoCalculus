# ส่วน A: ลิมิตและความต่อเนื่อง (Limits and Continuity)

⭐ ง่าย · ตัวอย่างที่ 1–4 · สัปดาห์ที่ 1–2 ในประมวลรายวิชา

## ตัวอย่างที่ 1 ⭐: ความเร็วของสถานี GPS หลังเกิดแผ่นดินไหว (ลิมิตของผลหารผลต่าง)

**บริบท** หลังแผ่นดินไหวขนาดใหญ่ สถานี GPS ยังคงเคลื่อนที่ต่อไปอีกหลายปีเพราะ *การเลื่อนหลังแผ่นดินไหว (afterslip)* บนรอยเลื่อน
แบบจำลองที่ใช้กันทั่วไปสำหรับระยะเลื่อนไปทางทิศตะวันออก (mm) ของสถานี เมื่อเวลาผ่านไป $t$ ปีหลังเหตุการณ์ คือ

$$x(t) = 50\,\ln\!\left(1+\frac{t}{0.5}\right).$$

**คำถาม** ณ เวลา $t = 1$ ปีพอดี สถานีเคลื่อนที่เร็วเท่าใด

**แคลคูลัส** ในช่วงเวลาสั้น ๆ GPS ให้ความเร็ว *เฉลี่ย* ซึ่งคือความชันของเส้นตัด (secant slope)
$\dfrac{x(1+h)-x(1)}{h}$ ส่วนความเร็ว *ขณะหนึ่ง* คือลิมิตของค่านี้เมื่อ $h\to 0$ โดยคำนวณทั้งจากทางซ้าย ($h<0$) และทางขวา ($h>0$)
ในรูปแบบตารางเดียวกับในตำรา (บทลิมิตของฟังก์ชัน)

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** สถานี GPS ให้ตำแหน่งเฉพาะบางช่วงเวลา ความเร็วที่หาได้ง่ายที่สุดจึงเป็น **ความเร็วเฉลี่ย** คือการเปลี่ยนตำแหน่งหารด้วยช่วงเวลา ซึ่งก็คือความชันของเส้นตัดโค้ง (secant line) ส่วน **ความเร็วขณะหนึ่ง** คือค่าที่ความเร็วเฉลี่ยเข้าใกล้เมื่อช่วงเวลา $h$ หดลงเข้าใกล้ศูนย์ ลิมิตนี้ก็คือนิยามของอนุพันธ์ (derivative) นั่นเอง

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. เขียนความเร็วเฉลี่ยระหว่าง $t_0$ ถึง $t_0+h$ เป็นผลหารเชิงผลต่าง (difference quotient): $$\bar v(h)=\frac{x(t_0+h)-x(t_0)}{h},\qquad x(t)=50\ln\!\left(1+\frac{t}{0.5}\right)$$
2. คำนวณที่ $t_0=1$ ปี โดยให้ $h$ เล็กลงเรื่อย ๆ เมื่อ $h=0.5$ ได้ $\bar v=28.77$ มม./ปี $h=0.1$ ได้ $32.27$ $h=0.01$ ได้ $33.22$ และ $h=0.001$ ได้ $33.32$ มม./ปี ถ้าใช้ $h<0$ (ช่วงเวลาก่อน $t_0$) จะได้ $40.55,\ 34.50,\ 33.44,\ 33.34$ มม./ปี
3. ค่าจากทั้งสองด้านบีบเข้าหาค่าเดียวกันจากคนละฝั่ง ลิมิตสองข้างจึงมีค่า ค่าจากด้านขวาน้อยเกินไปและค่าจากด้านซ้ายมากเกินไป เพราะกราฟโค้งลง (เว้าลง)
4. หาค่าลิมิตที่แท้จริงด้วยหลักเกณฑ์ลูกโซ่ (chain rule): $$v(t)=x'(t)=50\cdot\frac{1}{1+t/0.5}\cdot\frac{1}{0.5}=\frac{50}{0.5+t}$$
5. แทน $t_0=1$ ได้ $v(1)=50/1.5=33.33$ มม./ปี ตรงกับแนวโน้มในตาราง ในแอปค่าเริ่มต้น $h\approx0.50$ ให้ความชันเส้นตัดโค้ง $28.76$ และ $40.57$ มม./ปี คลาดเคลื่อนราว $4.6$ มม./ปี และความคลาดเคลื่อนนี้หายไปเมื่อ $h\to0$

```python
x = lambda t: 50*np.log(1 + t/0.5)          # displacement in mm
t0 = 1.0
print(f"{'h':>10} {'avg velocity (mm/yr), h<0':>28} {'avg velocity (mm/yr), h>0':>28}")
for h in [0.5, 0.1, 0.01, 0.001, 0.0001]:
    left  = (x(t0 - h) - x(t0)) / (-h)
    right = (x(t0 + h) - x(t0)) / h
    print(f"{h:>10} {left:>28.6f} {right:>28.6f}")

# Exact answer with SymPy: limit of the difference quotient
t, h = sp.symbols('t h', real=True)
X = 50*sp.log(1 + t/sp.Rational(1, 2))
v_exact = sp.limit((X.subs(t, 1 + h) - X.subs(t, 1))/h, h, 0)
show(r"\lim_{h\to0}\frac{x(1+h)-x(1)}{h}", v_exact)
print("≈", float(v_exact), "mm/yr")
```

```python
tt = np.linspace(0, 3, 300)
fig, ax = plt.subplots()
ax.plot(tt, x(tt), lw=2, label="x(t) = 50 ln(1 + t/0.5)")
for hh, c in [(1.5, "C3"), (0.5, "C2"), (0.1, "C1")]:
    slope = (x(t0+hh) - x(t0))/hh
    ax.plot(tt, x(t0) + slope*(tt - t0), "--", c=c, lw=1, label=f"secant, h = {hh}: {slope:.1f} mm/yr")
ax.plot(tt, x(t0) + float(v_exact)*(tt - t0), "k", lw=1.2, label=f"tangent (limit): {float(v_exact):.2f} mm/yr")
ax.plot(t0, x(t0), "ko")
ax.set(xlabel="time after earthquake (yr)", ylabel="displacement (mm)", ylim=(0, 90),
       title="Secant slopes approach the tangent slope as h → 0")
ax.legend(fontsize=9); plt.show()
```

**การตีความ** ลิมิตทางเดียวทั้งสองด้านเข้าใกล้ **33.33 mm/yr** ดังนั้นลิมิตหาค่าได้ และค่านี้คือความเร็วขณะหนึ่ง
สถานีเคลื่อนที่ช้าลงตามเวลา (ความชันลดลงเรื่อย ๆ) ซึ่งเป็นลักษณะของการเลื่อนหลังแผ่นดินไหวที่เห็นในข้อมูล GPS

**ความหมายของคำตอบ** หนึ่งปีหลังแผ่นดินไหว สถานียังเคลื่อนที่ราว **33 มม./ปี** ซึ่งพอ ๆ กับความเร็วของแผ่นเปลือกโลก การเลื่อนหลังแผ่นดินไหว (afterslip) จึงเป็นสัญญาณขนาดใหญ่ในข้อมูล GPS เนื่องจาก $v=50/(0.5+t)$ ลดลงตามเวลา สถานีจึงเคลื่อนช้าลง ทันทีหลังแผ่นดินไหว ($t=0$) สถานีเคลื่อนที่ 100 มม./ปี นักมาตรวิทยาธรณีจึงใช้เส้นโค้งแบบนี้แยกการเลื่อนหลังแผ่นดินไหวออกจากการเคลื่อนที่คงที่ของแผ่นเปลือกโลก

**ข้อผิดพลาดที่พบบ่อย**

- แทน $h=0$ ลงในผลหารเชิงผลต่างโดยตรงจะได้ $0/0$ ซึ่งไม่นิยาม ลิมิตพูดถึงสิ่งที่เกิดขึ้นเมื่อ $h$ เข้าใกล้ 0 ไม่ใช่ที่ $h=0$
- ลืมคูณอนุพันธ์ของฟังก์ชันข้างใน $1/0.5$ ตามหลักเกณฑ์ลูกโซ่ แล้วเขียน $v=50/(1+t/0.5)$ ซึ่งน้อยไป 2 เท่า
- ตรวจเพียงด้านเดียว ($h>0$) ลิมิตจะมีค่าก็ต่อเมื่อลิมิตซ้ายและลิมิตขวาเท่ากัน

**ลองทำดู** หาความเร็วที่ $t = 0$ และ $t = 5$ yr สถานีเคลื่อนที่เร็วกว่ากี่เท่าในช่วงทันทีหลังแผ่นดินไหว

## ตัวอย่างที่ 2 ⭐: เม็ดทรายจมเร็วแค่ไหน (ลิมิตที่อนันต์ (limit at infinity))

**บริบท** เม็ดควอตซ์ที่ปล่อยลงในน้ำนิ่งจะเร่งความเร็วจนแรงต้านสมดุลกับน้ำหนัก ความเร็วของมันเข้าใกล้**ความเร็วตกตะกอนสุดท้าย (terminal settling velocity)** จากกฎของสโตกส์ (Stokes' law):

$$v_t = \frac{(\rho_s-\rho_w)\,g\,D^2}{18\mu},\qquad v(t) = v_t\left(1-e^{-t/\tau}\right),\quad \tau = \frac{v_t}{g}.$$

ควอตซ์ $\rho_s = 2650$ kg/m³ น้ำ $\rho_w = 1000$ kg/m³ ความหนืด $\mu = 1.0\times10^{-3}$ Pa·s เส้นผ่านศูนย์กลางเม็ด $D = 0.1$ mm (ทรายละเอียดมาก)

**แคลคูลัส** $\displaystyle\lim_{t\to\infty} v(t) = v_t\left(1 - \lim_{t\to\infty}e^{-t/\tau}\right) = v_t$ เพราะ $e^{-t/\tau}\to0$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** เม็ดตะกอนที่จมลงในน้ำไม่มีวันถึงความเร็วสุดท้ายในเวลาจำกัด ความเร็วเพียง *เข้าใกล้* ค่านั้นเรื่อย ๆ **ลิมิตที่อนันต์ (limit at infinity)** เปลี่ยนการเข้าใกล้อย่างไม่สิ้นสุดนี้ให้เป็นตัวเลขเดียว คือความเร็วปลาย (terminal velocity) $v_t$ ซึ่งเป็นความเร็วที่ควบคุมการตกตะกอนจริง

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **เขียนแบบจำลอง** ความเร็วของเม็ดตะกอนคือ $v(t) = v_t\left(1-e^{-t/\tau}\right)$ เมื่อ $t=0$ ความเร็วเป็น $0$ และเพิ่มขึ้นเมื่อพจน์เอกซ์โพเนนเชียลเล็กลง
2. **หาลิมิต** เนื่องจาก $\tau>0$ เมื่อ $t\to\infty$ จะได้ $-t/\tau\to-\infty$ ดังนั้น $e^{-t/\tau}\to0$ ใช้สมบัติของลิมิตได้ $$\lim_{t\to\infty}v(t)=v_t\left(1-\lim_{t\to\infty}e^{-t/\tau}\right)=v_t(1-0)=v_t$$
3. **คำนวณ $v_t$ จากกฎของสโตกส์ (Stokes' law)** โดย $D = 0.1$ mm $=10^{-4}$ m $$v_t=\frac{(2650-1000)(9.81)(10^{-4})^2}{18(1.0\times10^{-3})}=\frac{1.619\times10^{-4}}{0.018}\approx 8.99\times10^{-3}\ \text{m/s}=8.99\ \text{mm/s}$$
4. **เข้าใกล้ลิมิตเร็วแค่ไหน** มาตราส่วนเวลาคือ $\tau=v_t/g=0.00899/9.81\approx0.917$ ms หลังผ่านไป $5\tau$ (ไม่ถึง 5 ms) ตัวประกอบ $e^{-5}\approx0.007$ เม็ดตะกอนจึงมีความเร็วต่างจาก $v_t$ ไม่ถึง 1 %
5. **นำลิมิตไปใช้** เนื่องจากช่วงเร่งความเร็วสั้นมาก เวลาที่ใช้จมผ่านน้ำลึก 10 m จึงเท่ากับ $10/0.00899\approx1112$ s $\approx18.5$ นาที
6. **ตรวจสอบสมมติฐาน** กฎของสโตกส์ใช้ได้เมื่อเลขเรย์โนลด์ส (Reynolds number) $Re=v_tD\rho_w/\mu<1$ ในที่นี้ $Re=0.00899\times10^{-4}\times1000/10^{-3}\approx0.90$ ผลลัพธ์จึงใช้ได้

```python
t, tau, vt_s = sp.symbols('t tau v_t', positive=True)
v = vt_s*(1 - sp.exp(-t/tau))
show(r"\lim_{t\to\infty}v(t)", sp.limit(v, t, sp.oo))

rho_s, rho_w, mu, g = 2650, 1000, 1.0e-3, 9.81
vt = lambda D: (rho_s - rho_w)*g*D**2/(18*mu)
for D_mm in (0.05, 0.1, 0.2):
    D = D_mm/1000; v_t = vt(D); Re = v_t*D/(mu/rho_w)
    print(f"D = {D_mm} mm: v_t = {v_t*1000:6.2f} mm/s, tau = {v_t/g*1000:.3f} ms, "
          f"time to settle 10 m = {10/v_t/60:6.1f} min, Reynolds number = {Re:.2f}")
```

```python
D = 0.1e-3; v_t = vt(D); tau_v = v_t/g
tt = np.linspace(0, 6*tau_v, 300)
fig, ax = plt.subplots()
ax.plot(tt*1000, v_t*(1 - np.exp(-tt/tau_v))*1000, lw=2, label="v(t)")
ax.axhline(v_t*1000, c="C3", ls="--", label=f"limit v_t = {v_t*1000:.2f} mm/s")
ax.set(xlabel="time (ms)", ylabel="velocity (mm/s)", title="The grain reaches terminal velocity in about a millisecond"); ax.legend(); plt.show()
```

**การตีความ** เม็ดขนาด 0.1 mm ถึงความเร็วสุดท้ายประมาณ **9.0 mm/s** ภายในไม่กี่มิลลิวินาที ดังนั้นในทางปฏิบัติมันจมด้วยความเร็ว $v_t$ และใช้เวลาประมาณ 18.5 นาทีในการตกผ่านน้ำลึก 10 m
เนื่องจาก $v_t\propto D^2$ การลดขนาดเม็ดลงครึ่งหนึ่งทำให้จมช้าลง 4 เท่า นี่คือเหตุที่แม่น้ำและทะเลคัดแยกตะกอนตามขนาด โดยโคลนถูกพาไปไกลจากฝั่ง
(กฎของสโตกส์สมมติว่าเลขเรย์โนลด์ส (Reynolds number) < 1 ดังนั้นผลสำหรับ 0.2 mm ซึ่ง Re ≈ 7 จึงเป็นค่าที่สูงเกินจริง)

**ความหมายของคำตอบ** เม็ดทรายละเอียดมากจมด้วยความเร็วประมาณ 9 mm/s และใช้เวลาราว 18 นาทีในการจมผ่านน้ำนิ่งลึก 10 m เนื่องจาก $v_t\propto D^2$ เม็ดที่เล็กลงครึ่งหนึ่งจะจมช้าลงสี่เท่า แม่น้ำและทะเลจึงคัดขนาดตะกอน และพาโคลไปได้ไกลจากฝั่ง ตรวจหน่วย: (kg/m³)(m/s²)(m²)/(Pa·s) = m/s

**ข้อผิดพลาดที่พบบ่อย**

- ลืมแปลงเส้นผ่านศูนย์กลางจาก mm เป็น m ก่อนยกกำลังสอง ถ้าใช้ $D=0.1$ แทน $10^{-4}$ จะได้ $v_t$ ใหญ่เกินจริงล้านเท่า
- คิดว่าลิมิต \"ไปไม่ถึงจึงไม่มีประโยชน์\" ที่จริงการเข้าใกล้เป็นแบบเอกซ์โพเนนเชียล หลังผ่านไปไม่กี่เท่าของ $\tau$ ผลต่างก็เล็กจนละเลยได้
- ใช้กฎของสโตกส์กับเม็ดหยาบที่ $Re>1$ เช่น $D=0.2$ mm ได้ $Re\approx7$ สูตรนี้จะให้ความเร็วสูงเกินจริง

**ลองทำดู** อนุภาคดินเหนียวมี $D \approx 0.002$ mm จะใช้เวลานานเท่าใดในการตกตะกอนผ่านน้ำนิ่งลึก 10 m

## ตัวอย่างที่ 3 ⭐: อุณหภูมิในเปลือกโลกสองชั้น (ความต่อเนื่องของฟังก์ชันนิยามเป็นช่วง)

**บริบท** อุณหภูมิเพิ่มขึ้นตามความลึก $z$ (km) ในเปลือกโลกที่ประกอบด้วยหินสองชั้น แต่ละชั้นมีเกรเดียนต์ความร้อนใต้พิภพ (geothermal gradient) ของตัวเอง:

$$T(z)=\begin{cases} 15 + 25z, & 0 \le z \le 10 \quad\text{(sedimentary layer)}\\ c + 15z, & z > 10 \quad\text{(granitic basement)}\end{cases}\qquad(^\circ\mathrm{C})$$

**คำถาม**
1. อุณหภูมิไม่สามารถกระโดดที่รอยสัมผัสระหว่างชั้นหินได้ จงหา $c$ ที่ทำให้ $T$ **ต่อเนื่อง** ที่ $z = 10$ km
2. ลิมิตทางเดียวของ **เกรเดียนต์** $T'(z)$ ที่ $z=10$ เท่ากันหรือไม่ และมีความหมายทางกายภาพอย่างไร

**แคลคูลัส** $T$ ต่อเนื่องที่ $a$ ถ้า $\displaystyle\lim_{z\to a^-}T(z)=\lim_{z\to a^+}T(z)=T(a)$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ฟังก์ชันที่นิยามเป็นช่วง (piecewise function) จะต่อเนื่องที่จุดเชื่อมได้ก็ต่อเมื่อลิมิตซ้าย ลิมิตขวา และค่าฟังก์ชันเท่ากันทั้งหมด อุณหภูมิเป็นปริมาณทางกายภาพที่กระโดดไม่ได้ภายในโลก เราจึง **เลือก** ค่าคงตัว $c$ ให้สองช่วงมาบรรจบกัน ส่วนอนุพันธ์ (ความชันอุณหภูมิ) กระโดดได้ และการตรวจลิมิตข้างเดียวของมันจะแสดงให้เห็นจุดหักมุม (kink)

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. เขียนแบบจำลองด้วยค่าเริ่มต้น $g_1=25$ °C/กม. $g_2=15$ °C/กม. และความลึกรอยต่อ $z_c=10$ กม.: $$T(z)=\begin{cases}15+25z,& z\le10\\ c+15z,& z>10\end{cases}$$
2. ลิมิตซ้าย (จากชั้นตะกอนด้านบน): $\lim_{z\to10^-}T=15+25(10)=265$ °C ลิมิตขวา (จากหินฐานด้านล่าง): $\lim_{z\to10^+}T=c+15(10)=c+150$
3. ความต่อเนื่องต้องการให้ลิมิตทั้งสองเท่ากัน: $c+150=265$ จึงได้ $c=115$ °C ในกรณีทั่วไป $c=15+(g_1-g_2)z_c$
4. ต่อไปดูอนุพันธ์: $T'(z)=25$ เมื่อ $z<10$ และ $T'(z)=15$ เมื่อ $z>10$ ลิมิตข้างเดียวของ $T'$ ไม่เท่ากัน ดังนั้น $T$ ต่อเนื่องแต่ **หาอนุพันธ์ไม่ได้** ที่ $z=10$ กม. (กราฟมีมุม)
5. ฟิสิกส์บอกว่าปริมาณที่ต้องเท่ากันคือ การไหลของความร้อน $q=k\,T'(z)$ ดังนั้น $k_1g_1=k_2g_2$ และ $k_2/k_1=g_1/g_2=25/15\approx1.67$ สุดท้าย $T(40)=115+15(40)=715$ °C

```python
z, c = sp.symbols('z c', real=True)
T_top, T_bot = 15 + 25*z, c + 15*z

L_minus = sp.limit(T_top, z, 10, dir='-')
L_plus  = sp.limit(T_bot, z, 10, dir='+')
c_val = sp.solve(sp.Eq(L_minus, L_plus), c)[0]
show(r"\lim_{z\to10^-}T", L_minus); show(r"\lim_{z\to10^+}T", L_plus); show("c", c_val)

T = sp.Piecewise((T_top, z <= 10), (T_bot.subs(c, c_val), True))
dT = sp.diff(T, z)
show(r"\lim_{z\to10^-}T'(z)", sp.limit(dT, z, 10, dir='-'))
show(r"\lim_{z\to10^+}T'(z)", sp.limit(dT, z, 10, dir='+'))
```

```python
zz = np.linspace(0, 30, 301)
TT = sp.lambdify(z, T, "numpy")(zz)
fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(TT, zz, lw=2); ax.axhline(10, c="gray", ls="--"); ax.text(30, 9.3, "rock contact, z = 10 km")
ax.invert_yaxis(); ax.set(xlabel="temperature (°C)", ylabel="depth z (km)", title="Continuous T, kinked gradient")
plt.show()
```

**การตีความ** $c = 115$ ทำให้อุณหภูมิต่อเนื่อง (265 °C ที่รอยสัมผัส) แต่เกรเดียนต์ **ไม่** ต่อเนื่อง:
25 °C/km ในชั้นบน และ 15 °C/km ในชั้นล่าง ซึ่งเป็นไปตามที่คาด กฎของฟูรีเย (Fourier's law) บอกว่าฟลักซ์ความร้อน $q = k\,T'(z)$ คือปริมาณที่ต้องต่อเนื่อง
ดังนั้นหินฐานต้องนำความร้อนได้ดีกว่า: $k_2/k_1 = 25/15 \approx 1.67$ และหินแกรนิตก็นำความร้อนได้ดีกว่าหินดินดานจริง

**ความหมายของคำตอบ** อุณหภูมิที่รอยต่อคือ **265 °C** และหินฐานนำความร้อนได้ดีกว่าชั้นตะกอนราว **1.67 เท่า** ซึ่งสมเหตุสมผลสำหรับหินแกรนิตที่อยู่ใต้หินดินดาน จุดหักมุมในโปรไฟล์อุณหภูมิที่วัดจากหลุมเจาะจึงเป็นเบาะแสของการเปลี่ยนชนิดหิน ไม่ใช่ความผิดพลาดของข้อมูล

**ข้อผิดพลาดที่พบบ่อย**

- ทำให้ความชันเท่ากันแทนที่จะทำให้อุณหภูมิเท่ากัน ความต่อเนื่องพูดถึง $T$ ไม่ใช่ $T'$
- คิดว่าต่อเนื่องแปลว่าหาอนุพันธ์ได้ ฟังก์ชันต่อเนื่องอาจมีมุมได้ เช่น $|x|$ ที่ 0
- ใช้หน่วยปนกัน ความชันหน่วย °C/กม. ต้องคูณกับความลึกหน่วยกิโลเมตร ไม่ใช่เมตร

**ลองทำดู** ถ้าเกรเดียนต์ของหินฐานเป็น 20 °C/km ค่า $c$ เท่าใดจึงทำให้ $T$ ต่อเนื่อง

## ตัวอย่างที่ 4 ⭐: เนื้อโลกเริ่มหลอมละลายที่ไหน (ความต่อเนื่องและทฤษฎีบทค่าระหว่างกลาง (Intermediate Value Theorem))

**บริบท** หินหลอมละลายเมื่ออุณหภูมิ $T(z)$ สูงกว่า**โซลิดัส (solidus)** (อุณหภูมิเริ่มหลอมเหลว) ซึ่งเพิ่มขึ้นตามความดัน (ความลึก):

$$T(z) = 15 + 1600\left(1 - e^{-z/60}\right),\qquad T_s(z) = 1100 + 2z\qquad (^\circ\mathrm{C},\ z\ \text{in km}).$$

**แคลคูลัส** $f(z) = T(z) - T_s(z)$ เป็นฟังก์ชันต่อเนื่อง ถ้า $f(50) < 0$ และ $f(100) > 0$ แล้ว**ทฤษฎีบทค่าระหว่างกลาง**รับประกันว่ามีความลึก $c\in(50, 100)$ ที่ $f(c) = 0$
**วิธีแบ่งครึ่งช่วง (bisection method)** หาค่านี้ได้โดยแบ่งช่วงครึ่งไปเรื่อย ๆ

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** เราต้องการความลึกที่อุณหภูมิถึงจุดหลอมเหลวเป็นครั้งแรก แต่สมการ $T(z)=T_s(z)$ แก้ด้วยพีชคณิตไม่ได้ **ทฤษฎีบทค่าระหว่างกลาง (Intermediate Value Theorem, IVT)** พิสูจน์ว่ามีคำตอบจากการตรวจเครื่องหมายเพียงสองครั้ง แล้ว **วิธีแบ่งครึ่งช่วง (bisection method)** บีบคำตอบให้อยู่ในช่วงที่แคบลงเรื่อย ๆ

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **สร้างฟังก์ชันเดียว** นิยาม $f(z)=T(z)-T_s(z)=15+1600\left(1-e^{-z/60}\right)-(1100+2z)$ หินเป็นของแข็งเมื่อ $f<0$ และเริ่มหลอมเมื่อ $f=0$
2. **ตรวจความต่อเนื่อง** $f$ ประกอบด้วยฟังก์ชันเลขชี้กำลังและพหุนาม ซึ่งต่อเนื่องทุกค่า $z$ ดังนั้น $f$ ต่อเนื่องบน $[50,100]$ ซึ่งเป็นเงื่อนไขที่ IVT ต้องการ
3. **ตรวจเครื่องหมายที่ปลายช่วง** $f(50)=15+1600(1-e^{-0.833})-1200=15+904.6-1200\approx-280.4$ และ $f(100)=15+1600(1-e^{-1.667})-1300=15+1297.8-1300\approx+12.8$ เครื่องหมายต่างกัน ตาม IVT จึงมี $c\in(50,100)$ ที่ $f(c)=0$
4. **แบ่งครึ่งช่วง** หาค่าที่จุดกึ่งกลางแล้วเก็บครึ่งที่เครื่องหมายเปลี่ยน: $f(75)\approx-93.4$ → รากอยู่ใน $[75,100]$; $f(87.5)\approx-32.2$ → $[87.5,100]$; $f(93.75)\approx-7.9$ → $[93.75,100]$; $f(96.875)\approx+2.9$ → $[93.75,96.875]$ ทำต่อไปเรื่อย ๆ
5. **นับจำนวนขั้น** หลังแบ่งครึ่ง $n$ ครั้ง ความกว้างช่วงคือ $50/2^n$ ถ้าต้องการความกว้างน้อยกว่า $0.01$ km ต้องให้ $2^n>5000$ นั่นคือ $n\ge13$ ($50/2^{13}\approx0.006$ km)
6. **คำตอบ** รากคือ $z\approx96.01$ km ซึ่งมีอุณหภูมิ $T_s\approx1100+2(96.01)\approx1292$ °C

```python
f = lambda z: 15 + 1600*(1 - np.exp(-z/60)) - (1100 + 2*z)
a, b = 50.0, 100.0
print(f"f(50) = {f(a):.1f} °C  (< 0, solid)   f(100) = {f(b):.1f} °C  (> 0, partly molten)")
print(f"{'step':>4} {'a':>9} {'b':>9} {'mid':>9} {'f(mid)':>9}")
step = 0
while b - a > 0.01:
    m = (a + b)/2; step += 1
    print(f"{step:>4} {a:9.4f} {b:9.4f} {m:9.4f} {f(m):9.3f}")
    if f(a)*f(m) <= 0: b = m
    else: a = m
root = optimize.brentq(f, 50, 100)
print(f"melting starts at z ≈ {root:.2f} km (brentq check), T ≈ {1100 + 2*root:.0f} °C")
```

```python
zz = np.linspace(0, 200, 400)
fig, ax = plt.subplots(figsize=(5.5, 5))
ax.plot(15 + 1600*(1 - np.exp(-zz/60)), zz, lw=2, label="geotherm T(z)")
ax.plot(1100 + 2*zz, zz, "C3", lw=2, label="solidus T_s(z)")
ax.axhline(root, c="gray", ls="--"); ax.text(50, root - 4, f"melting begins ≈ {root:.0f} km")
ax.invert_yaxis(); ax.set(xlabel="temperature (°C)", ylabel="depth (km)", title="Where the curves cross, rock starts to melt"); ax.legend(); plt.show()
```

**การตีความ** $f$ เปลี่ยนเครื่องหมายระหว่าง 50 และ 100 km ดังนั้นตาม IVT เส้นโค้งทั้งสองต้องตัดกัน วิธีแบ่งครึ่งช่วงระบุจุดตัดได้ที่ประมาณ **96 km** ซึ่งเป็นด้านบนของเขตที่หลอมละลายบางส่วน เช่น เขตความเร็วต่ำ (low-velocity zone) ในฐานธรณีภาค (asthenosphere)
การแบ่งครึ่งแต่ละครั้งลดความไม่แน่นอนลงครึ่งหนึ่ง ดังนั้น 13 ขั้นตอนลดช่วง 50 km ให้เหลือน้อยกว่า 0.01 km

**ความหมายของคำตอบ** ใต้ความลึกประมาณ 96 km แมนเทิลร้อนพอที่จะเริ่มหลอมบางส่วน ซึ่งตรงกับขอบบนของเขตความเร็วต่ำในแอสทีโนสเฟียร์ ที่คลื่นไหวสะเทือนช้าลงเพราะมีหินหลอมเล็กน้อยแทรกระหว่างเม็ดแร่ IVT บอกเราว่าจุดตัด *มีอยู่จริง* ก่อนที่เราจะลงแรงหามัน

**ข้อผิดพลาดที่พบบ่อย**

- ใช้ IVT โดยไม่ตรวจความต่อเนื่อง หรือสรุปว่ามีรากเพียงหนึ่งราก ที่จริง IVT รับประกันเพียงว่ามี *อย่างน้อย* หนึ่งราก
- เก็บครึ่งช่วงผิดด้านในวิธีแบ่งครึ่ง ต้องเก็บครึ่งที่ค่าปลายทั้งสองยังมีเครื่องหมายตรงข้าม คือเมื่อ $f(a)f(m)\le0$
- คิดว่าเครื่องหมายเหมือนกันที่ปลายทั้งสองแปลว่าไม่มีราก ที่จริงอาจมีจุดตัดสองจุด IVT เพียงแต่ไม่ให้ข้อมูลในกรณีนี้

**ลองทำดู** เพิ่มความชันของโซลิดัสจาก 2 เป็น 3 °C/km (เนื้อโลกที่แห้งกว่า) ยังมีรากระหว่าง 50 และ 100 km หรือไม่ หมายความว่าอย่างไร

# ส่วน B: อนุพันธ์ของฟังก์ชันพื้นฐาน (Derivatives of Elementary Functions)

⭐ ง่าย → ⭐⭐ ปานกลาง · ตัวอย่างที่ 5–15 · สัปดาห์ที่ 3–5 ในประมวลรายวิชา

## ตัวอย่างที่ 5 ⭐: ระดับน้ำทะเลสูงขึ้นเร็วขึ้นหรือไม่ (อนุพันธ์ของพหุนาม)

**บริบท** เครื่องวัดระดับน้ำขึ้นน้ำลงบันทึกระดับน้ำทะเลเฉลี่ยรายปี (mm) นักวิทยาศาสตร์ภูมิอากาศปรับเส้นโค้งกำลังสอง
$h(t) = a + b\,s + c\,s^2$ โดย $s = t - 1900$ แล้วถามว่า **อัตรา** $h'(t)$ เพิ่มขึ้นหรือไม่

**แคลคูลัส** กฎยกกำลัง (power rule): $h'(t) = b + 2c\,s$ (mm/yr) และ $h''(t) = 2c$ (mm/yr² คือ *ความเร่ง*)

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** แนวโน้มกำลังสอง $h(s)=bs+cs^2$ เป็นพหุนาม จึงหาอนุพันธ์ได้ทันทีด้วย **กฎยกกำลัง (power rule)** อนุพันธ์อันดับหนึ่งคืออัตราการสูงขึ้นของระดับน้ำทะเล และอนุพันธ์อันดับสองคือความเร่ง อนุพันธ์อันดับสองที่เป็นบวกคือความหมายทางคณิตศาสตร์ของคำว่า \"ระดับน้ำทะเลสูงขึ้นเร็วขึ้นเรื่อย ๆ\"

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. ใช้แบบจำลองค่าเริ่มต้น $h(s)=1.2s+0.0085s^2$ มม. โดย $s=t-1900$ (จำนวนปีนับจาก ค.ศ. 1900)
2. ใช้กฎยกกำลัง $\frac{d}{ds}s^n=ns^{n-1}$ ทีละพจน์: $$h'(s)=1.2+2(0.0085)s=1.2+0.017s\ \text{มม./ปี}$$
3. หาอนุพันธ์อีกครั้ง: $h''(s)=0.017$ มม./ปี² มีค่าบวกและคงที่ อัตราจึงเพิ่มขึ้น 0.017 มม./ปี ทุกปี
4. สำหรับ ค.ศ. 2024 $s=124$: $h'(124)=1.2+0.017(124)=3.31$ มม./ปี เทียบกับ $h'(0)=1.2$ มม./ปี ใน ค.ศ. 1900 (ในสมุดบันทึกเราปรับเส้นโค้งกำลังสองเข้ากับข้อมูลมาตรวัดน้ำขึ้นน้ำลงจำลองที่มีสัญญาณรบกวน อัตราที่ได้จึงต่างเล็กน้อย คือ 1.26 มม./ปี ใน ค.ศ. 1900 และ 3.34 มม./ปี ใน ค.ศ. 2024)
5. การสูงขึ้นรวม: $h(124)=1.2(124)+0.0085(124)^2=148.8+130.7\approx279$ มม. นับจาก ค.ศ. 1900 ถ้าแนวโน้มนี้ดำเนินต่อไป จะสูงขึ้นอีก $h(200)-h(124)\approx301$ มม. ระหว่าง ค.ศ. 2024 ถึง 2100

```python
rng = np.random.default_rng(111)
years = np.arange(1900, 2025)
s = years - 1900
true = 1.2*s + 0.0085*s**2                         # synthetic "truth"
gauge = true + rng.normal(0, 15, years.size)        # + interannual noise

c2, b1, a0 = np.polyfit(s, gauge, 2)                # least-squares quadratic
S = sp.symbols('s')
h = a0 + b1*S + c2*S**2
dh, d2h = sp.diff(h, S), sp.diff(h, S, 2)
show("h(s)", sp.N(h, 4)); show("h'(s)", sp.N(dh, 4)); show("h''(s)", sp.N(d2h, 4))
for yr in (1900, 1950, 2000, 2024):
    print(f"rate in {yr}: {float(dh.subs(S, yr-1900)):.2f} mm/yr")
```

```python
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4))
a1.plot(years, gauge, ".", ms=4, alpha=.6, label="tide gauge (synthetic)")
a1.plot(years, sp.lambdify(S, h)(s), "C3", lw=2, label="quadratic fit h(t)")
a1.set(xlabel="year", ylabel="sea level (mm)"); a1.legend()
a2.plot(years, sp.lambdify(S, dh)(s), "C3", lw=2)
a2.set(xlabel="year", ylabel="h'(t)  (mm/yr)", title="rate of sea-level rise")
plt.tight_layout(); plt.show()
```

**การตีความ** $h''>0$ แสดงว่าอัตราเองกำลังเพิ่มขึ้น นั่นคือระดับน้ำทะเลสูงขึ้นแบบเร่ง อัตราเพิ่มเกือบสามเท่าระหว่างปี 1900 ถึง 2024 (≈1.3 → 3.3 mm/yr)

**ความหมายของคำตอบ** อัตราเพิ่มขึ้นเกือบ **สามเท่า** (1.2 → 3.3 มม./ปี) ใกล้กับค่าที่วัดด้วยดาวเทียมวัดความสูง (satellite altimetry) ราว 3–4 มม./ปี การคาดการณ์ถึง ค.ศ. 2100 เป็นเพียงฉากทัศน์ เพราะพหุนามที่ปรับเข้ากับข้อมูลในอดีตไม่มีฟิสิกส์อยู่ในนั้น การพยากรณ์นอกช่วงข้อมูลจึงต้องระวัง

**ข้อผิดพลาดที่พบบ่อย**

- แทนปี ค.ศ. $t=2024$ แทนที่จะแทน $s=124$ ลงใน $h'(s)$
- สับสนระหว่างอัตรา $h'$ (มม./ปี) กับความเร่ง $h''$ (มม./ปี²) หรือลืมตัวคูณ 2 จากกฎยกกำลัง
- ถือว่าพหุนามที่ปรับเข้ากับข้อมูลเป็นกฎธรรมชาติ แล้วใช้คาดการณ์ไกลเกินช่วงข้อมูล

**ลองทำดู** ใช้เส้นโค้งที่ปรับได้ประมาณอัตราในปี 2050 การประมาณค่านอกช่วง (extrapolate) ด้วยพหุนามไกลขนาดนั้นปลอดภัยหรือไม่ เพราะเหตุใด

## ตัวอย่างที่ 6 ⭐: ลาดดินพังเมื่อใด (กฎผลหาร (quotient rule))

**บริบท** สำหรับลาดเขายาว ("อนันต์") ที่มีชั้นดินหนา $z$ **อัตราส่วนความปลอดภัย (factor of safety)** ต่อการเลื่อนไถลคือ แรงต้าน / แรงขับ:

$$FS(\beta) = \frac{c}{\gamma z\sin\beta\cos\beta} + \frac{\tan\varphi}{\tan\beta},$$

โดยแรงยึดเหนี่ยว (cohesion) $c = 5$ kPa หน่วยน้ำหนัก $\gamma = 19$ kN/m³ $z = 2$ m และมุมเสียดทาน $\varphi = 30^\circ$ ลาดจะพังเมื่อ $FS < 1$

**แคลคูลัส** กฎผลหาร: $\left(\dfrac{u}{v}\right)' = \dfrac{u'v - uv'}{v^2}$ เครื่องหมายของ $FS'(\beta)$ บอกว่าการเพิ่มความชันทำให้ลาดปลอดภัยขึ้นหรือลดลง

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** อัตราส่วนความปลอดภัยเป็นผลบวกของ **ผลหาร (quotient)** สองพจน์ของฟังก์ชันตรีโกณมิติของมุมลาด $\beta$ อนุพันธ์ $FS'(\beta)$ ที่หาด้วย **กฎผลหาร (quotient rule)** บอกว่าความมั่นคงลดลงเร็วแค่ไหนเมื่อลาดชันขึ้น

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ทำค่าคงตัวให้ง่าย** เมื่อ $c=5$ kPa, $\gamma=19$ kN/m³, $z=2$ m จะได้ $c/(\gamma z)=5/38\approx0.1316$ ดังนั้น $$FS(\beta)=\frac{0.1316}{\sin\beta\cos\beta}+\frac{\tan30^\circ}{\tan\beta}$$
2. **หาอนุพันธ์ของพจน์แรก** ให้ $u=c$ (ค่าคงตัว $u'=0$) และ $v=\gamma z\sin\beta\cos\beta$ จากกฎผลคูณ $v'=\gamma z(\cos^2\beta-\sin^2\beta)=\gamma z\cos2\beta$ กฎผลหารให้ $$\left(\frac{u}{v}\right)'=\frac{0\cdot v-c\,v'}{v^2}=-\frac{c\cos2\beta}{\gamma z\sin^2\beta\cos^2\beta}$$
3. **หาอนุพันธ์ของพจน์ที่สอง** ให้ $u=\tan\varphi$ (ค่าคงตัว) และ $v=\tan\beta$ ซึ่ง $v'=\sec^2\beta$ $$\left(\frac{\tan\varphi}{\tan\beta}\right)'=-\frac{\tan\varphi\sec^2\beta}{\tan^2\beta}=-\frac{\tan\varphi}{\sin^2\beta}$$
4. **แทนค่าที่ $\beta=30^\circ$** $\sin30^\circ\cos30^\circ=0.4330$ ดังนั้น $FS=0.1316/0.4330+1=1.30$ และเนื่องจาก $\cos60^\circ=0.5$ และ $\sin^2\beta\cos^2\beta=0.1875$ จะได้ $FS'=-0.1316(0.5)/0.1875-0.5774/0.25=-0.351-2.309\approx-2.66$ ต่อเรเดียน
5. **อ่านเครื่องหมาย** เมื่อ $\beta<45^\circ$ ทั้งสองพจน์ของ $FS'$ เป็นลบ (เพราะ $\cos2\beta>0$) $FS$ จึงลดลง ความชันที่เพิ่มขึ้นทุกองศาทำให้ลาดพังง่ายขึ้น
6. **หามุมวิกฤต** แก้สมการ $FS(\beta)=1$ เชิงตัวเลข: $FS(35^\circ)=1.10$, $FS(40^\circ)=0.96$ และรากคือ $\beta\approx38.4^\circ$

```python
beta = sp.symbols('beta', positive=True)
c_, gam, zz_, phi = 5, 19, 2, sp.rad(30)
FS = c_/(gam*zz_*sp.sin(beta)*sp.cos(beta)) + sp.tan(phi)/sp.tan(beta)
dFS = sp.simplify(sp.diff(FS, beta))
show(r"FS'(\beta)", dFS)

FSf, dFSf = sp.lambdify(beta, FS), sp.lambdify(beta, dFS)
b_crit = optimize.brentq(lambda b: FSf(b) - 1, np.radians(10), np.radians(60))
b_min = optimize.brentq(dFSf, np.radians(45), np.radians(89))
print(f"FS'(β) < 0 for β < {np.degrees(b_min):.1f}°  (steeper = less safe); FS has its minimum {FSf(b_min):.2f} there")
print(f"critical slope (FS = 1): β = {np.degrees(b_crit):.1f}°")
for bd in (20, 30, 35, 40):
    print(f"  β = {bd}°: FS = {FSf(np.radians(bd)):.2f}, FS' = {dFSf(np.radians(bd)):.2f} per radian")
```

```python
bb = np.radians(np.linspace(10, 85, 300))
fig, ax = plt.subplots()
ax.plot(np.degrees(bb), FSf(bb), lw=2, label="FS(β)")
ax.axhline(1, c="C3", ls="--", label="FS = 1 (failure)"); ax.axvline(np.degrees(b_crit), c="gray", ls=":")
ax.fill_between(np.degrees(bb), 0, FSf(bb), where=FSf(bb) < 1, color="C3", alpha=.15, label="unstable")
ax.set(xlabel="slope angle β (°)", ylabel="factor of safety", ylim=(0, 4), title="Infinite-slope stability"); ax.legend(); plt.show()
```

**การตีความ** $FS'(\beta) < 0$ สำหรับทุกความชันที่ต่ำกว่าประมาณ 67° ซึ่งครอบคลุมลาดเขาที่มีดินปกคลุมจริง ดังนั้นความชันที่เพิ่มขึ้นทุกองศาจะลดเสถียรภาพ ลาดจะไม่เสถียรเมื่อชันเกินประมาณ **38°**
(เมื่อเกินประมาณ 67° พจน์แรงยึดเหนี่ยวทำให้ $FS$ เพิ่มขึ้นอีก แต่ชั้นดินหนา 2 m ไม่สามารถคงอยู่บนลาดชันขนาดนั้นได้อยู่แล้ว) แผนที่พื้นที่เสี่ยงดินถล่มในภาคเหนือของไทยระบุลาดที่ชันกว่าประมาณ 30–35° ด้วยเหตุผลนี้

**ความหมายของคำตอบ** ด้วยสมบัติดินชุดนี้ ลาดจะมั่นคงจนถึงประมาณ 38° และพังเมื่อชันกว่านั้น ที่ 30° การเพิ่มความชัน 1° (0.0175 rad) ทำให้ $FS$ ลดลงราว $2.66\times0.0175\approx0.047$ ฝนที่ลดแรงยึดเหนี่ยว $c$ ทำให้มุมวิกฤตต่ำลง ดินถล่มในฤดูมรสุมทางภาคเหนือจึงเกิดบนลาดที่มั่นคงในฤดูแล้ง

**ข้อผิดพลาดที่พบบ่อย**

- เขียนกฎผลหารเป็น $(u'v+uv')/v^2$ เครื่องหมายในตัวเศษต้องเป็น **ลบ** คือ $u'v-uv'$
- แทนค่ามุมเป็นองศาในสูตรอนุพันธ์ อนุพันธ์ของฟังก์ชันตรีโกณมิติใช้หน่วยเรเดียน $FS'$ จึงมีหน่วยต่อเรเดียน ถ้าต้องการต่อองศาให้หารด้วย 57.3
- ลืมใช้กฎผลคูณกับ $v=\sin\beta\cos\beta$ (อนุพันธ์คือ $\cos2\beta$ ไม่ใช่ $\cos\beta\cdot(-\sin\beta)$)

**ลองทำดู** ฝนมรสุมหนักลดแรงยึดเหนี่ยวเหลือ $c = 1$ kPa มุมวิกฤตใหม่เป็นเท่าใด

## ตัวอย่างที่ 7 ⭐: การเปลี่ยนแปลงความดันบนบอลลูนตรวจอากาศ (กฎลูกโซ่ (chain rule))

**บริบท** ความดันอากาศลดลงตามความสูง: $P(z) = 1013.25\,e^{-z/8.4}$ hPa โดย $z$ หน่วย km (8.4 km คือ*ความสูงมาตรา (scale height)*)
เรดิโอซอนด์ (radiosonde หรือบอลลูนตรวจอากาศ) ลอยขึ้นด้วยอัตรา $dz/dt = 5$ m/s เซนเซอร์ความดันจะเห็นความดันลดลงเร็วเท่าใดที่ $z = 3$ km

**แคลคูลัส** กฎลูกโซ่: $\dfrac{dP}{dt} = \dfrac{dP}{dz}\cdot\dfrac{dz}{dt}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ความดันขึ้นกับความสูง และความสูงขึ้นกับเวลา **หลักเกณฑ์ลูกโซ่ (chain rule)** เชื่อมอัตราทั้งสองเข้าด้วยกัน: อัตราที่เซนเซอร์วัดได้ตามเวลาเท่ากับอัตราการเปลี่ยนตามความสูงคูณด้วยอัตราการลอยขึ้น

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **หาอนุพันธ์เทียบกับความสูง** สำหรับ $P(z)=1013.25\,e^{-z/8.4}$ ใช้หลักเกณฑ์ลูกโซ่กับเลขชี้กำลังได้ $$\frac{dP}{dz}=1013.25\,e^{-z/8.4}\cdot\left(-\frac{1}{8.4}\right)=-\frac{P(z)}{8.4}$$
2. **แทนค่าที่ 3 km** $P(3)=1013.25\,e^{-0.3571}=1013.25(0.6997)\approx708.9$ hPa ดังนั้น $dP/dz=-708.9/8.4\approx-84.40$ hPa/km
3. **แปลงอัตราการลอยขึ้นให้หน่วยเข้ากัน** $dz/dt=5$ m/s $=300$ m/min $=0.3$ km/min
4. **ใช้หลักเกณฑ์ลูกโซ่** $$\frac{dP}{dt}=\frac{dP}{dz}\cdot\frac{dz}{dt}=(-84.40\ \text{hPa/km})(0.3\ \text{km/min})\approx-25.32\ \text{hPa/min}$$ หน่วย km ตัดกันเหลือ hPa/min
5. **เปรียบเทียบที่ 10 km** $P(10)\approx308.2$ hPa ดังนั้น $dP/dt=-(308.2/8.4)(0.3)\approx-11.00$ hPa/min การลอยขึ้นเท่าเดิมทำให้ความดันลดช้าลงเมื่ออยู่สูงขึ้น

```python
z, tt_ = sp.symbols('z t', positive=True)
P = 1013.25*sp.exp(-z/sp.Rational(84, 10))
dPdz = sp.diff(P, z)
show(r"\frac{dP}{dz}", dPdz)
dzdt = 5*60/1000                          # 5 m/s = 0.3 km/min
z0 = 3
print(f"P(3 km) = {float(P.subs(z, z0)):.1f} hPa, dP/dz = {float(dPdz.subs(z, z0)):.2f} hPa/km")
print(f"dP/dt = dP/dz · dz/dt = {float(dPdz.subs(z, z0)):.2f} × {dzdt} = {float(dPdz.subs(z, z0))*dzdt:.2f} hPa/min")
print(f"at 10 km the rate is only {float(dPdz.subs(z, 10))*dzdt:.2f} hPa/min")
```

```python
Z = np.linspace(0, 30, 300)
Pf, dPf = sp.lambdify(z, P), sp.lambdify(z, dPdz)
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4))
a1.plot(Pf(Z), Z, lw=2); a1.set(xlabel="pressure (hPa)", ylabel="height (km)", title="P(z)")
a2.plot(dPf(Z)*dzdt, Z, "C3", lw=2); a2.set(xlabel="dP/dt (hPa/min) for a 5 m/s balloon", ylabel="height (km)", title="rate seen by the balloon")
plt.tight_layout(); plt.show()
```

**การตีความ** ที่ 3 km บอลลูนเห็นความดันลดลงประมาณ **25 hPa ทุกนาที** ที่สูงขึ้นไปการลดลงจะช้าลงเพราะ $dP/dz$ แปรผันตาม $P$ เอง
นักอุตุนิยมวิทยาใช้เหตุผลนี้ในทางกลับกัน: จาก $dP/dt$ ที่วัดได้และ $dP/dz$ ที่ทราบ จะได้ความสูงและอัตราการลอยขึ้นของบอลลูน

**ความหมายของคำตอบ** ที่ความสูง 3 km เซนเซอร์บนบอลลูนวัดความดันที่ลดลงราว 25 hPa ต่อนาที เพราะ $dP/dz$ แปรผันตาม $P$ การลดลงจึงช้าลงเมื่อสูงขึ้น นักอุตุนิยมวิทยาใช้ความสัมพันธ์เดียวกันย้อนกลับ คือคำนวณความสูงและอัตราการลอยของบอลลูนจาก $dP/dt$ ที่วัดได้

**ข้อผิดพลาดที่พบบ่อย**

- ใช้หน่วยปนกัน การคูณ hPa/km กับ 5 m/s ให้ตัวเลขที่ไม่มีความหมาย ต้องแปลงหน่วยให้หน่วยความยาวตัดกัน
- ลืมอนุพันธ์ภายใน $-1/8.4$ ของเลขชี้กำลัง ทำให้ได้ขนาดและเครื่องหมายผิด
- ทิ้งเครื่องหมายลบ ความดัน *ลดลง* เมื่อบอลลูนลอยขึ้น ดังนั้น $dP/dt<0$

**ลองทำดู** กระแสอากาศขึ้นในพายุยกบอลลูนด้วยความเร็ว 15 m/s แทน $dP/dt$ ที่ 3 km เป็นเท่าใด

## ตัวอย่างที่ 8 ⭐: การหาอายุด้วยคาร์บอนกัมมันตรังสี (อนุพันธ์ของฟังก์ชันเอกซ์โพเนนเชียลและลอการิทึม)

**บริบท** $^{14}$C สลายตัวด้วยครึ่งชีวิต $t_{1/2} = 5730$ yr สัดส่วนที่เหลืออยู่หลังผ่านไป $t$ ปีคือ
$f(t) = e^{-\lambda t}$ โดย $\lambda = \ln 2 / t_{1/2}$ เมื่อแก้สมการหาอายุจะได้ $t(f) = -\dfrac{\ln f}{\lambda}$

**คำถาม**
1. **อัตราการสลายตัว** $f'(t)$: เหตุใดจึงแปรผันตรงกับ $f$ เอง
2. **ความไวของอายุ** $\dfrac{dt}{df}$: ห้องปฏิบัติการวัดได้ $f = 0.25$ อายุจะเปลี่ยนไปเท่าใดต่อความคลาดเคลื่อน 0.01 ใน $f$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** การสลายตัวของกัมมันตรังสีอธิบายด้วยฟังก์ชันเลขชี้กำลัง $f=e^{-\lambda t}$ และหาอายุได้โดยใช้ลอการิทึมธรรมชาติกลับฟังก์ชัน อนุพันธ์ของอายุเทียบกับสัดส่วนที่วัดได้ $dt/df$ เป็น **ความไว (sensitivity)** เมื่อคูณด้วยความคลาดเคลื่อนในการวัด $\Delta f$ จะได้ความคลาดเคลื่อนของอายุ นี่คือการประมาณเชิงเส้น (เส้นสัมผัส) ที่ใช้ทั่วไปในการวิเคราะห์ความคลาดเคลื่อน

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. ค่าคงตัวการสลายตัวจากครึ่งชีวิต: $\lambda=\dfrac{\ln2}{t_{1/2}}=\dfrac{0.6931}{5730}=1.2097\times10^{-4}$ ต่อปี
2. อนุพันธ์ของกฎการสลายตัว: $f'(t)=-\lambda e^{-\lambda t}=-\lambda f$ อัตราการลดลงแปรผันตามปริมาณที่เหลือ ดังนั้น $f'/f=-\lambda$ เป็นค่าคงตัว
3. หาฟังก์ชันผกผัน: ใส่ $\ln$ ทั้งสองข้าง $\ln f=-\lambda t$ จึงได้ $t(f)=-\dfrac{\ln f}{\lambda}$ เมื่อ $f=0.25$: $t=-\ln(0.25)/\lambda=1.3863/1.2097\times10^{-4}=11{,}460$ ปี ซึ่งเท่ากับสองครึ่งชีวิตพอดี
4. หาอนุพันธ์ของอายุโดยใช้ $\frac{d}{df}\ln f=\frac1f$: $$\frac{dt}{df}=-\frac{1}{\lambda f}=-\frac{1}{(1.2097\times10^{-4})(0.25)}=-33{,}067\ \text{ปี}$$
5. ประมาณความคลาดเคลื่อน: $\Delta t\approx\left|\dfrac{dt}{df}\right|\Delta f=33{,}067\times0.01\approx331$ ปี ทำซ้ำกับตัวอย่างอายุมากที่ $f=0.02$: $|dt/df|=413{,}332$ ปี ความคลาดเคลื่อนเท่าเดิมจึงให้ $\pm4{,}133$ ปี

```python
t, f = sp.symbols('t f', positive=True)
lam = sp.log(2)/5730
F = sp.exp(-lam*t)
age = -sp.log(f)/lam

show("f'(t)", sp.diff(F, t))
print("f'(t) / f(t) =", sp.simplify(sp.diff(F, t)/F), "  -> constant: decay rate is proportional to amount")
dage = sp.diff(age, f)
show(r"\frac{dt}{df}", sp.simplify(dage))
print(f"age at f = 0.25 : {float(age.subs(f, 0.25)):.0f} yr")
print(f"dt/df at f=0.25 : {float(dage.subs(f, 0.25)):.0f} yr per unit f  ->  {abs(float(dage.subs(f, 0.25)))*0.01:.0f} yr per 0.01 error")
print(f"dt/df at f=0.02 : {float(dage.subs(f, 0.02)):.0f} yr per unit f  ->  {abs(float(dage.subs(f, 0.02)))*0.01:.0f} yr per 0.01 error")
```

```python
ff = np.linspace(0.01, 1, 300)
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4))
tt = np.linspace(0, 40000, 300)
a1.plot(tt, np.exp(-float(lam)*tt), lw=2); a1.set(xlabel="time (yr)", ylabel="fraction ¹⁴C remaining f(t)")
a2.plot(ff, np.abs(sp.lambdify(f, dage)(ff)), "C3", lw=2); a2.set_yscale("log")
a2.set(xlabel="measured fraction f", ylabel="|dt/df|  (yr)", title="age uncertainty explodes for old samples")
plt.tight_layout(); plt.show()
```

**การตีความ** $f'(t) = -\lambda f$: ยิ่งมีอะตอมมาก ก็ยิ่งสลายตัวเร็ว สำหรับอายุ $\dfrac{dt}{df} = -\dfrac{1}{\lambda f}$
มีค่ามากเมื่อ $f$ มีค่าน้อย นี่คือเหตุผลที่การหาอายุด้วยคาร์บอนใช้ไม่ได้เกินประมาณ 50,000 yr เพราะความคลาดเคลื่อนในการวัดเพียงเล็กน้อยทำให้อายุคลาดเคลื่อนมหาศาล

**ความหมายของคำตอบ** ตัวอย่างมีอายุ **11,460 ± 331 ปี** เครื่องหมายลบใน $dt/df$ บอกเพียงว่าคาร์บอน-14 น้อยลงแปลว่าอายุมากขึ้น เนื่องจาก $|dt/df|\propto1/f$ ความไม่แน่นอนจึงพุ่งสูงมากสำหรับตัวอย่างเก่า นี่คือเหตุที่การหาอายุด้วยคาร์บอนกัมมันตรังสีใช้ไม่ได้เมื่อใกล้ 50,000 ปี และนักธรณีวิทยาเปลี่ยนไปใช้นาฬิกาที่ช้ากว่า เช่น K–Ar หรือ U–Pb

**ข้อผิดพลาดที่พบบ่อย**

- ใช้ $\log_{10}$ แทน $\ln$ กฎการสลายตัวใช้ฐาน $e$ ฟังก์ชันผกผันจึงต้องเป็นลอการิทึมธรรมชาติ
- ใช้ครึ่งชีวิตแทน $\lambda$ จำไว้ว่า $\lambda=\ln2/t_{1/2}$
- เก็บเครื่องหมายลบไว้ในช่วงความคลาดเคลื่อน ความคลาดเคลื่อนเป็นขนาด จึงต้องใช้ $|dt/df|$

**ลองทำดู** ทำซ้ำกับ $^{40}$K ($t_{1/2} = 1.25\times10^{9}$ yr) นาฬิกานี้เหมาะกับหินประเภทใด

## ตัวอย่างที่ 9 ⭐: แผ่นดินไหวขนาด 7 มีพลังงานมากกว่าขนาด 6 เท่าใด (อนุพันธ์ของ $a^x$ และ $\log_a x$)

**บริบท** พลังงานคลื่นไหวสะเทือนที่แผ่ออกมา (J) สัมพันธ์กับขนาดผ่านความสัมพันธ์พลังงานของกูเทนแบร์ก–ริกเตอร์ (Gutenberg–Richter)
$\log_{10}E = 1.5M + 4.8$ ดังนั้น $E(M) = 10^{1.5M+4.8}$
ขนาดโมเมนต์ (moment magnitude) นิยามจากโมเมนต์แผ่นดินไหว $M_0$ (N·m) โดย $M_w = \tfrac{2}{3}\left(\log_{10}M_0 - 9.1\right)$

**แคลคูลัส** $\dfrac{d}{dx}a^{u} = a^{u}\ln a\,\dfrac{du}{dx}$ และ $\dfrac{d}{dx}\log_a u = \dfrac{1}{u\ln a}\dfrac{du}{dx}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** มาตราขนาดแผ่นดินไหวเป็นแบบลอการิทึม พลังงานจึงเป็นฟังก์ชันเลขชี้กำลังฐาน 10 ของขนาด: $E=10^{1.5M+4.8}$ กฎ $\frac{d}{dx}a^{u}=a^{u}\ln a\cdot u'$ แสดงว่าอัตราการเติบโต **สัมพัทธ์** $(1/E)(dE/dM)$ เป็นค่าคงตัว ซึ่งเป็นสมบัติเฉพาะของการเติบโตแบบเลขชี้กำลัง และอธิบายว่าทำไมขนาดที่เพิ่มขึ้นหนึ่งหน่วยจึงคูณพลังงานด้วยตัวเลขเดิมเสมอ

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. พลังงานที่ขนาดค่าเริ่มต้น $M=7.7$: $E=10^{1.5(7.7)+4.8}=10^{16.35}=2.24\times10^{16}$ จูล หารด้วย $4.184\times10^9$ จูลต่อทีเอ็นทีหนึ่งตัน ได้ประมาณ 5.35 ล้านตัน
2. หาอนุพันธ์โดยใช้ $\frac{d}{dM}10^{u}=10^{u}\ln10\cdot\frac{du}{dM}$ และ $u=1.5M+4.8$: $$\frac{dE}{dM}=1.5\ln10\cdot E\approx3.454\,E$$
3. ที่ $M=7.7$: $dE/dM=3.454\times2.24\times10^{16}=7.73\times10^{16}$ จูลต่อขนาดหนึ่งหน่วย อัตราสัมพัทธ์ $(1/E)(dE/dM)=1.5\ln10\approx3.45$ ไม่ขึ้นกับ $M$
4. เนื่องจากอนุพันธ์เป็นอัตราเฉพาะจุด สำหรับการเพิ่มขึ้นจำกัด $\Delta M=1$ ให้ใช้อัตราส่วนแท้จริง: $\dfrac{E(M+1)}{E(M)}=10^{1.5}\approx31.6$ เพิ่มสองหน่วยได้ $10^{3}=1000$
5. เทียบกับแผ่นดินไหวขนาดปานกลาง: $E(7.7)/E(5)=10^{1.5(2.7)}=10^{4.05}\approx11{,}220$ แผ่นดินไหวขนาด 7.7 หนึ่งครั้งปล่อยพลังงานเท่ากับแผ่นดินไหวขนาด 5.0 ราว 11,200 ครั้ง

```python
M, M0 = sp.symbols('M M_0', positive=True)
E = 10**(sp.Rational(3, 2)*M + sp.Rational(48, 10))
dE = sp.diff(E, M)
show(r"\frac{dE}{dM}", dE)
print("relative rate (1/E)(dE/dM) =", sp.simplify(dE/E), "=", float(sp.simplify(dE/E)))
print(f"E(7)/E(6) = {float(E.subs(M, 7)/E.subs(M, 6)):.1f}   (one magnitude unit ≈ ×31.6 energy)")

Mw = sp.Rational(2, 3)*(sp.log(M0, 10) - sp.Rational(91, 10))
dMw = sp.diff(Mw, M0)
show(r"\frac{dM_w}{dM_0}", sp.simplify(dMw))
M0_tohoku = 4.0e22
print(f"Tohoku 2011: M0 = {M0_tohoku:.1e} N·m -> Mw = {float(Mw.subs(M0, M0_tohoku)):.2f}")
```

```python
MM = np.linspace(3, 9.5, 200)
Ef = sp.lambdify(M, E)
fig, ax = plt.subplots()
ax.semilogy(MM, Ef(MM), lw=2)
for m, name in [(5.0, "moderate"), (7.7, "Myanmar 2025"), (9.1, "Sumatra–Andaman 2004")]:
    ax.plot(m, Ef(m), "o"); ax.annotate(f"{name}\nM{m}: {Ef(m):.1e} J", (m, Ef(m)), xytext=(-110, 10), textcoords="offset points", fontsize=9)
ax.set(xlabel="magnitude M", ylabel="energy E (J), log scale", title="E = 10^(1.5M + 4.8)")
plt.show()
```

**การตีความ** $\dfrac{1}{E}\dfrac{dE}{dM} = 1.5\ln 10 \approx 3.45$ เป็นอัตราการเติบโต *สัมพัทธ์* ที่คงที่ ขนาดที่เพิ่มขึ้นทุก +1 ทำให้พลังงานคูณด้วย $10^{1.5}\approx 31.6$
เนื่องจาก $\dfrac{dM_w}{dM_0}=\dfrac{2}{3M_0\ln 10}$ ลดลงเมื่อ $M_0$ เพิ่มขึ้น การเปลี่ยนแปลงของโมเมนต์จำนวนมหาศาลจึงทำให้ขนาดเปลี่ยนไปเพียงเล็กน้อย

**ความหมายของคำตอบ** แผ่นดินไหวขนาด 7.7 (เช่น เมียนมา ค.ศ. 2025) ปล่อยพลังงานราว $2\times10^{16}$ จูล แผ่นดินไหวเล็ก ๆ จึง \"ระบายแรงดัน\" ของรอยเลื่อนใหญ่ไม่ได้ ต้องใช้แผ่นดินไหวขนาด 5 มากกว่าหนึ่งหมื่นครั้งจึงเท่ากับขนาด 7.7 หนึ่งครั้ง บนกราฟสเกลลอการิทึม เส้นโค้งพลังงานเป็นเส้นตรง ซึ่งเป็นวิธีสังเกตการเติบโตแบบเลขชี้กำลัง

**ข้อผิดพลาดที่พบบ่อย**

- เขียน $\frac{d}{dM}10^{u}=u\cdot10^{u-1}$ (ใช้กฎยกกำลังผิดที่) ตัวแปรอยู่ในเลขชี้กำลัง จึงต้องใช้ $10^{u}\ln10\cdot u'$
- คิดว่าขนาด 7 มากกว่าขนาด 6 แค่ \"หนึ่ง\" จึงแรงกว่าเพียงเล็กน้อย ที่จริงพลังงานมากกว่าราว 31.6 เท่า
- ใช้ $dE/dM\cdot\Delta M$ กับการเปลี่ยนแปลงขนาดใหญ่ เช่น $\Delta M=1$ สำหรับการเปลี่ยนแปลงมากต้องใช้อัตราส่วนแท้จริง $10^{1.5\Delta M}$

**ลองทำดู** แผ่นดินไหวขนาด M 5.0 กี่ครั้งจึงปลดปล่อยพลังงานเท่ากับแผ่นดินไหวขนาด M 7.7 หนึ่งครั้ง

## ตัวอย่างที่ 10 ⭐: ภาวะกรดในมหาสมุทร (อนุพันธ์ของลอการิทึม (derivative of a logarithm))

**บริบท** $\mathrm{pH} = -\log_{10}[\mathrm{H}^+]$ ตั้งแต่ยุคก่อนอุตสาหกรรม pH เฉลี่ยของผิวมหาสมุทรลดลงจากประมาณ **8.18 เป็น 8.07** เพราะมหาสมุทรดูดซับ CO₂
การเปลี่ยนแปลง "เพียง 0.11" ฟังดูน้อย แต่ในแง่ของ $[\mathrm{H}^+]$ มากเพียงใด

**แคลคูลัส** $\dfrac{d\,\mathrm{pH}}{d[\mathrm{H}^+]} = -\dfrac{1}{[\mathrm{H}^+]\ln10}$ ดังนั้นผลต่างเชิงอนุพันธ์ (differential) คือ $d\,\mathrm{pH} \approx -\dfrac{1}{\ln 10}\dfrac{d[\mathrm{H}^+]}{[\mathrm{H}^+]}$: pH ตอบสนองต่อการเปลี่ยนแปลง**สัมพัทธ์**ของ $[\mathrm{H}^+]$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** pH เป็นลอการิทึมฐาน 10 และอนุพันธ์ของลอการิทึมแปรผันตาม $1/x$ pH จึงตอบสนองต่อการเปลี่ยนแปลง **สัมพัทธ์** ของความเข้มข้นไฮโดรเจนไอออน การเปลี่ยน pH ที่ดูเล็กน้อยจึงซ่อนการเปลี่ยนความเป็นกรดที่มาก

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **เปลี่ยนฐาน** $\mathrm{pH}=-\log_{10}H=-\dfrac{\ln H}{\ln10}$ โดย $H=[\mathrm{H}^+]$
2. **หาอนุพันธ์** เนื่องจาก $\dfrac{d}{dH}\ln H=\dfrac1H$ จะได้ $$\frac{d\,\mathrm{pH}}{dH}=-\frac{1}{H\ln10}$$
3. **แปลง pH เป็นความเข้มข้น** $H_0=10^{-8.18}\approx6.607\times10^{-9}$ mol/L (ก่อนยุคอุตสาหกรรม) และ $H_1=10^{-8.07}\approx8.511\times10^{-9}$ mol/L (ปัจจุบัน) เพิ่มขึ้นสัมพัทธ์ $(H_1-H_0)/H_0\approx0.288=28.8\%$
4. **ประมาณด้วยผลต่างเชิงอนุพันธ์** $\Delta\mathrm{pH}\approx\dfrac{d\,\mathrm{pH}}{dH}\Delta H=-\dfrac{1}{\ln10}\dfrac{\Delta H}{H_0}=-0.4343\times0.288\approx-0.125$
5. **เทียบกับค่าจริง** การเปลี่ยนแปลงจริงคือ $8.07-8.18=-0.110$ ค่าประมาณเชิงเส้นเกินไปเล็กน้อยเพราะการเปลี่ยน 29 % ไม่ใช่ค่าเล็ก เส้นโค้งลอการิทึมจึงโค้งออกจากเส้นสัมผัส
6. **มองไปข้างหน้า** ถ้า pH ลดลงเหลือ 7.95 ในปี 2100 จะได้ $10^{-7.95}/H_0-1\approx0.70$ คือไฮโดรเจนไอออนเพิ่มขึ้นราว 70 % จากก่อนยุคอุตสาหกรรม

```python
H = sp.symbols('H', positive=True)
pH = -sp.log(H, 10)
show(r"\frac{d\,pH}{d[H^+]}", sp.simplify(sp.diff(pH, H)))
H0, H1 = 10**-8.18, 10**-8.07
rel = (H1 - H0)/H0
print(f"[H+] pre-industrial = {H0:.3e} mol/L, today = {H1:.3e} mol/L -> increase {rel*100:.1f}%")
approx = float(sp.diff(pH, H).subs(H, H0))*(H1 - H0)
print(f"differential estimate ΔpH ≈ {approx:.3f}   exact ΔpH = {8.07-8.18:.3f}")
print(f"if pH falls to 7.95 by 2100: [H+] up {(10**-7.95/H0 - 1)*100:.0f}% from pre-industrial")
```

```python
pHs = np.linspace(7.7, 8.3, 200)
fig, ax = plt.subplots()
ax.plot(pHs, 10**(-pHs)*1e9, lw=2)
for p_, lab in [(8.18, "pre-industrial"), (8.07, "today"), (7.95, "2100 (high emissions)")]:
    ax.plot(p_, 10**(-p_)*1e9, "o"); ax.annotate(lab, (p_, 10**(-p_)*1e9), xytext=(5, 5), textcoords="offset points")
ax.invert_xaxis(); ax.set(xlabel="pH", ylabel="[H⁺] (nmol/L)", title="A small pH change is a large change in acidity"); plt.show()
```

**การตีความ** การลดลง 0.11 หน่วย pH คือ**ความเข้มข้นไอออนไฮโดรเจนเพิ่มขึ้น 29 %** ค่าประมาณเชิงเส้น (จากผลต่างเชิงอนุพันธ์) −0.125 ใกล้เคียงค่าจริง −0.11 แต่เกินไปเล็กน้อย เพราะการเปลี่ยนแปลง 29 % ไม่ "เล็ก"
ความเป็นกรดที่สูงขึ้นทำให้ปะการังและสัตว์มีเปลือก (รวมถึงสิ่งมีชีวิตสร้างแนวปะการังในทะเลอันดามัน) สร้างโครงสร้างแคลเซียมคาร์บอเนตได้ยากขึ้น

**ความหมายของคำตอบ** การลดลงของ pH \"เพียง\" 0.11 หน่วย หมายความว่าผิวมหาสมุทรมีไฮโดรเจนไอออนมากขึ้นราว 29 % จากก่อนยุคอุตสาหกรรม ซึ่งเปลี่ยนเคมีของคาร์บอเนต และทำให้ปะการังและสัตว์มีเปลือก รวมถึงสิ่งมีชีวิตสร้างแนวปะการังในทะเลอันดามัน สร้างโครงร่างแคลเซียมคาร์บอเนตได้ยากขึ้น

**ข้อผิดพลาดที่พบบ่อย**

- ลืมตัวประกอบ $1/\ln10$ เมื่อหาอนุพันธ์ของ $\log_{10}$ เพราะ $\frac{d}{dx}\log_{10}x=\frac{1}{x\ln10}$ ไม่ใช่ $\frac1x$
- อ่าน pH เหมือนมาตราส่วนเชิงเส้น การเปลี่ยน 0.11 ไม่ได้แปลว่า \"เป็นกรดเพิ่ม 0.11 %\" เพราะ pH หนึ่งหน่วยคือ 10 เท่า
- ใช้ผลต่างเชิงอนุพันธ์กับการเปลี่ยนแปลงขนาดใหญ่แล้วเชื่อว่าแม่นยำ ที่จริงเป็นเพียงการประมาณอันดับหนึ่ง

**ลองทำดู** ใช้ผลต่างเชิงอนุพันธ์ประมาณการลดลงของ pH จากการเพิ่มขึ้น 10 % ของ $[\mathrm{H}^+]$ แล้วเปรียบเทียบกับค่าแม่นตรง $\log_{10}1.1$

## ตัวอย่างที่ 11 ⭐: น้ำขึ้นเร็วที่สุดเมื่อใด (อนุพันธ์ของฟังก์ชันตรีโกณมิติ)

**บริบท** น้ำขึ้นน้ำลงในอ่าวไทยและทะเลอันดามันเกิดจากการรวมกันของ *องค์ประกอบ (constituents)* รูปไซน์หลายตัว ถ้าใช้สององค์ประกอบกึ่งรายวันที่ใหญ่ที่สุด
คือ M2 (คาบ 12.42 h) และ S2 (คาบ 12.00 h):

$$\eta(t) = 1.2\cos\!\left(\frac{2\pi t}{12.42}\right) + 0.4\cos\!\left(\frac{2\pi t}{12.00} + 0.5\right)\quad(\mathrm{m}).$$

**แคลคูลัส** $\dfrac{d}{dt}\cos(\omega t+\phi) = -\omega\sin(\omega t+\phi)$ อัตรา $\eta'(t)$ (m/h) มีค่ามากที่สุดที่ตำแหน่งซึ่ง $\eta''(t)=0$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** น้ำขึ้นน้ำลงคือผลรวมของคลื่นโคไซน์ หนึ่งคลื่นต่อหนึ่งองค์ประกอบทางดาราศาสตร์ อัตราที่ระดับน้ำสูงขึ้นคืออนุพันธ์ $\eta'(t)$ ซึ่งหาได้จาก $\frac{d}{dt}\cos(\omega t+\phi)=-\omega\sin(\omega t+\phi)$ การขึ้นเร็วที่สุดคือค่าสูงสุดของ $\eta'$ ซึ่งเกิดที่ $\eta''=0$ คือที่จุดเปลี่ยนเว้า (inflection point) ของกราฟน้ำขึ้นน้ำลง ไม่ใช่ตอนน้ำขึ้นสูงสุด

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. แบบจำลองด้วยค่าเริ่มต้น $A_1=1.2$ ม. (M2, 12.42 ชม.) $A_2=0.4$ ม. (S2, 12 ชม.) $\phi=0.5$ เรเดียน: $$\eta(t)=1.2\cos\omega_1t+0.4\cos(\omega_2t+0.5),\quad \omega_1=\tfrac{2\pi}{12.42}=0.5059,\ \omega_2=\tfrac{2\pi}{12}=0.5236\ \text{rad/ชม.}$$
2. หาอนุพันธ์ทีละพจน์ด้วยหลักเกณฑ์ลูกโซ่: $$\eta'(t)=-1.2\,\omega_1\sin\omega_1t-0.4\,\omega_2\sin(\omega_2t+0.5)\ \ \text{ม./ชม.}$$
3. หาจุดที่ขึ้นเร็วที่สุดโดยให้ $\eta''(t)=0$ เมื่อ $\eta''=-1.2\omega_1^2\cos\omega_1t-0.4\omega_2^2\cos(\omega_2t+0.5)$ แก้เชิงตัวเลขในรอบน้ำขึ้นน้ำลงแรก ได้ $t\approx8.98$ ชม. ที่ $\eta'=0.783$ ม./ชม. และ $\eta\approx-0.01$ ม. (ระดับน้ำกลาง)
4. ขอบเขตบนได้จาก $|\sin|\le1$: $|\eta'|\le A_1\omega_1+A_2\omega_2=0.607+0.209=0.817$ ม./ชม. ซึ่งเกิดได้เฉพาะเมื่อสององค์ประกอบมีเฟสตรงกัน (น้ำเกิด) ในช่วง 15 วันของแอป ค่าสูงสุดคือ $0.816$ ม./ชม. ที่ $t\approx332$ ชม.
5. จังหวะการเหลื่อมของคาบ: ผลต่างเฟสเปลี่ยนไป $\omega_2-\omega_1$ ต่อชั่วโมง น้ำเกิดจึงวนกลับมาทุก $2\pi/(\omega_2-\omega_1)\approx354$ ชม. $=14.77$ วัน พิสัยน้ำขึ้นน้ำลงที่มากที่สุดคือ $2(A_1+A_2)=3.2$ ม.

```python
t = sp.symbols('t', real=True)
eta = 1.2*sp.cos(2*sp.pi*t/12.42) + 0.4*sp.cos(2*sp.pi*t/12 + 0.5)
deta, d2eta = sp.diff(eta, t), sp.diff(eta, t, 2)
show(r"\eta'(t)", deta)

f_eta, f_d = sp.lambdify(t, eta), sp.lambdify(t, deta)
tt = np.linspace(0, 30, 3001)
i = np.argmax(f_d(tt[:1300]))                        # fastest rise in the first ~13 h
t_star = optimize.brentq(sp.lambdify(t, d2eta), tt[i]-1, tt[i]+1)
print(f"fastest rise at t ≈ {t_star:.2f} h, rate = {f_d(t_star):.3f} m/h, tide level = {f_eta(t_star):.2f} m")
```

```python
fig, ax = plt.subplots()
ax.plot(tt, f_eta(tt), lw=2, label="η(t), tide height (m)")
ax.plot(tt, f_d(tt), "C1", lw=1.5, label="η'(t), rate (m/h)")
ax.axvline(t_star, c="gray", ls="--"); ax.axhline(0, c="k", lw=.6)
ax.set(xlabel="time (h)", title="Fastest rise happens near mid-tide, not at high tide"); ax.legend(); plt.show()
```

**การตีความ** น้ำขึ้นเร็วที่สุดใกล้ **ระดับน้ำกึ่งกลาง (mid-tide)** ($\eta \approx 0$) และอัตราเป็นศูนย์ ณ เวลาน้ำขึ้นสูงสุดและน้ำลงต่ำสุด
นี่คือพื้นฐานทางคณิตศาสตร์ของ "กฎหนึ่งในสิบสอง (rule of twelfths)" ของนักเดินเรือ และเป็นเหตุที่กระแสน้ำขึ้นน้ำลงแรงที่สุดในช่วงระดับน้ำกึ่งกลาง

**ความหมายของคำตอบ** น้ำขึ้นเร็วที่สุดราว **0.8 เมตรต่อชั่วโมง** เมื่อระดับน้ำอยู่ใกล้ระดับกลาง และหยุดนิ่งตอนน้ำขึ้นสูงสุดและลงต่ำสุด นี่คือเหตุที่กระแสน้ำในปากแม่น้ำและท่าเรือแรงที่สุดตอนระดับน้ำกลาง และ \"กฎหนึ่งในสิบสอง (rule of twelfths)\" ของชาวเรือให้การเปลี่ยนแปลงมากที่สุดในชั่วโมงกลาง ๆ น้ำเกิดที่เกิดทุกราว 14.8 วันให้ทั้งพิสัยมากที่สุดและน้ำขึ้นเร็วที่สุด

**ข้อผิดพลาดที่พบบ่อย**

- ลืมตัวคูณ $\omega$ จากหลักเกณฑ์ลูกโซ่ ทำให้ได้ \"อัตรา\" ที่มีหน่วยเป็นเมตร แทนที่จะเป็นเมตรต่อชั่วโมง
- หาจุดที่ขึ้นเร็วที่สุดจาก $\eta'=0$ เงื่อนไขนี้ให้ตำแหน่งน้ำขึ้นสูงสุดและลงต่ำสุด ค่าสูงสุดของ $\eta'$ ต้องใช้ $\eta''=0$
- ใช้องศาแทนเรเดียน สูตรแคลคูลัสของ $\sin$ และ $\cos$ ใช้ได้เฉพาะเรเดียน

**ลองทำดู** เปลี่ยนแอมพลิจูดของ S2 เป็น 1.2 m (น้ำเกิด องค์ประกอบมีเฟสตรงกัน) อัตราสูงสุดเปลี่ยนไปอย่างไร

## ตัวอย่างที่ 12 ⭐: จุดที่ชันที่สุดของผารอยเลื่อน (อนุพันธ์ของ $\arctan$)

**บริบท** ผารอยเลื่อน (fault scarp) คือขั้นบนผิวดินที่เกิดจากการแตกของรอยเลื่อน มีความสูง $H$ และถูกการกร่อนทำให้มนลง แบบจำลองรูปร่างอย่างง่ายคือ

$$z(x) = \frac{H}{\pi}\arctan\!\left(\frac{x}{w}\right) + \frac{H}{2},$$

โดย $w$ (m) บอกว่าผาถูกกร่อนไปมากเพียงใด **มุมความชัน** คือ $\theta(x) = \arctan\big(z'(x)\big)$

**คำถาม** จงหา $z'(x)$ และมุมความชันสูงสุด แล้วตรวจสอบว่ามุมนี้เกินมุมทรงตัว (angle of repose) ของเศษหินร่วน (≈ 35°) หรือไม่

**แคลคูลัส** $\dfrac{d}{dx}\arctan u = \dfrac{1}{1+u^2}\dfrac{du}{dx}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ฟังก์ชันอาร์กแทนเจนต์มีรูปเป็นขั้นบันไดที่ราบเรียบ จึงเหมาะเป็นแบบจำลองผารอยเลื่อน (fault scarp) ที่ถูกกัดเซาะจนมน อนุพันธ์ $\frac{d}{dx}\arctan u=\frac{u'}{1+u^2}$ ให้ความชันที่ทุกจุด และมุมลาดคือ $\theta=\arctan(z')$ จุดที่ชันที่สุดคือจุดที่ $z'$ มากที่สุด ซึ่งด้วยความสมมาตรอยู่ที่กลางผา

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. โปรไฟล์ด้วยค่าเริ่มต้น $H=6$ ม. และ $w=3$ ม.: $z(x)=\dfrac{6}{\pi}\arctan\dfrac{x}{3}+3$ ทางซ้ายสุด $z\to0$ ทางขวาสุด $z\to6$ ม. ระยะเหลื่อมรวมจึงเท่ากับ $H$
2. ใช้หลักเกณฑ์ลูกโซ่กับ $u=x/w$, $u'=1/w$: $$z'(x)=\frac{H}{\pi}\cdot\frac{1/w}{1+x^2/w^2}=\frac{H}{\pi}\cdot\frac{w}{w^2+x^2}$$
3. ตัวส่วน $w^2+x^2$ มีค่าน้อยที่สุดที่ $x=0$ ความชันจึงมากที่สุดที่นั่น: $z'(0)=\dfrac{H}{\pi w}=\dfrac{6}{3\pi}=0.637$
4. แปลงความชัน (ระยะตั้ง/ระยะราบ) เป็นมุม: $\theta_{\max}=\arctan(0.637)=32.5^\circ$ ซึ่งต่ำกว่ามุมพัก (angle of repose) ราว 35° ผาจึงถูกกัดเซาะไปแล้ว เปรียบเทียบ $w=1$ ม. ให้ 62.4° (ยังสด) และ $w=8$ ม. ให้ 13.4°
5. แก้ย้อนกลับหาความกว้างที่ทำให้ $\theta_{\max}=35^\circ$: $\tan35^\circ=\dfrac{H}{\pi w}\Rightarrow w=\dfrac{6}{\pi\tan35^\circ}=2.73$ ม.

```python
x = sp.symbols('x', real=True); H, w = sp.symbols('H w', positive=True)
zs = H/sp.pi*sp.atan(x/w) + H/2
dz = sp.simplify(sp.diff(zs, x))
show("z'(x)", dz)
theta = sp.atan(dz)
show(r"\theta_{\max}=\theta(0)", theta.subs(x, 0))

for Hv, wv in [(6, 1.0), (6, 3.0), (6, 8.0)]:
    th = float(sp.deg(theta.subs({x: 0, H: Hv, w: wv})))
    print(f"H = {Hv} m, w = {wv} m -> max slope {th:5.1f}°  {'(steeper than repose: unstable, still fresh)' if th > 35 else '(below repose: degraded)'}")
```

```python
xx = np.linspace(-30, 30, 400)
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4))
for wv in [1, 3, 8]:
    a1.plot(xx, sp.lambdify(x, zs.subs({H: 6, w: wv}))(xx), label=f"w = {wv} m")
    a2.plot(xx, np.degrees(sp.lambdify(x, theta.subs({H: 6, w: wv}))(xx)), label=f"w = {wv} m")
a2.axhline(35, c="k", ls="--", lw=1); a2.text(-29, 37, "angle of repose ≈ 35°")
a1.set(xlabel="distance across scarp x (m)", ylabel="elevation z (m)", title="scarp profile"); a1.legend()
a2.set(xlabel="x (m)", ylabel="slope angle θ (°)", title="θ(x) = arctan z'(x)"); plt.tight_layout(); plt.show()
```

**การตีความ** $\theta_{\max} = \arctan\!\big(H/(\pi w)\big)$ ที่จุดกึ่งกลางของผา เมื่อการกร่อนทำให้ $w$ เพิ่มขึ้น ผาก็จะลาดลง
นักธรณีสัณฐานวิทยาใช้หลักนี้ประมาณว่าแผ่นดินไหวเกิดขึ้น *นานเท่าใดแล้ว* (การหาอายุผารอยเลื่อน)

**ความหมายของคำตอบ** ผาสูง 6 ม. ที่มี $w=3$ ม. มีความชันสูงสุด **32.5°** ซึ่งลาดน้อยกว่าที่เศษหินร่วนจะตั้งอยู่ได้ จึงไม่ใช่รอยแตกที่เพิ่งเกิดแล้ว การกัดเซาะทำให้ $w$ กว้างขึ้นตามเวลา การวัดความชันสูงสุดของผาในภาคสนาม (หรือจากแบบจำลองความสูงเชิงเลข DEM) จึงเป็นวิธีประมาณว่าแผ่นดินไหวเกิดขึ้นนานเท่าใดแล้ว

**ข้อผิดพลาดที่พบบ่อย**

- ถือว่าความชัน $z'$ เป็นมุม ความชัน 0.637 ไม่ใช่ 0.637 เรเดียนหรือ 0.637° ต้องใส่ $\arctan$ ก่อน
- ลืมอนุพันธ์ของฟังก์ชันข้างใน $1/w$ เมื่อหาอนุพันธ์ของ $\arctan(x/w)$
- ตั้งเครื่องคิดเลขเป็นโหมดองศาขณะใช้สูตรที่พิสูจน์มาในหน่วยเรเดียน

**ลองทำดู** สำหรับ $H = 6$ m จงหาค่า $w$ ที่ทำให้ $\theta_{\max}$ เท่ากับ 35° พอดี

## ตัวอย่างที่ 13 ⭐: มุมเทปรากฏในหน้าตัดถนน (อนุพันธ์ของ arctan)

**บริบท** ชั้นหินเอียงเทด้วย**มุมเทจริง (true dip)** $\delta = 40^\circ$ หน้าตัดถนนหรือภาพตัดขวางที่ไม่ตั้งฉากกับแนวระดับ (strike) จะแสดง**มุมเทปรากฏ (apparent dip)** $\alpha$ ที่เล็กกว่า:

$$\tan\alpha = \tan\delta\,\sin\beta \quad\Longrightarrow\quad \alpha(\beta) = \arctan(\tan\delta\,\sin\beta),$$

โดยที่ $\beta$ คือมุมระหว่างแนวระดับกับแนวภาพตัดขวาง

**แคลคูลัส** $\dfrac{d\alpha}{d\beta} = \dfrac{\tan\delta\cos\beta}{1+\tan^2\delta\,\sin^2\beta}$ (กฎลูกโซ่กับ $\dfrac{d}{du}\arctan u = \dfrac1{1+u^2}$)

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** มุมเทปรากฏ (apparent dip) ที่เห็นในภาคตัดขวางคือ $\alpha=\arctan(\tan\delta\sin\beta)$ การหาอนุพันธ์ของ **ฟังก์ชันตรีโกณมิติผกผัน (inverse trigonometric function)** ด้วยหลักเกณฑ์ลูกโซ่บอกว่ามุมเทปรากฏไวต่อทิศทางของภาคตัดเพียงใด

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **คำนวณมุมเทปรากฏที่ค่าตั้งต้น** เมื่อ $\delta=40^\circ$ และ $\beta=30^\circ$: $\tan40^\circ=0.8391$ ดังนั้น $u=\tan\delta\sin\beta=0.8391\times0.5=0.4195$ และ $\alpha=\arctan0.4195\approx22.76^\circ$
2. **หาอนุพันธ์ด้วยหลักเกณฑ์ลูกโซ่** ให้ $u=\tan\delta\sin\beta$ ดังนั้น $du/d\beta=\tan\delta\cos\beta$ ($\delta$ คงที่) และ $\dfrac{d}{du}\arctan u=\dfrac1{1+u^2}$ จะได้ $$\frac{d\alpha}{d\beta}=\frac{\tan\delta\cos\beta}{1+\tan^2\delta\sin^2\beta}$$
3. **แทนค่าที่ $\beta=30^\circ$** ตัวเศษ $0.8391\times0.8660=0.7267$ ตัวส่วน $1+0.4195^2=1.1760$ ดังนั้น $d\alpha/d\beta\approx0.618$ เนื่องจากมุมทั้งสองใช้หน่วยเดียวกัน ค่านี้จึงเท่ากับ 0.618 องศาต่อองศาด้วย
4. **ตรวจกรณีสุดขั้ว** ที่ $\beta=90^\circ$: $\cos\beta=0$ ดังนั้น $d\alpha/d\beta=0$ และ $\alpha=\delta=40^\circ$ ที่ $\beta=0$: $d\alpha/d\beta=\tan\delta\approx0.839$ ซึ่งเป็นความไวสูงสุด
5. **ใช้ผลต่างเชิงอนุพันธ์** ถ้าทิศทางของภาคตัดคลาดเคลื่อน $\pm2^\circ$ ที่ $\beta=30^\circ$ จะได้ $d\alpha\approx0.618\times2\approx1.2^\circ$

```python
b = sp.symbols('beta', real=True); dlt = sp.rad(40)
alpha = sp.atan(sp.tan(dlt)*sp.sin(b))
da = sp.simplify(sp.diff(alpha, b))
show(r"\frac{d\alpha}{d\beta}", da)
af, daf = sp.lambdify(b, alpha), sp.lambdify(b, da)
print(f"{'β (°)':>6} {'apparent dip α (°)':>20} {'dα/dβ (°/°)':>14}")
for bd in range(10, 91, 10):
    print(f"{bd:>6} {np.degrees(af(np.radians(bd))):>20.2f} {daf(np.radians(bd)):>14.3f}")
```

```python
B = np.radians(np.linspace(0, 90, 200))
fig, ax = plt.subplots()
for dd in (20, 40, 60):
    ax.plot(np.degrees(B), np.degrees(np.arctan(np.tan(np.radians(dd))*np.sin(B))), lw=2, label=f"true dip δ = {dd}°")
ax.set(xlabel="angle between strike and section β (°)", ylabel="apparent dip α (°)", title="Apparent dip is never larger than the true dip"); ax.legend(); plt.show()
```

**การตีความ** เมื่อภาพตัดขวางตั้งฉากกับแนวระดับ ($\beta = 90^\circ$) จะได้ $\alpha = \delta = 40^\circ$ และ $d\alpha/d\beta = 0$ ดังนั้นความคลาดเคลื่อนเล็กน้อยของทิศทางภาพตัดขวางแทบไม่มีผล
ที่ $\beta$ เล็ก อนุพันธ์มีค่าใกล้ค่าสูงสุด ($\tan40^\circ \approx 0.84$ °/°) มุมเทปรากฏจึงเปลี่ยนเร็ว นักธรณีวิทยาต้องวัดทิศทางภาพตัดขวางอย่างระมัดระวังเมื่อแนวตัดเกือบขนานกับแนวระดับ

**ความหมายของคำตอบ** มุมเทปรากฏไม่มีวันมากกว่ามุมเทจริง ในภาคตัดที่ทำมุม 30° กับแนวระดับ (strike) ชั้นหินที่เท 40° จะดูเหมือนเทเพียงประมาณ 23° เมื่อภาคตัดเกือบตั้งฉาก ความคลาดเคลื่อนของทิศทางแทบไม่มีผล แต่เมื่อใกล้แนวระดับจะมีผลมาก นักธรณีวิทยาภาคสนามจึงต้องวัดทิศทางของหน้าตัดถนนอย่างระมัดระวัง

**ข้อผิดพลาดที่พบบ่อย**

- ลืมอนุพันธ์ภายใน $\tan\delta\cos\beta$ เมื่อหาอนุพันธ์ของ $\arctan u$
- ถือว่า $\delta$ เป็นตัวแปร มุมเทจริงคงที่ มีเพียง $\beta$ ที่เปลี่ยน $\tan\delta$ จึงเป็นค่าคงตัว
- ตั้งเครื่องคิดเลขผิดโหมดมุมเมื่อคำนวณ $\tan40^\circ$ หรือ $\arctan$

**ลองทำดู** หน้าตัดถนนแสดงมุมเทปรากฏ 25° และแนวระดับทำมุม 35° กับถนน มุมเทจริงเป็นเท่าใด

## ตัวอย่างที่ 14 ⭐⭐: ทำไมสึนามิเคลื่อนที่เร็วเท่าเครื่องบินไอพ่น (ฟังก์ชันไฮเพอร์โบลิก)

**บริบท** คลื่นน้ำที่มีเลขคลื่น $k = 2\pi/\lambda$ ในน้ำลึก $h$ เป็นไปตาม **ความสัมพันธ์การกระจาย (dispersion relation)**
$\omega^2 = gk\tanh(kh)$ อัตราเร็วเฟส (phase speed) คือ

$$c(k,h) = \sqrt{\frac{g}{k}\tanh(kh)}.$$

**คำถาม**
1. จงแสดงว่าสำหรับ **น้ำตื้น** ($kh\to0$) จะได้ $c\to\sqrt{gh}$ และสำหรับ **น้ำลึก** ($kh\to\infty$) จะได้ $c\to\sqrt{g/k}$
2. จงหา **ความเร็วกลุ่ม (group velocity)** $c_g = \dfrac{d\omega}{dk}$ ซึ่งเป็นอัตราเร็วที่พลังงานคลื่นเคลื่อนที่

**แคลคูลัส** $\dfrac{d}{dx}\tanh x = \operatorname{sech}^2 x$; $\displaystyle\lim_{x\to0}\frac{\tanh x}{x}=1$; $\displaystyle\lim_{x\to\infty}\tanh x = 1$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ความเร็วคลื่นน้ำคือ $c=\sqrt{(g/k)\tanh(kh)}$ ฟังก์ชันแทนเจนต์ไฮเพอร์โบลิก (hyperbolic tangent) มีพฤติกรรมต่างกันมากที่สองปลาย: $\tanh x\approx x$ เมื่อ $x$ เล็ก และ $\tanh x\to1$ เมื่อ $x$ ใหญ่ การหา **ลิมิต** ทั้งสองทำให้สูตรเดียวกลายเป็นกฎง่าย ๆ สองข้อ และสึนามิอยู่ในกรณี $kh$ เล็ก ซึ่งความเร็วขึ้นกับความลึกเท่านั้น

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. ค่าเริ่มต้น: ความลึก $h=10^{3.6}\approx3981$ ม. และความยาวคลื่น $\lambda=10^{5.3}\approx199.5$ กม. เลขคลื่น $k=2\pi/\lambda=3.149\times10^{-5}$ ต่อเมตร จึงได้ $kh=0.1254$
2. ลิมิตน้ำตื้น: จากอนุกรม $\tanh x=x-\tfrac{x^3}{3}+\dots$ ได้ $\tanh x\approx x$ เมื่อ $x\to0$ ดังนั้น $$\lim_{kh\to0}c=\sqrt{\frac gk\cdot kh}=\sqrt{gh}$$ ลิมิตน้ำลึก: $\tanh(kh)\to1$ จึงได้ $c\to\sqrt{g/k}$
3. ความเร็วแท้จริง: $\tanh(0.1254)=0.1247$ ดังนั้น $c=\sqrt{\dfrac{9.81}{3.149\times10^{-5}}\times0.1247}=197.1$ ม./วินาที $\approx710$ กม./ชม. ลิมิต $\sqrt{gh}=\sqrt{9.81\times3981}=197.6$ ม./วินาที สูงกว่าเพียง 0.3%
4. ความเร็วกลุ่ม (ความเร็วที่พลังงานเดินทาง) จาก $\omega=ck$: $$c_g=\frac{d\omega}{dk}=\frac c2\left(1+\frac{2kh}{\sinh2kh}\right)=196.1\ \text{ม./วินาที}$$ เนื่องจาก $\sinh x\approx x$ เมื่อ $x$ เล็ก วงเล็บจึงเข้าใกล้ 2 และ $c_g\to c$ ในน้ำตื้น ในน้ำลึกวงเล็บเข้าใกล้ 1 และ $c_g\to c/2$
5. จำแนกประเภท: $kh<\pi/10\approx0.314$ หมายถึงคลื่นน้ำตื้น เมื่อ $kh=0.125$ สึนามิจึงเป็นคลื่นน้ำตื้นแม้ในมหาสมุทรลึก 4 กม.

```python
k, hh, g = sp.symbols('k h g', positive=True)
c = sp.sqrt(g/k*sp.tanh(k*hh))
show(r"\lim_{k\to0} c", sp.limit(c, k, 0))
show(r"\lim_{h\to\infty} c", sp.limit(c, hh, sp.oo))

omega = sp.sqrt(g*k*sp.tanh(k*hh))
cg = sp.simplify(sp.diff(omega, k))
show(r"c_g=\frac{d\omega}{dk}", cg)
show(r"\frac{c_g}{c}", sp.simplify((cg/c).rewrite(sp.exp)).rewrite(sp.sinh).simplify())

gv = 9.81
for depth in [4000, 1000, 100, 10]:
    v = np.sqrt(gv*depth)
    print(f"tsunami over {depth:>5} m of water: c = √(gh) = {v:6.1f} m/s = {v*3.6:6.0f} km/h")
```

```python
lam = np.logspace(0, 6, 400)                      # wavelength 1 m .. 1000 km
kk = 2*np.pi/lam
fig, ax = plt.subplots()
for depth in [10, 100, 4000]:
    ax.loglog(lam, np.sqrt(gv/kk*np.tanh(kk*depth)), lw=2, label=f"h = {depth} m")
    ax.axhline(np.sqrt(gv*depth), ls=":", c="gray")
ax.loglog(lam, np.sqrt(gv/kk), "k--", lw=1, label="deep-water limit √(g/k)")
ax.set(xlabel="wavelength (m)", ylabel="phase speed c (m/s)", title="Long waves feel the bottom: c → √(gh)")
ax.legend(); plt.show()
```

**การตีความ** ความยาวคลื่นของสึนามิ (≈ 100–500 km) ยาวกว่าความลึกของมหาสมุทรมาก จึงเป็นคลื่น *น้ำตื้น* ที่เคลื่อนที่ด้วยอัตราเร็ว $\sqrt{gh}\approx 200$ m/s (≈ 700 km/h) ในมหาสมุทรลึก
เมื่อเข้าใกล้ชายฝั่ง คลื่นจะช้าลง น้ำจึงพอกตัวและคลื่นสูงขึ้น ซึ่งเป็นสิ่งที่เกิดขึ้นตามชายฝั่งอันดามันของประเทศไทยในปี 2004
ผลลัพธ์ $c_g/c = \tfrac12\left(1+\dfrac{2kh}{\sinh 2kh}\right)$ มีค่าตั้งแต่ ½ (น้ำลึก) ถึง 1 (น้ำตื้น)

**ความหมายของคำตอบ** สึนามิเคลื่อนที่ข้ามมหาสมุทรลึกด้วยความเร็วราว **200 ม./วินาที (≈ 710 กม./ชม.)** เท่าเครื่องบินไอพ่น เนื่องจาก $c=\sqrt{gh}$ มันจึงช้าลงเหลือราว 31 ม./วินาที ในน้ำลึก 100 ม. และ 10 ม./วินาที ในน้ำลึก 10 ม. ด้านหลังของคลื่นจึงไล่ทันด้านหน้า และคลื่นสูงขึ้นใกล้ชายฝั่ง เช่นที่ชายฝั่งทะเลอันดามันของไทยใน ค.ศ. 2004 ส่วนคลื่นลมมี $kh\gg1$ จึงไม่ \"รู้สึก\" ถึงพื้นทะเล

**ข้อผิดพลาดที่พบบ่อย**

- คิดว่ามหาสมุทรลึก 4 กม. เป็น \"น้ำลึก\" สำหรับคลื่นทุกชนิด น้ำลึกหรือน้ำตื้นขึ้นกับ $kh$ (ความลึกเทียบกับความยาวคลื่น) ไม่ใช่ความลึกอย่างเดียว
- ใช้ความยาวคลื่น $\lambda$ ในตำแหน่งที่สูตรต้องการเลขคลื่น $k=2\pi/\lambda$
- สับสน $\tanh$ กับ $\tan$: $\tan x$ พุ่งเป็นอนันต์ที่ $\pi/2$ แต่ $\tanh x$ อยู่ระหว่าง $-1$ ถึง $1$ เสมอ

**ลองทำดู** สึนามิปี 2004 ใช้เวลาประมาณ 2 h เดินทางจากแหล่งกำเนิดถึงภูเก็ต (~600 km) แสดงว่าความลึก *เฉลี่ย* เป็นเท่าใด

## ตัวอย่างที่ 15 ⭐⭐: การวัดความลึกน้ำจากภาพถ่ายดาวเทียมของคลื่น (ฟังก์ชันไฮเพอร์โบลิกผกผัน (inverse hyperbolic functions))

**บริบท** ดาวเทียมถ่ายภาพคลื่นที่เคลื่อนเข้าหาชายฝั่งและวัดคาบ $T$ และความยาวคลื่น $L$ ได้ การผกผันความสัมพันธ์การกระจาย (dispersion relation) $\omega^2 = gk\tanh(kh)$ (ตัวอย่างที่ 8) ให้ความลึก:

$$h = \frac{1}{k}\,\operatorname{artanh}\!\left(\frac{\omega^2}{gk}\right),\qquad \omega = \frac{2\pi}{T},\ k = \frac{2\pi}{L}.$$

วิธีนี้ใช้ได้เฉพาะเมื่อ $\omega^2/(gk) < 1$ มิฉะนั้นคลื่นเป็นคลื่นน้ำลึกที่ไม่ "รับรู้" พื้นทะเล

**แคลคูลัส** $\dfrac{d}{dx}\operatorname{artanh}x = \dfrac{1}{1-x^2}$ อนุพันธ์ $dh/dL$ บอกว่าค่าประมาณความลึกไวต่อความคลาดเคลื่อนของความยาวคลื่นที่วัดได้เพียงใด

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ฟิสิกส์ของคลื่นให้ $\omega^2=gk\tanh(kh)$ เพื่อหาความลึกต้องย้อนฟังก์ชัน tanh ซึ่งต้องใช้ **ฟังก์ชันไฮเพอร์โบลิกผกผัน (inverse hyperbolic function)** artanh อนุพันธ์ $1/(1-x^2)$ ของมันยังบอกว่าความคลาดเคลื่อนของความยาวคลื่นที่วัดได้ขยายเป็นความคลาดเคลื่อนของความลึกเท่าใด

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **แก้หา $h$** หารด้วย $gk$: $\tanh(kh)=\dfrac{\omega^2}{gk}=x$ ดังนั้น $kh=\operatorname{artanh}x$ และ $h=\dfrac1k\operatorname{artanh}x$ ซึ่งต้องมี $|x|<1$
2. **แทนค่า $T=10$ s, $L=100$ m** $\omega=2\pi/10=0.6283$ rad/s, $k=2\pi/100=0.06283$ m⁻¹ ดังนั้น $x=\dfrac{0.3948}{9.81\times0.06283}=\dfrac{0.3948}{0.6164}\approx0.6405<1$
3. **คำนวณ artanh ด้วยลอการิทึม** $\operatorname{artanh}x=\tfrac12\ln\dfrac{1+x}{1-x}=\tfrac12\ln\dfrac{1.6405}{0.3595}=\tfrac12\ln4.563\approx0.7590$ จากนั้น $h=0.7590/0.06283\approx12.08$ m
4. **หาอนุพันธ์เทียบกับ $L$** เขียน $h=\dfrac{L}{2\pi}\operatorname{artanh}x$ โดย $x=\dfrac{\omega^2L}{2\pi g}$ ดังนั้น $dx/dL=x/L$ ใช้กฎผลคูณและหลักเกณฑ์ลูกโซ่ได้ $$\frac{dh}{dL}=\frac{1}{2\pi}\left[\operatorname{artanh}x+\frac{x}{1-x^2}\right]$$
5. **คำนวณความไว** $x/(1-x^2)=0.6405/0.5898\approx1.086$ ดังนั้น $dh/dL=(0.7590+1.086)/6.283\approx0.294$ m ต่อ m ความคลาดเคลื่อนของความยาวคลื่น 5 m ทำให้ความลึกคลาดเคลื่อนประมาณ $1.5$ m
6. **หาจุดที่วิธีนี้ใช้ไม่ได้** $x\to1$ เมื่อ $L\to gT^2/(2\pi)=9.81(100)/6.283\approx156.1$ m (ความยาวคลื่นน้ำลึก) ที่จุดนั้น $\operatorname{artanh}x\to\infty$ และ $1/(1-x^2)\to\infty$

```python
L, Tp = sp.symbols('L T', positive=True); gs = sp.Float(9.81)
k = 2*sp.pi/L; w = 2*sp.pi/Tp
x_ = w**2/(gs*k)
h = sp.atanh(x_)/k
dhdL = sp.diff(h, L)
show(r"\frac{dh}{dL}", sp.simplify(dhdL))
vals = {Tp: 10, L: 100}
print(f"ω²/(gk) = {float(x_.subs(vals)):.4f} (< 1, so the wave feels the bottom)")
print(f"depth h = {float(h.subs(vals)):.2f} m")
print(f"dh/dL = {float(dhdL.subs(vals)):.3f} m per m -> a 5 m error in wavelength gives ≈ {5*float(dhdL.subs(vals)):.1f} m error in depth")
```

```python
Tv = 10; Ls = np.linspace(40, 150, 300)
xx = (2*np.pi/Tv)**2/(9.81*2*np.pi/Ls)
ok = xx < 1
fig, ax = plt.subplots()
ax.plot(Ls[ok], np.arctanh(xx[ok])/(2*np.pi/Ls[ok]), lw=2)
ax.plot(100, float(h.subs(vals)), "C3o")
ax.set(xlabel="measured wavelength L (m), period 10 s", ylabel="depth h (m)", title="Depth from wavelength: artanh blows up near deep water")
ax.set_ylim(0, 60); plt.show()
```

**การตีความ** คลื่นคาบ 10 s ที่มีความยาวคลื่น 100 m บ่งชี้ความลึกประมาณ **12.1 m** ความไวอยู่ที่ประมาณ 0.29 m ของความลึกต่อความยาวคลื่น 1 เมตร
เมื่อ $\omega^2/(gk)\to1$ ค่า artanh (และอนุพันธ์ $1/(1-x^2)$) จะพุ่งสู่อนันต์ วิธีนี้จึงใช้ไม่ได้ในน้ำลึก ในทางปฏิบัติใช้ทำแผนที่ความลึกได้ถึงประมาณ 15–20 m ซึ่งมีประโยชน์มากสำหรับการทำแผนที่ชายฝั่งน้ำตื้นที่เรือเข้าไม่ถึง

**ความหมายของคำตอบ** คลื่นที่มีคาบ 10 s และความยาวคลื่น 100 m บอกว่าน้ำลึกประมาณ 12 m วิธีนี้ใช้ได้ดีในน้ำตื้นชายฝั่งลึกถึงราว 15–20 m แต่ใช้ไม่ได้เมื่อคลื่นไม่ \"รู้สึก\" ถึงพื้นทะเลอีกต่อไป จึงใช้การหาความลึกจากภาพดาวเทียมเพื่อทำแผนที่ชายฝั่งตื้นและแนวปะการังที่เรือสำรวจเข้าไม่ถึง

**ข้อผิดพลาดที่พบบ่อย**

- ใช้ artanh เมื่อ $x\ge1$ ฟังก์ชัน artanh นิยามเฉพาะ $|x|<1$ ในกรณีนั้นเป็นคลื่นน้ำลึกและหาความลึกไม่ได้
- สับสน $\operatorname{artanh}x$ กับ $1/\tanh x$ ฟังก์ชันผกผันไม่ใช่ส่วนกลับ
- ลืมว่า $k$ ขึ้นกับ $L$ ด้วยเมื่อหา $dh/dL$ ทำให้ขาดพจน์จากกฎผลคูณ

**ลองทำดู** คง $T = 10$ s ไว้ ความยาวคลื่นเท่าใดที่สอดคล้องกับความลึก 5 m

# ส่วน C: อนุพันธ์อันดับสูง การหาอนุพันธ์โดยปริยาย และผลต่างเชิงอนุพันธ์ (Higher-Order Derivatives, Implicit Differentiation and Differentials)

⭐⭐ ปานกลาง · ตัวอย่างที่ 16–21 · สัปดาห์ที่ 5–6 ในประมวลรายวิชา

## ตัวอย่างที่ 16 ⭐⭐: การกระจัด ความเร็ว และความเร่งจากคลื่นไหวสะเทือน (อนุพันธ์อันดับสูง)

**บริบท** วิศวกรออกแบบอาคารโดยพิจารณา **ความเร่งสูงสุดของพื้นดิน (peak ground acceleration, PGA)** แบบจำลองอย่างง่ายของการกระจัดของพื้นดิน ณ จุดหนึ่งคือการสั่นแบบหน่วง

$$u(t) = A\,e^{-\zeta\omega t}\sin(\omega t),\qquad A = 5\ \text{cm},\ \omega = 2\pi\cdot 1.5\ \text{rad/s},\ \zeta = 0.15.$$

**แคลคูลัส** ความเร็ว $v = u'(t)$ ความเร่ง $a = u''(t)$ (ใช้กฎผลคูณและกฎลูกโซ่สองครั้ง)

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ความเร็วคืออนุพันธ์อันดับหนึ่งของการกระจัด และความเร่งคืออนุพันธ์อันดับสอง สำหรับการสั่นแบบหน่วง $u=Ae^{-\alpha t}\sin\omega t$ ต้องใช้ **กฎผลคูณ (product rule)** และ **หลักเกณฑ์ลูกโซ่** สองครั้ง การหาอนุพันธ์แต่ละครั้งนำตัวคูณประมาณ $\omega$ ออกมา ความเร่งจึงมีขนาดราว $\omega^2A$ นี่คือเหตุที่การสั่นความถี่สูงสร้างแรงขนาดใหญ่ต่ออาคาร

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. ค่าเริ่มต้น: $A=5$ ซม. $f=1.5$ เฮิรตซ์ $\zeta=0.15$ ความถี่เชิงมุม $\omega=2\pi f=9.425$ ต่อวินาที และอัตราการหน่วง $\alpha=\zeta\omega=1.414$ ต่อวินาที
2. อนุพันธ์อันดับหนึ่งด้วยกฎผลคูณ $(fg)'=f'g+fg'$ โดย $f=Ae^{-\alpha t}$ และ $g=\sin\omega t$: $$v=u'=-\alpha Ae^{-\alpha t}\sin\omega t+Ae^{-\alpha t}\omega\cos\omega t=Ae^{-\alpha t}(\omega\cos\omega t-\alpha\sin\omega t)$$
3. หาอนุพันธ์อีกครั้ง (ใช้กฎผลคูณกับ $v$) แล้วจัดพจน์: $$a=u''=Ae^{-\alpha t}\big[(\alpha^2-\omega^2)\sin\omega t-2\alpha\omega\cos\omega t\big]$$
4. ค่าสูงสุด (ค่า $|u|,|v|,|a|$ ที่มากที่สุดตลอดบันทึก): การกระจัด $3.99$ ซม. PGV $=47.1$ ซม./วินาที (ที่ $t=0$ ซึ่ง $v=A\omega$) PGA $=379$ ซม./วินาที² $=0.387$ g เมื่อหารด้วย $g=981$ ซม./วินาที²
5. ตรวจสอบอย่างเร็วด้วย $\omega^2A$: $(9.425)^2\times5=444$ ซม./วินาที² $=0.453$ g ซึ่งสูงไปเล็กน้อย เพราะการหน่วงและการกระจัดสูงสุดจริง (3.99 ซม. ไม่ใช่ 5 ซม.) ทำให้ค่าจริงลดลง

```python
t = sp.symbols('t', nonnegative=True)
A, z_, w_ = 0.05, 0.15, 2*sp.pi*1.5
u = A*sp.exp(-z_*w_*t)*sp.sin(w_*t)
v = sp.diff(u, t); a = sp.diff(u, t, 2)
show("v(t)=u'(t)", sp.simplify(v))
show("a(t)=u''(t)", sp.simplify(a))

tt = np.linspace(0, 6, 4000)
U, V, Aa = (sp.lambdify(t, e)(tt) for e in (u, v, a))
print(f"peak displacement {np.abs(U).max()*100:.2f} cm | PGV {np.abs(V).max()*100:.1f} cm/s | "
      f"PGA {np.abs(Aa).max():.2f} m/s² = {np.abs(Aa).max()/9.81:.2f} g")
```

```python
fig, axs = plt.subplots(3, 1, figsize=(8, 6.5), sharex=True)
for ax, y, lab, col in zip(axs, (U*100, V*100, Aa/9.81), ("u (cm)", "v (cm/s)", "a (g)"), ("C0", "C1", "C3")):
    ax.plot(tt, y, col); ax.set_ylabel(lab)
axs[0].set_title("Each derivative shifts the phase and multiplies by about ω ≈ 9.4 s⁻¹")
axs[-1].set_xlabel("time (s)"); plt.tight_layout(); plt.show()
```

**การตีความ** การหาอนุพันธ์แต่ละครั้งคูณแอมพลิจูดด้วยประมาณ $\omega$ การกระจัดเพียงไม่กี่ cm จึงกลายเป็นความเร่งประมาณ 0.4 g ซึ่งมากพอจะทำให้อาคารที่ไม่แข็งแรงเสียหาย
การสั่นความถี่สูง ($\omega$ มาก) ให้ความเร่งสูง ข้อกำหนดการออกแบบต้านแผ่นดินไหวจึงให้ความสำคัญกับการสั่นความถี่สูงสำหรับอาคารเตี้ยที่แข็งเกร็ง

**ความหมายของคำตอบ** การกระจัดของพื้นดินเพียง 4 ซม. ที่ความถี่ 1.5 เฮิรตซ์ ให้ความเร่งสูงสุดราว **0.39 g** ซึ่งพอจะทำให้อาคารที่ไม่แข็งแรงเสียหาย การเพิ่มความถี่เป็นสองเท่าทำให้ PGA เพิ่มราวสี่เท่า ($\propto\omega^2$) อาคารเตี้ยและแข็งที่ตอบสนองต่อความถี่สูงจึงเสี่ยงที่สุด วิศวกรหาปริพันธ์ของบันทึกความเร่งเพื่อได้ความเร็วและการกระจัด ซึ่งเป็นการคำนวณย้อนกลับของตัวอย่างนี้

**ข้อผิดพลาดที่พบบ่อย**

- หาอนุพันธ์ของ $e^{-\alpha t}\sin\omega t$ เป็น $(-\alpha e^{-\alpha t})(\omega\cos\omega t)$ อนุพันธ์ของผลคูณไม่ใช่ผลคูณของอนุพันธ์
- ใช้ $f$ (เฮิรตซ์) แทน $\omega=2\pi f$ (เรเดียน/วินาที) ทำให้ทุกค่าน้อยไป $2\pi$ เท่า
- ลืมแปลง ซม./วินาที² เป็นหน่วย g (หารด้วย 981 ซม./วินาที²)

**ลองทำดู** เปลี่ยนความถี่เป็น 5 Hz โดยคง $A$ ไว้เท่าเดิม PGA เพิ่มขึ้นกี่เท่า เปรียบเทียบกับ $\omega^2$

## ตัวอย่างที่ 17 ⭐⭐: ลาดเขาถูกกัดเซาะตรงไหน (อนุพันธ์อันดับสองและความเว้า (concavity))

**บริบท** บนเนินเขาที่มีดินปกคลุม ดินจะคืบลงตามลาด (soil creep) ด้วยอัตราที่แปรผันตามความชัน ซึ่งนำไปสู่**สมการการแพร่ของลาดเขา (hillslope diffusion equation)**

$$\frac{\partial z}{\partial t} = D\,\frac{\partial^2 z}{\partial x^2},\qquad D = 0.01\ \text{m}^2/\text{yr}.$$

การเปลี่ยนแปลงระดับความสูงขึ้นกับ**อนุพันธ์อันดับสอง** (ความโค้ง) โปรไฟล์: $z(x) = \dfrac{H}{1+e^{(x-x_0)/w}}$ โดย $H = 50$ m, $x_0 = 100$ m, $w = 20$ m

**แคลคูลัส** $z''<0$ (โค้งนูน ด้านยอดเขา) ⇒ ระดับลดลง (การกัดเซาะ); $z''>0$ (โค้งเว้า ตีนลาด) ⇒ ระดับสูงขึ้น (การทับถม) **จุดเปลี่ยนเว้า (inflection point)** $z''=0$ แบ่งสองบริเวณนี้

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ในสมการการแพร่ของลาดเขา $\partial z/\partial t=D\,z''$ การกัดเซาะหรือการสะสมขึ้นกับ **อนุพันธ์อันดับสอง (second derivative)** หรือความโค้งเท่านั้น ส่วนที่นูน ($z''<0$) พื้นจะต่ำลง ส่วนที่เว้า ($z''>0$) พื้นจะสูงขึ้น และ **จุดเปลี่ยนเว้า (inflection point)** แบ่งสองส่วนนี้

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **อนุพันธ์อันดับหนึ่ง** ให้ $u=(x-x_0)/w$ และ $s=e^u$ ดังนั้น $z=H/(1+s)$ ใช้หลักเกณฑ์ลูกโซ่ได้ $$z'=-\frac{H}{w}\,\frac{s}{(1+s)^2}<0$$ เนินจึงลาดลงตลอด
2. **อนุพันธ์อันดับสอง** หาอนุพันธ์ของ $s/(1+s)^2$ ด้วยกฎผลหาร (และ $ds/dx=s/w$) $$z''=-\frac{H}{w^2}\,\frac{s(1-s)}{(1+s)^3}$$
3. **จุดเปลี่ยนเว้า** $z''=0$ เมื่อ $s=1$ คือ $u=0$ ดังนั้น $x=x_0=100$ m เมื่อ $x<100$ จะได้ $s<1$ และ $z''<0$ (นูน กัดเซาะ) เมื่อ $x>100$ จะได้ $s>1$ และ $z''>0$ (เว้า สะสม)
4. **กัดเซาะเร็วที่สุดตรงไหน** หาค่าสูงสุดของ $g(s)=s(1-s)/(1+s)^3$ จาก $g'(s)=0$ ได้ $1-4s+s^2=0$ ดังนั้น $s=2-\sqrt3\approx0.268$ คือ $x=100+20\ln(0.268)\approx73.7$ m ซึ่ง $g\approx0.0962$
5. **แปลงเป็นอัตรา** $H/w^2=50/400=0.125$ m⁻¹ ดังนั้น $\partial z/\partial t=-D(0.125)(0.0962)=-0.01\times0.01203\approx-1.20\times10^{-4}$ m/yr $=-0.120$ mm/yr จากความสมมาตร การสะสมสูงสุด $+0.120$ mm/yr อยู่ที่ $x\approx126.3$ m

```python
x = sp.symbols('x', real=True)
H, x0, w, Dc = 50, 100, 20, 0.01
zx = H/(1 + sp.exp((x - x0)/w))
z2 = sp.diff(zx, x, 2)
infl = sp.nsolve(z2, x, 95)
print(f"inflection point: x = {float(infl):.2f} m (= x0)")
z2f = sp.lambdify(x, z2)
X = np.linspace(0, 200, 2001)
rate = Dc*z2f(X)                                   # dz/dt in m/yr
print(f"max erosion rate    = {-rate.min()*1000:.3f} mm/yr at x = {X[np.argmin(rate)]:.1f} m")
print(f"max deposition rate = {rate.max()*1000:.3f} mm/yr at x = {X[np.argmax(rate)]:.1f} m")
```

```python
zf = sp.lambdify(x, zx)
fig, (a1, a2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)
a1.plot(X, zf(X), lw=2); a1.axvline(float(infl), c="gray", ls=":"); a1.set(ylabel="elevation z (m)", title="Convex top erodes, concave foot fills")
a2.plot(X, rate*1000, "C3", lw=2); a2.axhline(0, c="k", lw=.6)
a2.fill_between(X, rate*1000, 0, where=rate < 0, color="C1", alpha=.4, label="erosion (z″ < 0)")
a2.fill_between(X, rate*1000, 0, where=rate > 0, color="C0", alpha=.4, label="deposition (z″ > 0)")
a2.set(xlabel="x (m)", ylabel="dz/dt (mm/yr)"); a2.legend(); plt.tight_layout(); plt.show()
```

**การตีความ** จุดเปลี่ยนเว้าอยู่ที่ $x = 100$ m ด้านบนลาดของจุดนี้โปรไฟล์โค้งนูนและถูกกัดเซาะในอัตราสูงสุดประมาณ **0.12 mm/yr** ด้านล่างลาดโค้งเว้าและได้รับดินเพิ่มในปริมาณเท่ากัน
เมื่อเวลาผ่านไปกระบวนการนี้ทำให้ยอดเขามนและเติมพื้นหุบเขา ภูมิประเทศเก่าจึงดูเรียบ

**ความหมายของคำตอบ** ส่วนบนของเนินที่นูนสูญเสียดินได้ถึงประมาณ 0.12 mm ต่อปี และส่วนล่างที่เว้าได้รับดินเท่ากัน ตลอดหลายพันปีกระบวนการนี้ทำให้ยอดเนินมนและหุบเขาถูกถม ภูมิประเทศเก่าจึงดูราบเรียบ เนื่องจาก $z''\propto H/w^2$ เนินที่กว้างขึ้นสองเท่าจะกัดเซาะช้าลงสี่เท่า

**ข้อผิดพลาดที่พบบ่อย**

- คิดว่าจุดที่ชันที่สุดถูกกัดเซาะเร็วที่สุด การกัดเซาะขึ้นกับความโค้ง $z''$ ไม่ใช่ความชัน $z'$ จุดที่ชันที่สุดคือจุดเปลี่ยนเว้าซึ่งไม่มีการเปลี่ยนแปลง
- สับสนเครื่องหมาย $z''<0$ หมายถึงเว้าคว่ำ (ยอดเนินนูน) และให้ $\partial z/\partial t<0$ คือพื้นต่ำลง
- ลืมตัวประกอบ $1/w$ จากหลักเกณฑ์ลูกโซ่ทุกครั้งที่หาอนุพันธ์ ทำให้ $z''$ ผิดไป $w^2$ เท่า

**ลองทำดู** เพิ่ม $w$ เป็นสองเท่าเป็น 40 m (เนินที่ลาดน้อยกว่า) อัตราการกัดเซาะสูงสุดเปลี่ยนไปกี่เท่า (คำใบ้: $z''\propto H/w^2$)

## ตัวอย่างที่ 18 ⭐⭐: ระดับน้ำใต้ดินระหว่างแม่น้ำสองสาย (การหาอนุพันธ์โดยปริยาย)

**บริบท** น้ำใต้ดินไหลผ่านชั้นหินอุ้มน้ำไม่ถูกปิดทับ (unconfined aquifer) จากแม่น้ำที่มีระดับน้ำ $h_1 = 20$ m ไปยังแม่น้ำอีกสายที่ $h_2 = 12$ m ซึ่งห่างกัน $L = 500$ m
ทฤษฎีของดูปุย–ฟอร์ชไฮเมอร์ (Dupuit–Forchheimer) ให้ความสูงของระดับน้ำใต้ดิน $h(x)$ ใน **รูปโดยปริยาย**:

$$h^2 = h_1^2 - \left(h_1^2-h_2^2\right)\frac{x}{L}.$$

**คำถาม** จงหาความชันของระดับน้ำใต้ดิน $\dfrac{dh}{dx}$ โดยการหาอนุพันธ์โดยปริยาย แล้วแสดงว่าอัตราการไหลต่อหนึ่งหน่วยความกว้าง
$q = -K\,h\,\dfrac{dh}{dx}$ มีค่า **คงที่** (การอนุรักษ์มวล) ใช้ $K = 10$ m/day

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** สมการดูปุย (Dupuit) ให้ระดับน้ำใต้ดินในรูป $h^2$ ไม่ใช่ $h$ ความสัมพันธ์ระหว่าง $h$ กับ $x$ จึงหาอนุพันธ์ได้ง่ายที่สุดแบบ **โดยปริยาย (implicit)** คือหาอนุพันธ์ทั้งสองข้างเทียบกับ $x$ โดยถือว่า $h$ เป็นฟังก์ชันของ $x$ แล้วแก้หา $dh/dx$ จากนั้นกฎของดาร์ซี (Darcy's law) เปลี่ยนความชันเป็นอัตราการไหล

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. ค่าเริ่มต้น: $h_1=20$ ม. $h_2=12$ ม. $L=500$ ม. $K=10$ ม./วัน ระดับน้ำใต้ดินสอดคล้องกับ $h^2=400-256\,\dfrac{x}{500}$ เพราะ $h_1^2-h_2^2=400-144=256$ ม.²
2. หาอนุพันธ์ทั้งสองข้างเทียบกับ $x$ ด้านซ้ายใช้หลักเกณฑ์ลูกโซ่ได้ $\frac{d}{dx}h^2=2h\frac{dh}{dx}$: $$2h\frac{dh}{dx}=-\frac{256}{500}\ \Rightarrow\ \frac{dh}{dx}=-\frac{256}{1000\,h}$$
3. คำนวณที่แม่น้ำทั้งสองสาย: ที่ $x=0$ ($h=20$) $dh/dx=-0.0128$ ที่ $x=L$ ($h=12$) $dh/dx=-0.0213$ ระดับน้ำใต้ดินชันขึ้นทางท้ายน้ำซึ่งชั้นน้ำบางกว่า ที่กึ่งกลาง $h=\sqrt{400-128}=16.49$ ม. สูงกว่าค่าจากเส้นตรง 16 ม. เล็กน้อย
4. กฎของดาร์ซีสำหรับชั้นหินอุ้มน้ำไร้แรงดัน: อัตราการไหลต่อความยาวตลิ่งหนึ่งเมตร $q=-Kh\dfrac{dh}{dx}$ แทนผลจากการหาอนุพันธ์โดยปริยาย $h$ ตัดกันหมด: $$q=Kh\cdot\frac{256}{1000h}=\frac{K(h_1^2-h_2^2)}{2L}=\frac{10\times256}{1000}=2.56\ \text{ม.}^2/\text{วัน}$$
5. ขยายขนาด: ตลอดแม่น้ำยาว 1 กม. น้ำไหลจากแม่น้ำสายบนไปสายล่าง $2.56\times1000=2560$ ม.³/วัน

```python
x, h = sp.symbols('x h', positive=True)
h1, h2, L, K = 20, 12, 500, 10
G = h**2 - (h1**2 - (h1**2 - h2**2)*x/L)          # G(x, h) = 0
dhdx = sp.idiff(G, h, x)                            # implicit derivative
show(r"\frac{dh}{dx}", dhdx)
q = sp.simplify(-K*h*dhdx)
show(r"q=-Kh\frac{dh}{dx}", q)
print(f"q = {float(q):.2f} m²/day per metre of river bank (does not depend on x or h)")
```

```python
xx = np.linspace(0, L, 200)
hx = np.sqrt(h1**2 - (h1**2 - h2**2)*xx/L)
fig, ax = plt.subplots()
ax.fill_between(xx, 0, hx, color="C0", alpha=.25, label="saturated aquifer")
ax.plot(xx, hx, "C0", lw=2, label="water table h(x)")
ax.plot(xx, h1 - (h1 - h2)*xx/L, "k--", lw=1, label="straight line (for comparison)")
ax.set(xlabel="x (m)", ylabel="head h (m)", ylim=(0, 22), title="Water table steepens downstream as the aquifer thins")
ax.legend(); plt.show()
```

**การตีความ** $\dfrac{dh}{dx} = -\dfrac{h_1^2-h_2^2}{2Lh}$: ความชัน **มากขึ้นในบริเวณที่ชั้นหินอุ้มน้ำบางลง** เพราะปริมาณน้ำเท่าเดิมต้องไหลผ่านความหนาอิ่มตัวที่น้อยลง
ผลคูณ $h\,h'$ คงที่ ดังนั้น $q$ มีค่าเท่ากันทุกตำแหน่ง ไม่มีน้ำเกิดขึ้นใหม่หรือสูญหายไป

**ความหมายของคำตอบ** ระดับน้ำใต้ดินเป็นเส้นโค้ง ไม่ใช่เส้นตรง และชันขึ้นในบริเวณที่ชั้นน้ำบางลง เพื่อให้ $h\,dh/dx$ คงที่ ค่าคงที่นี้คือการแสดง **การอนุรักษ์มวล** คือน้ำ 2.56 ม.²/วัน เท่ากันไหลผ่านทุกหน้าตัด นักอุทกธรณีวิทยาใช้หลักนี้ประมาณว่าแม่น้ำได้รับน้ำจากหรือสูญเสียน้ำให้น้ำใต้ดินเท่าใด

**ข้อผิดพลาดที่พบบ่อย**

- เขียน $\frac{d}{dx}h^2=2h$ แล้วลืมตัวคูณ $dh/dx$ ในการหาอนุพันธ์โดยปริยาย ทุกพจน์ที่มี $h$ ต้องมีตัวคูณนี้
- สมมติว่าระดับน้ำใต้ดินเป็นเส้นตรงระหว่างแม่น้ำสองสาย ซึ่งจะทำให้อัตราการไหลไม่คงที่และผิด
- ใช้หน่วยปนกัน $K$ หน่วย ม./วัน ให้ $q$ หน่วย ม.²/วัน ถ้าใช้ $K$ หน่วย ม./วินาที โดยไม่แปลง คำตอบจะต่างกัน 86,400 เท่า

**ลองทำดู** ระดับน้ำในแม่น้ำ $h_2$ เท่าใดจึงทำให้อัตราการไหล $q$ เพิ่มเป็นสองเท่า

## ตัวอย่างที่ 19 ⭐⭐: รูปทรงของโลกและละติจูดสองแบบ (การหาอนุพันธ์โดยปริยาย (implicit differentiation))

**บริบท** ภาคตัดตามแนวเมริเดียนของโลกเป็นวงรี $\dfrac{x^2}{a^2}+\dfrac{z^2}{b^2}=1$ โดยรัศมี WGS84 คือ $a = 6378.137$ km (เส้นศูนย์สูตร) และ $b = 6356.752$ km (ขั้วโลก)
GPS ใช้**ละติจูดเชิงมาตรภูมิศาสตร์ (geodetic latitude)** $\varphi$ ซึ่งเป็นมุมของแนวดิ่งเฉพาะที่ (เส้นแนวฉากของวงรี) ส่วน**ละติจูดเชิงศูนย์กลางโลก (geocentric latitude)** $\psi$ คือมุมที่วัดจากจุดศูนย์กลางโลก

**แคลคูลัส** การหาอนุพันธ์โดยปริยายให้ความชันของเส้นสัมผัส $\dfrac{dz}{dx} = -\dfrac{b^2x}{a^2z}$ เส้นแนวฉากตั้งฉากกับเส้นสัมผัส ดังนั้น $\tan\varphi = \dfrac{a^2z}{b^2x}$ ขณะที่ $\tan\psi = \dfrac zx$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** เส้นเมริเดียนของโลกเป็นวงรีที่กำหนดด้วยสมการใน $x$ และ $z$ ไม่ใช่ในรูป $z=f(x)$ **การหาอนุพันธ์โดยปริยาย (implicit differentiation)** หาความชันได้โดยตรง และเส้นตั้งฉากกับความชันนั้นกำหนดละติจูดที่ GPS ใช้

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **หาอนุพันธ์ทั้งสองข้างเทียบกับ $x$** โดยถือว่า $z$ เป็นฟังก์ชันของ $x$ $$\frac{2x}{a^2}+\frac{2z}{b^2}\frac{dz}{dx}=0\quad\Longrightarrow\quad\frac{dz}{dx}=-\frac{b^2x}{a^2z}$$
2. **หาเส้นตั้งฉาก** ความชันของเส้นตั้งฉากคือส่วนกลับที่ติดลบ $\dfrac{a^2z}{b^2x}$ และมุมที่ทำกับเส้นศูนย์สูตรคือละติจูดเชิงภูมิมาตร (geodetic latitude) $\tan\varphi=\dfrac{a^2z}{b^2x}$
3. **เชื่อมละติจูดทั้งสอง** ละติจูดเชิงศูนย์กลางโลก (geocentric latitude) มี $\tan\psi=z/x$ ดังนั้น $$\tan\varphi=\left(\frac ab\right)^2\tan\psi,\qquad \left(\frac ab\right)^2=\left(\frac{6378.137}{6356.752}\right)^2\approx1.006739$$
4. **แทนค่าที่ $\psi=45^\circ$** $\tan\varphi=1.006739$ ดังนั้น $\varphi=\arctan1.006739\approx45.1924^\circ$ และ $\varphi-\psi\approx0.1924^\circ$
5. **ทำไมค่าสูงสุดอยู่ใกล้ 45°** ให้ $q=(a/b)^2$ ตั้ง $\dfrac{d}{d\psi}\left[\arctan(q\tan\psi)-\psi\right]=0$ ได้ $q\sec^2\psi=1+q^2\tan^2\psi$ คือ $\tan^2\psi=1/q$ ดังนั้น $\psi=\arctan(1/\sqrt q)\approx44.9^\circ$
6. **แปลงเป็นระยะทาง** $0.1924^\circ=0.003358$ rad และ $0.003358\times6371\approx21.4$ km ตามผิวโลก

```python
x, z = sp.symbols('x z', positive=True)
a, b = 6378.137, 6356.752
E = x**2/a**2 + z**2/b**2 - 1
dzdx = sp.idiff(E, z, x)
show(r"\frac{dz}{dx}", sp.simplify(dzdx))
psi = np.radians(np.linspace(0.01, 89.99, 2000))
phi = np.arctan((a/b)**2*np.tan(psi))
diff = np.degrees(phi - psi)
i = np.argmax(diff)
print(f"maximum φ − ψ = {diff[i]:.4f}° at geocentric latitude ψ = {np.degrees(psi[i]):.1f}°")
print(f"that is {np.radians(diff[i])*6371:.1f} km along the surface")
```

```python
fig, ax = plt.subplots()
ax.plot(np.degrees(psi), diff, lw=2)
ax.set(xlabel="geocentric latitude ψ (°)", ylabel="φ − ψ (°)", title="Geodetic minus geocentric latitude"); plt.show()
```

**การตีความ** ละติจูดทั้งสองเท่ากันที่เส้นศูนย์สูตรและขั้วโลก แต่ต่างกันได้ถึง **0.19°** ใกล้ 45° ซึ่งคิดเป็นประมาณ 21 km บนพื้นดิน
เครื่องรับ GPS แผนที่ และกรมแผนที่ทหารล้วนใช้ละติจูดเชิงมาตรภูมิศาสตร์ ดังนั้นการหาอนุพันธ์โดยปริยายจึงอยู่ในทุกพิกัดที่คุณอ่านจากโทรศัพท์

**ความหมายของคำตอบ** เพราะโลกแป้นเล็กน้อย แนวดิ่งในท้องถิ่นจึงไม่ได้ชี้ตรงไปยังศูนย์กลางโลก ละติจูดทั้งสองแบบเท่ากันที่เส้นศูนย์สูตรและขั้วโลก แต่ต่างกันได้ถึงประมาณ 0.19° หรือราว 21 km บนพื้นผิว เครื่องรับ GPS และแผนที่ของประเทศใช้ละติจูดเชิงภูมิมาตร อนุพันธ์โดยปริยายนี้จึงอยู่ในทุกพิกัดที่อ่านจากโทรศัพท์

**ข้อผิดพลาดที่พบบ่อย**

- ลืม $dz/dx$ เมื่อหาอนุพันธ์ของ $z^2$ เพราะ $\frac{d}{dx}z^2=2z\,\frac{dz}{dx}$ ไม่ใช่ $2z$
- ใช้ความชันของเส้นสัมผัสเป็นละติจูด ละติจูดนิยามจาก **เส้นตั้งฉาก** (แนวดิ่งท้องถิ่น) ซึ่งมีความชันเป็นส่วนกลับที่ติดลบ
- สมมติว่าโลกเป็นทรงกลมเมื่อทำงานในระดับกิโลเมตร ผลต่าง 21 km มากกว่าความแม่นยำของ GPS มาก

**ลองทำดู** ใช้ $\tan\varphi = (a/b)^2\tan\psi$ หา $\varphi$ สำหรับละติจูดเชิงศูนย์กลางโลกของกรุงเทพฯ $\psi = 13.69^\circ$

## ตัวอย่างที่ 20 ⭐⭐: การปรับแก้ค่าความโน้มถ่วงแบบอากาศอิสระ (ผลต่างเชิงอนุพันธ์และการประมาณเชิงเส้น)

**บริบท** ความโน้มถ่วงลดลงตามความสูง $h$ เหนือระดับน้ำทะเล: $g(h) = \dfrac{GM}{(R+h)^2}$
ก่อนที่การสำรวจความโน้มถ่วงจะเผยให้เห็นธรณีวิทยาใต้ดินได้ ค่าที่อ่านได้ทุกค่าต้องได้รับการปรับแก้ตามความสูงของสถานี

**แคลคูลัส** ผลต่างเชิงอนุพันธ์ (differential): $dg = g'(0)\,dh = -\dfrac{2GM}{R^3}\,dh = -\dfrac{2g_0}{R}\,dh$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** แรงโน้มถ่วงเป็นไปตามกฎกำลังสองผกผัน $g(h)=GM/(R+h)^2$ สำหรับความสูงที่น้อยกว่ารัศมีโลกมาก กราฟเกือบเป็นเส้นตรง **ผลต่างเชิงอนุพันธ์ (differential)** $dg=g'(0)\,dh$ (การประมาณด้วยเส้นสัมผัส) จึงเพียงพอ ตัวเลขเดียวนี้ คือเกรเดียนต์อากาศอิสระ (free-air gradient) ใช้แก้ค่าการวัดแรงโน้มถ่วงทุกครั้งตามความสูงของสถานี

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. แรงโน้มถ่วงที่ระดับน้ำทะเล: $g_0=\dfrac{GM}{R^2}=\dfrac{3.986\times10^{14}}{(6.371\times10^6)^2}=9.8202$ ม./วินาที²
2. หาอนุพันธ์ด้วยกฎยกกำลังกับ $(R+h)^{-2}$: $$g'(h)=-\frac{2GM}{(R+h)^3},\qquad g'(0)=-\frac{2GM}{R^3}=-\frac{2g_0}{R}$$
3. ค่าตัวเลข: $g'(0)=-\dfrac{2\times9.8202}{6.371\times10^6}=-3.083\times10^{-6}$ ต่อวินาที² เมื่อ 1 มิลลิแกล $=10^{-5}$ ม./วินาที² จะได้ $-0.3083$ มิลลิแกลต่อเมตร
4. ใช้ผลต่างเชิงอนุพันธ์ที่ดอยอินทนนท์ $dh=2565$ ม.: $dg=-0.3083\times2565=-790.74$ มิลลิแกล ค่าแท้จริงคือ $g(2565)-g_0=-790.26$ มิลลิแกล
5. ความคลาดเคลื่อนสัมพัทธ์: $\dfrac{|-790.74+790.26|}{790.26}\approx0.06\%$ แม้ที่ความสูง 20 กม. ความคลาดเคลื่อนก็เพียง 0.47% เพราะ $h/R$ เล็กมาก

```python
hs = sp.symbols('h', real=True)
GM, R = 3.986004e14, 6.371e6
gfun = GM/(R + hs)**2
g0 = float(gfun.subs(hs, 0))
dg_dh = sp.diff(gfun, hs).subs(hs, 0)
print(f"g0 = {g0:.4f} m/s²")
print(f"dg/dh at h=0 = {float(dg_dh):.4e} s⁻² = {float(dg_dh)*1e5*1:.4f} mGal per metre")
print("   (1 mGal = 1e-5 m/s²) -> almost exactly the standard free-air gradient 0.3086 mGal/m")

for H in [10, 1000, 5000, 20000]:
    exact = (float(gfun.subs(hs, H)) - g0)*1e5
    approx = float(dg_dh)*H*1e5
    print(f"h = {H:>6} m: exact Δg = {exact:10.3f} mGal | linear dg = {approx:10.3f} mGal | rel. error {abs(approx-exact)/abs(exact)*100:.3f}%")
```

```python
HH = np.linspace(0, 50000, 300)
fig, ax = plt.subplots()
ax.plot(HH/1000, (sp.lambdify(hs, gfun)(HH) - g0)*1e5, lw=2, label="exact Δg")
ax.plot(HH/1000, float(dg_dh)*HH*1e5, "--", label="linear approximation dg = g'(0) dh")
ax.set(xlabel="height above sea level (km)", ylabel="change in gravity (mGal)",
       title="The tangent line is excellent for every land survey (h < 9 km)"); ax.legend(); plt.show()
```

**การตีความ** การประมาณเชิงเส้นให้ค่าประมาณ $-0.308$ mGal ต่อเมตร (ค่าในตำรา 0.3086 ใช้ความโน้มถ่วงอ้างอิงที่ต่างกันเล็กน้อย) โดยมีความคลาดเคลื่อนต่ำกว่า 0.1 % มากสำหรับทุกระดับความสูงบนแผ่นดิน
ความผิดปกติของความโน้มถ่วงจากแหล่งแร่หรือแอ่งตะกอนมักมีขนาดเพียง 1–10 mGal ความคลาดเคลื่อนของความสูงสถานีเพียง 3 m (≈ 1 mGal) จึงมีผล นักสำรวจจึงต้องใช้ความสูงจาก GPS ที่แม่นยำ

**ความหมายของคำตอบ** แรงโน้มถ่วงลดลงราว **0.308 มิลลิแกลทุก ๆ หนึ่งเมตร** ที่สูงขึ้น (ตำราใช้ 0.3086 ซึ่งมาจากค่าอ้างอิงที่ต่างเล็กน้อย) บนดอยอินทนนท์ ค่าที่อ่านได้ต่ำกว่าที่ระดับน้ำทะเลราว 790 มิลลิแกล จึงต้อง **บวก** ค่าแก้อากาศอิสระนี้กลับเข้าไป เนื่องจากแหล่งแร่และแอ่งตะกอนให้ค่าผิดปกติเพียง 1–10 มิลลิแกล ความคลาดเคลื่อนของความสูง 3 ม. (≈ 1 มิลลิแกล) ก็มีผลแล้ว การสำรวจแรงโน้มถ่วงจึงต้องใช้ความสูงจาก GPS ที่แม่นยำ

**ข้อผิดพลาดที่พบบ่อย**

- ลืมเครื่องหมายลบ แรงโน้มถ่วงลดลงตามความสูง ค่าที่อ่านได้จึงต้องแก้ให้สูงขึ้น
- ใช้หน่วย ม./วินาที² กับมิลลิแกลปนกัน 1 มิลลิแกล $=10^{-5}$ ม./วินาที² ไม่ใช่ $10^{-3}$
- ใช้ $R$ หน่วยกิโลเมตรร่วมกับ $GM$ หน่วย ม.³/วินาที² ความยาวทุกตัวต้องเป็นเมตร

**ลองทำดู** สถานีวัดความโน้มถ่วงบนดอยอินทนนท์ (2565 m) อ่านค่าได้ 978,xxx mGal ต้อง **บวก** ค่าปรับแก้แบบอากาศอิสระเท่าใด (หน่วย mGal)

## ตัวอย่างที่ 21 ⭐⭐: แผ่นดินไหวอยู่ไกลแค่ไหน (ผลต่างเชิงอนุพันธ์และการประมาณความคลาดเคลื่อน (differentials and error estimates))

**บริบท** คลื่น P ($v_p = 6.0$ km/s) มาถึงก่อนคลื่น S ($v_s = 3.5$ km/s) จากเวลา S–P $\Delta t$ เครื่องวัดแผ่นดินไหวเพียงเครื่องเดียวให้ระยะทาง:

$$d = \Delta t\,\frac{v_pv_s}{v_p - v_s}.$$

สถานีบันทึกได้ $\Delta t = 12$ s

**แคลคูลัส** ผลต่างเชิงอนุพันธ์: $dd = \dfrac{\partial d}{\partial\Delta t}\,d(\Delta t)$ และ $dd = \dfrac{\partial d}{\partial v_s}\,dv_s$ ใช้ประมาณความคลาดเคลื่อนของระยะทาง

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ระยะทางถึงศูนย์กลางแผ่นดินไหวขึ้นกับเวลา S–P และความเร็วคลื่น ซึ่งไม่เคยรู้ค่าได้แน่นอน **ผลต่างเชิงอนุพันธ์ (differential)** ประมาณว่าความคลาดเคลื่อนของข้อมูลแต่ละตัวทำให้ระยะทางที่คำนวณได้เปลี่ยนไปเท่าใด โดยไม่ต้องคำนวณใหม่ทั้งหมด

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **หาสูตร** คลื่น P มาถึงที่เวลา $d/v_p$ และคลื่น S ที่ $d/v_s$ ดังนั้น $\Delta t=d\left(\frac1{v_s}-\frac1{v_p}\right)=d\,\frac{v_p-v_s}{v_pv_s}$ ได้ $d=\Delta t\,\dfrac{v_pv_s}{v_p-v_s}$
2. **คำนวณระยะทาง** $\dfrac{v_pv_s}{v_p-v_s}=\dfrac{6.0\times3.5}{2.5}=8.4$ km/s ดังนั้น $d=12\times8.4=100.8$ km
3. **ความคลาดเคลื่อนของเวลา** $\partial d/\partial\Delta t=8.4$ km/s ความคลาดเคลื่อนในการอ่านเวลา $d(\Delta t)=0.2$ s ให้ $dd=8.4\times0.2=1.68$ km
4. **ความคลาดเคลื่อนของความเร็ว** ใช้กฎผลหาร $$\frac{\partial d}{\partial v_s}=\Delta t\,v_p\,\frac{(v_p-v_s)+v_s}{(v_p-v_s)^2}=\frac{\Delta t\,v_p^2}{(v_p-v_s)^2}=\frac{12\times36}{6.25}=69.12\ \text{s}$$
5. **เปรียบเทียบ** ความคลาดเคลื่อน $dv_s=0.1$ km/s ให้ $dd=69.12\times0.1\approx6.91$ km ประมาณสี่เท่าของผลจากความคลาดเคลื่อนของเวลา

```python
Dt, vp, vs = sp.symbols('Delta_t v_p v_s', positive=True)
d = Dt*vp*vs/(vp - vs)
vals = {Dt: 12, vp: 6.0, vs: 3.5}
print(f"distance d = {float(d.subs(vals)):.1f} km")
dd_t = sp.diff(d, Dt); dd_s = sp.simplify(sp.diff(d, vs))
show(r"\frac{\partial d}{\partial v_s}", dd_s)
print(f"∂d/∂Δt = {float(dd_t.subs(vals)):.2f} km/s -> timing error 0.2 s gives dd = {float(dd_t.subs(vals))*0.2:.2f} km")
print(f"∂d/∂v_s = {float(dd_s.subs(vals)):.2f} s -> velocity error 0.1 km/s gives dd = {float(dd_s.subs(vals))*0.1:.2f} km")
```

```python
DT = np.linspace(0, 30, 200); dfun = sp.lambdify(Dt, d.subs({vp: 6.0, vs: 3.5}))
fig, ax = plt.subplots()
ax.plot(DT, dfun(DT), lw=2, label="d(Δt) for v_s = 3.5 km/s")
for vv, cc in [(3.4, "C1"), (3.6, "C2")]:
    ax.plot(DT, sp.lambdify(Dt, d.subs({vp: 6.0, vs: vv}))(DT), "--", c=cc, lw=1, label=f"v_s = {vv} km/s")
ax.plot(12, dfun(12), "ko")
ax.set(xlabel="S − P time (s)", ylabel="distance (km)", title="An uncertain v_s matters more than a 0.2 s timing error"); ax.legend(); plt.show()
```

**การตีความ** แผ่นดินไหวอยู่ห่างออกไปประมาณ **100.8 km** ความคลาดเคลื่อนในการอ่านเวลา 0.2 s ทำให้ระยะเปลี่ยนเพียง 1.7 km แต่ความไม่แน่นอน 0.1 km/s ของ $v_s$ ทำให้ระยะเปลี่ยนประมาณ 6.9 km
ดังนั้นแบบจำลองความเร็วของเปลือกโลกที่ดีขึ้นจึงคุ้มค่ากว่านาฬิกาที่แม่นยำขึ้น จากนั้นนำระยะทางจากสามสถานี (วงกลม) มารวมกันเพื่อหาตำแหน่งจุดเหนือศูนย์เกิดแผ่นดินไหว (epicentre)

**ความหมายของคำตอบ** แผ่นดินไหวอยู่ห่างจากสถานีประมาณ 101 km ระยะทางไวต่อความเร็วคลื่น S มากกว่านาฬิกามาก เพราะ $v_p-v_s$ เป็นผลต่างเล็ก ๆ ในตัวส่วน แบบจำลองความเร็วของเปลือกโลกที่ดีขึ้นจึงมีค่ากว่าการจับเวลาที่แม่นขึ้น จากนั้นนำระยะทางจากสามสถานีมารวมกันเพื่อหาตำแหน่งศูนย์กลาง

**ข้อผิดพลาดที่พบบ่อย**

- มองว่า $v_s$ อยู่เฉพาะในตัวเศษ ที่จริงอยู่ในตัวส่วนด้วย จึงต้องใช้กฎผลหาร
- บวกค่าประมาณความคลาดเคลื่อนทั้งสองเสมือนว่าเกิดพร้อมกันเสมอ ทั้งสองเป็นแหล่งที่แยกกัน ให้รวมเฉพาะเมื่อเกิดขึ้นพร้อมกัน
- ใช้หน่วยวินาทีปนกับนาทีสำหรับ $\Delta t$ หรือ km ปนกับ m สำหรับความเร็ว

**ลองทำดู** สำหรับแผ่นดินไหวแม่ลาว (เชียงราย) ปี 2014 สถานีที่อยู่ห่าง 60 km จะวัดเวลา S–P ได้เท่าใด

- (d) ทุกจุดในอัตราเท่ากัน

<details><summary><b>เฉลย C3</b></summary>

**(c)** ที่ $z''<0$ (นูน ใกล้สันเขา) $\partial z/\partial t<0$ ตรงที่ $z''<0$ พอดี สันเขาที่นูนถูกกัดเซาะ และแอ่งที่เว้าถูกถม ความชัน ($z'$) อย่างเดียวไม่ได้เป็นตัวกำหนด

</details>

**C4** (ตัวอย่างที่ 18) ระหว่างแม่น้ำสองสาย ระดับน้ำใต้ดินเป็นไปตาม $h^2=400-300\,\dfrac{x}{1000}$ ($h$, $x$ เป็นเมตร) จงใช้การหาอนุพันธ์โดยปริยายหา $\dfrac{dh}{dx}$ ที่ $x=500$ ม.

</details>

**C5** (ตัวอย่างที่ 19) เส้นเมริเดียนของโลกคือวงรี $\dfrac{x^2}{a^2}+\dfrac{y^2}{b^2}=1$ จงหา $\dfrac{dy}{dx}$

- (a) $-\dfrac{b^2x}{a^2y}$
- (b) $\dfrac{b^2y}{a^2x}$
- (c) $-\dfrac{x}{y}$
- (d) $-\dfrac{a^2x}{b^2y}$

# ส่วน D: การประยุกต์อนุพันธ์ (Applications of Derivatives)

⭐⭐ ปานกลาง · ตัวอย่างที่ 22–28 · สัปดาห์ที่ 7–8 ในประมวลรายวิชา

## ตัวอย่างที่ 22 ⭐⭐: กรวยภูเขาไฟสกอเรียโตเร็วเพียงใด (อัตราสัมพัทธ์)

**บริบท** กรวยภูเขาไฟสกอเรีย (cinder cone) เติบโตขึ้นเมื่อเศษภูเขาไฟกองทับกันที่ **มุมทรงตัว (angle of repose)** $\varphi \approx 33^\circ$ ดังนั้น $h = r\tan\varphi$ ตลอดเวลา
การปะทุเพิ่มเทฟรา (tephra) ด้วยอัตรา $\dfrac{dV}{dt} = 10\ \text{m}^3/\text{s}$

**คำถาม** ความสูงของกรวยเพิ่มขึ้นเร็วเพียงใดเมื่อสูง 50 m และเมื่อสูง 200 m

**แคลคูลัส** $V = \tfrac13\pi r^2 h = \dfrac{\pi h^3}{3\tan^2\varphi}$ หาอนุพันธ์ทั้งสองข้างเทียบกับ $t$ (กฎลูกโซ่): $\dfrac{dV}{dt} = \dfrac{\pi h^2}{\tan^2\varphi}\dfrac{dh}{dt}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ในโจทย์ **อัตราการเปลี่ยนแปลงสัมพัทธ์ (related rates)** ปริมาณสองตัวเปลี่ยนไปพร้อมกันและเชื่อมกันด้วยสมการ ในที่นี้ปริมาตรกับความสูงของกรวยเชื่อมกันด้วยเรขาคณิต และทั้งสองเปลี่ยนตามเวลา การหาอนุพันธ์ของสูตรปริมาตรเทียบกับ $t$ (หลักเกณฑ์ลูกโซ่) เชื่อมอัตราการปะทุที่รู้ค่า $dV/dt$ เข้ากับอัตราการสูงขึ้นที่ไม่รู้ค่า $dh/dt$

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. เริ่มจากเรขาคณิต: เทฟราวางตัวที่มุมพัก $\varphi$ จึงได้ $\tan\varphi=h/r$ และ $r=h/\tan\varphi$ แทนลงใน $V=\tfrac13\pi r^2h$ เพื่อกำจัด $r$: $$V=\frac{\pi h^3}{3\tan^2\varphi}$$
2. หาอนุพันธ์ทั้งสองข้างเทียบกับเวลา เนื่องจาก $h$ ขึ้นกับ $t$ หลักเกณฑ์ลูกโซ่จึงให้ $$\frac{dV}{dt}=\frac{\pi h^2}{\tan^2\varphi}\frac{dh}{dt}\ \Rightarrow\ \frac{dh}{dt}=\frac{\tan^2\varphi}{\pi h^2}\frac{dV}{dt}$$
3. แทนค่าเริ่มต้น $\varphi=33^\circ$ ($\tan^2\varphi=0.4217$) $h=50$ ม. และ $dV/dt=10$ ม.³/วินาที: $$\frac{dh}{dt}=\frac{0.4217}{\pi(2500)}\times10=5.37\times10^{-4}\ \text{ม./วินาที}$$
4. แปลงเป็นหน่วยที่ใช้ง่าย: $5.37\times10^{-4}\times86{,}400=46.4$ ม./วัน ที่ $h=200$ ม. อัตราลดลง 16 เท่าเหลือ $2.90$ ม./วัน เพราะ $dh/dt\propto1/h^2$
5. ขนาดของกรวยตอนนี้: $r=50/\tan33^\circ=77.0$ ม. และ $V=\pi(50)^3/(3\times0.4217)=0.310$ ล้าน ม.³ ซึ่งใช้เวลาสร้าง $V/(dV/dt)=31{,}039$ วินาที $\approx0.36$ วัน ที่อัตรานี้

```python
tvar = sp.symbols('t'); hfun = sp.Function('h')(tvar); phi = sp.rad(33)
V = sp.pi*hfun**3/(3*sp.tan(phi)**2)
dVdt = sp.diff(V, tvar)
show(r"\frac{dV}{dt}", dVdt)
dhdt = sp.solve(sp.Eq(dVdt, 10), sp.diff(hfun, tvar))[0]
for H in (50, 200):
    rate = float(dhdt.subs(hfun, H))
    print(f"h = {H:>3} m: dh/dt = {rate:.2e} m/s = {rate*86400:.2f} m/day")
```

```python
Hs = np.linspace(5, 300, 300)
rate = 10*np.tan(np.radians(33))**2/(np.pi*Hs**2)*86400
fig, ax = plt.subplots()
ax.plot(Hs, rate, lw=2); ax.set_yscale("log")
ax.set(xlabel="cone height h (m)", ylabel="dh/dt (m/day)", title="Same eruption rate, slower growth: dh/dt ∝ 1/h²")
plt.show()
```

**การตีความ** การเติบโตช้าลงตาม $1/h^2$ เพราะความสูงที่เพิ่มขึ้นแต่ละเมตรต้องใช้เปลือกเทฟราที่ใหญ่ขึ้นมาก
ภูเขาไฟปาริกุติน (Parícutin, เม็กซิโก, 1943) สูงขึ้นประมาณ 50 m ในวันแรก แต่ใช้เวลาหลายปีกว่าจะสูงถึง 400 m ซึ่งสอดคล้องกับพฤติกรรมนี้

**ความหมายของคำตอบ** เมื่อสูง 50 ม. กรวยสูงขึ้นราว **46 ม. ต่อวัน** แต่เมื่อสูง 200 ม. สูงขึ้นเพียงราว 3 ม. ต่อวัน แม้อัตราการปะทุจะเท่าเดิม ความสูงที่เพิ่มขึ้นแต่ละเมตรต้องใช้เปลือกเทฟราที่ใหญ่กว่ามาก (พื้นที่ผิวกรวยเพิ่มตาม $h^2$) ซึ่งตรงกับภูเขาไฟปาริกุติน (Parícutin, เม็กซิโก ค.ศ. 1943) ที่สูงขึ้นราว 50 ม. ในวันแรก แต่ใช้เวลาหลายปีกว่าจะสูงถึง 400 ม.

**ข้อผิดพลาดที่พบบ่อย**

- แทน $h=50$ ก่อนหาอนุพันธ์ จะทำให้ $V$ เป็นค่าคงตัวและอนุพันธ์เป็น 0 ต้องแทนตัวเลขหลังจากหาอนุพันธ์แล้วเท่านั้น
- เก็บทั้ง $r$ และ $h$ เป็นตัวแปรโดยไม่ใช้ $r=h/\tan\varphi$ ทำให้เหลือตัวไม่ทราบค่า $dr/dt$
- ลืมแปลง ม./วินาที เป็น ม./วัน (×86,400) หรือใส่ $\varphi$ เป็นองศาในเครื่องคิดเลขที่ตั้งเป็นเรเดียน

**ลองทำดู** ถ้ากรวยประกอบด้วยเถ้าละเอียดกว่าซึ่งมี $\varphi = 28^\circ$ ความสูงจะเพิ่มเร็วขึ้นหรือช้าลง อธิบายโดยใช้สูตร

## ตัวอย่างที่ 23 ⭐⭐: กระเปาะแมกมาที่กำลังพองตัว (อัตราสัมพัทธ์)

**บริบท** แมกมาไหลเข้าสู่กระเปาะทรงกลมรัศมี $r = 2$ km ด้วยอัตรา $\dfrac{dV}{dt} = 10^6$ m³/yr **แบบจำลองโมกิ (Mogi model)** ให้การยกตัวของผิวดินเหนือกระเปาะที่ความลึก $d = 5$ km โดยตรงเป็น

$$u_z = \frac{3\,\Delta V}{4\pi d^2}\quad(\text{Poisson ratio } 0.25).$$

**แคลคูลัส** $V = \tfrac43\pi r^3 \Rightarrow \dfrac{dV}{dt} = 4\pi r^2\dfrac{dr}{dt}$ ในทำนองเดียวกัน $\dfrac{du_z}{dt} = \dfrac{3}{4\pi d^2}\dfrac{dV}{dt}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ปริมาตร รัศมี และการยกตัวของผิวดินล้วนเปลี่ยนตามเวลา และสัมพันธ์กันด้วยสูตร **อัตราสัมพัทธ์ (related rates)** หาอนุพันธ์ของสูตรเหล่านั้นเทียบกับเวลา อัตราการไหลเข้าของแมกมาที่ทราบจึงให้อัตราที่สังเกตได้จริง คือผนังโพรงที่ขยายและพื้นดินที่ยกตัว

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **หาอนุพันธ์ของปริมาตร** $V=\tfrac43\pi r^3$ โดย $r=r(t)$ ใช้หลักเกณฑ์ลูกโซ่ได้ $\dfrac{dV}{dt}=4\pi r^2\dfrac{dr}{dt}$
2. **แก้หา $dr/dt$** เมื่อ $dV/dt=10^6$ m³/yr และ $r=2000$ m $$\frac{dr}{dt}=\frac{10^6}{4\pi(2000)^2}=\frac{1}{16\pi}\approx0.0199\ \text{m/yr}\approx19.9\ \text{mm/yr}$$
3. **หาอนุพันธ์ของสูตรโมกิ (Mogi)** $u_z=\dfrac{3\,\Delta V}{4\pi d^2}$ โดยความลึก $d$ คงที่ ดังนั้น $\dfrac{du_z}{dt}=\dfrac{3}{4\pi d^2}\dfrac{dV}{dt}$
4. **คำนวณอัตราการยกตัว** เมื่อ $d=5000$ m $$\frac{du_z}{dt}=\frac{3\times10^6}{4\pi(5000)^2}=\frac{3}{100\pi}\approx0.00955\ \text{m/yr}=9.55\ \text{mm/yr}$$
5. **มองออกด้านข้าง** ห่างจากจุดกึ่งกลาง อัตราการยกตัวแปรผันตาม $d/(x^2+d^2)^{3/2}$ ที่ $x=d=5$ km ลดลงเหลือ $2^{-3/2}\approx35\%$ ของค่าสูงสุด หรือราว 3.4 mm/yr

```python
t = sp.symbols('t'); r = sp.Function('r')(t)
V = sp.Rational(4, 3)*sp.pi*r**3
show(r"\frac{dV}{dt}", sp.diff(V, t))
Qv, r0, dep = 1e6, 2000.0, 5000.0
drdt = Qv/(4*np.pi*r0**2)
duz = 3*Qv/(4*np.pi*dep**2)
print(f"dr/dt = {drdt:.4f} m/yr = {drdt*1000:.1f} mm/yr (chamber wall)")
print(f"duz/dt = {duz*1000:.2f} mm/yr (ground surface above the chamber)")
```

```python
xx = np.linspace(-20, 20, 400)*1000
uz = 3*Qv/(4*np.pi)*dep/(xx**2 + dep**2)**1.5
fig, ax = plt.subplots()
ax.plot(xx/1000, uz*1000, lw=2)
ax.set(xlabel="distance from the point above the chamber (km)", ylabel="uplift rate (mm/yr)", title="Mogi uplift pattern from an inflating chamber"); plt.show()
```

**การตีความ** ผนังกระเปาะขยายออกเพียงประมาณ **2 cm/yr** แต่พื้นดินด้านบนยกตัวขึ้นประมาณ **9.5 mm/yr** ซึ่ง GPS และดาวเทียม InSAR วัดได้ไม่ยาก
หอสังเกตการณ์ภูเขาไฟใช้ความสัมพันธ์นี้ย้อนกลับ คือจากอัตราการยกตัวที่วัดได้ ประมาณว่าแมกมากำลังสะสมเร็วเพียงใด และเตือนภัยการปะทุที่อาจเกิดขึ้น

**ความหมายของคำตอบ** ผนังโพรงขยายออกเพียงประมาณ 2 cm ต่อปี แต่พื้นดินด้านบนยกตัวราว 9.5 mm ต่อปี ซึ่ง GPS และดาวเทียม InSAR วัดได้ หอสังเกตการณ์ภูเขาไฟคำนวณย้อนกลับ คือประมาณอัตราการสะสมของแมกมาจากอัตราการยกตัวที่วัดได้ สังเกตว่าอัตราการยกตัวขึ้นกับความลึก ไม่ขึ้นกับรัศมีของโพรง

**ข้อผิดพลาดที่พบบ่อย**

- เขียน $dV/dt=\tfrac43\pi\cdot3r^2$ แล้วลืมตัวประกอบ $dr/dt$ จากหลักเกณฑ์ลูกโซ่
- แทนตัวเลขก่อนหาอนุพันธ์ $r$ ต้องเป็นตัวแปรจนกว่าจะหาอนุพันธ์เสร็จ
- ใช้หน่วย km ปนกับ m ถ้าใช้ $r$ เป็น km แต่ $dV/dt$ เป็น m³/yr คำตอบจะผิดไป $10^6$ เท่า

**ลองทำดู** หากกระเปาะเดียวกันนี้อยู่ลึก 10 km ผิวดินจะยกตัวเร็วเท่าใด

## ตัวอย่างที่ 24 ⭐⭐: หลักของแฟร์มา: คลื่นไหวสะเทือนเลือกเส้นทางที่เร็วที่สุด (การหาค่าเหมาะที่สุด → กฎของสเนลล์)

**บริบท** คลื่นไหวสะเทือนเดินทางจากแหล่งกำเนิด $A$ ในชั้นที่คลื่นเคลื่อนที่ช้า ($v_1 = 3$ km/s) ซึ่งอยู่เหนือรอยต่อ 2 km ไปยังเครื่องรับ $B$ ในชั้นที่คลื่นเคลื่อนที่เร็ว ($v_2 = 5$ km/s)
ซึ่งอยู่ใต้รอยต่อ 3 km และห่างในแนวราบ 6 km คลื่นผ่านรอยต่อที่ตำแหน่ง $x$ เวลาเดินทางรวมคือ

$$T(x) = \frac{\sqrt{2^2 + x^2}}{v_1} + \frac{\sqrt{3^2 + (6-x)^2}}{v_2}.$$

**คำถาม** จงหาค่าต่ำสุดของ $T(x)$ และแสดงว่าค่าต่ำสุดสอดคล้องกับ **กฎของสเนลล์ (Snell's law)** $\dfrac{\sin\theta_1}{v_1} = \dfrac{\sin\theta_2}{v_2}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** คลื่นไหวสะเทือนไม่ได้เดินทางเป็นเส้นตรงเมื่อผ่านสองชั้นหิน แต่เลือกเส้นทางที่ใช้ **เวลาน้อยที่สุด** (หลักของแฟร์มา (Fermat's principle)) เมื่อเขียนเวลาเดินทางเป็นฟังก์ชันของจุดตัด $x$ ปัญหานี้จึงกลายเป็น **การหาค่าสูงสุดต่ำสุด (optimisation)** คือหาจุดที่ $T'(x)=0$ แล้วตรวจว่าเป็นค่าต่ำสุด ซึ่งเงื่อนไข $T'(x)=0$ นี้ก็คือกฎของสเนลล์ (Snell's law) นั่นเอง

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ตั้งฟังก์ชัน** แหล่งกำเนิดอยู่เหนือรอยต่อ $a=2$ km ตัวรับอยู่ใต้รอยต่อ $b=3$ km และห่างออกไปในแนวราบ $d=6$ km จากทฤษฎีบทพีทาโกรัส ความยาวของสองช่วงคือ $\sqrt{a^2+x^2}$ และ $\sqrt{b^2+(d-x)^2}$ และเวลา = ระยะทาง / อัตราเร็ว จึงได้ $$T(x)=\frac{\sqrt{4+x^2}}{3}+\frac{\sqrt{9+(6-x)^2}}{5}.$$
2. **หาอนุพันธ์ด้วยหลักเกณฑ์ลูกโซ่ (chain rule)** $\dfrac{d}{dx}\sqrt{4+x^2}=\dfrac{x}{\sqrt{4+x^2}}$ และ $\dfrac{d}{dx}\sqrt{9+(6-x)^2}=\dfrac{-(6-x)}{\sqrt{9+(6-x)^2}}$ (อนุพันธ์ภายในของ $6-x$ คือ $-1$) ดังนั้น $$T'(x)=\frac{x}{3\sqrt{4+x^2}}-\frac{6-x}{5\sqrt{9+(6-x)^2}}.$$
3. **มองให้เห็นค่าไซน์** จากรูป $\dfrac{x}{\sqrt{4+x^2}}=\sin\theta_1$ และ $\dfrac{6-x}{\sqrt{9+(6-x)^2}}=\sin\theta_2$ โดยวัดมุมจากแนวดิ่ง การตั้ง $T'(x)=0$ จึงหมายถึง $\dfrac{\sin\theta_1}{v_1}=\dfrac{\sin\theta_2}{v_2}$ กฎของสเนลล์จึงได้มาจากอนุพันธ์โดยตรง
4. **แก้สมการเชิงตัวเลข** สมการ $T'(x)=0$ ไม่มีคำตอบรูปปิดที่สวยงาม จึงแก้ด้วยวิธีเชิงตัวเลข (วิธีของนิวตันหรือการแบ่งครึ่งช่วง) ได้ $x^*\approx 1.1836$ km จากนั้น $\sin\theta_1=1.1836/\sqrt{4+1.1836^2}=0.5093$ จึงได้ $\theta_1\approx30.62^\circ$ และ $\theta_2\approx58.08^\circ$ ทั้งสองอัตราส่วนมีค่าเท่ากันคือ $0.169762$ s/km
5. **ตรวจว่าเป็นค่าต่ำสุด** $T''(x^*)\approx0.1161>0$ กราฟจึงเว้าขึ้นที่จุดนั้น และ $x^*$ เป็นจุดต่ำสุด (การทดสอบด้วยอนุพันธ์อันดับสอง) เปรียบเทียบ: $T(x^*)=1.9095$ s ขณะที่จุดตัดทดลอง $x=3.5$ km ให้ $2.1247$ s และที่ปลายช่วงได้ $T(0)=2.008$ s และ $T(6)=2.708$ s

```python
x = sp.symbols('x', real=True)
v1, v2, a, b, d = 3, 5, 2, 3, 6
T = sp.sqrt(a**2 + x**2)/v1 + sp.sqrt(b**2 + (d - x)**2)/v2
dT = sp.diff(T, x)
show("T'(x)", dT)
x_star = sp.nsolve(dT, x, 2)                        # T'(x)=0
d2 = sp.diff(T, x, 2).subs(x, x_star)
print(f"critical point x* = {float(x_star):.4f} km, T'' = {float(d2):.4f} > 0  -> minimum")
print(f"minimum travel time = {float(T.subs(x, x_star)):.4f} s")

sin1 = x_star/sp.sqrt(a**2 + x_star**2); sin2 = (d - x_star)/sp.sqrt(b**2 + (d - x_star)**2)
print(f"sinθ1/v1 = {float(sin1/v1):.6f}   sinθ2/v2 = {float(sin2/v2):.6f}   -> Snell's law holds")
print(f"θ1 = {np.degrees(np.arcsin(float(sin1))):.2f}°, θ2 = {np.degrees(np.arcsin(float(sin2))):.2f}°; "
      f"critical angle arcsin(v1/v2) = {np.degrees(np.arcsin(v1/v2)):.2f}°")
```

```python
xx = np.linspace(-1, 7, 300); Tf = sp.lambdify(x, T)
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4))
a1.plot(xx, Tf(xx), lw=2); a1.plot(float(x_star), Tf(float(x_star)), "ro")
a1.set(xlabel="crossing point x (km)", ylabel="travel time T (s)", title="T(x) has a single minimum")
a2.axhspan(0, 2.5, color="C1", alpha=.15); a2.axhspan(-3.5, 0, color="C0", alpha=.15)
a2.text(4.2, 1.8, "v₁ = 3 km/s"); a2.text(0.2, -2.8, "v₂ = 5 km/s")
for xc in [0.5, 1.5, 3.5, float(x_star)]:
    style = dict(c="r", lw=2.5) if xc == float(x_star) else dict(c="gray", lw=1, ls="--")
    a2.plot([0, xc, d], [a, 0, -b], **style)
a2.plot([0, d], [a, -b], "k*", ms=12); a2.set(xlabel="x (km)", ylabel="z (km)", title="fastest path bends at the interface")
plt.tight_layout(); plt.show()
```

**การตีความ** การให้ $T'(x)=0$ ได้ $\dfrac{x}{v_1\sqrt{a^2+x^2}} = \dfrac{d-x}{v_2\sqrt{b^2+(d-x)^2}}$ พอดี ซึ่งก็คือกฎของสเนลล์
การหักเหของรังสีคลื่นเป็นพื้นฐานของการสำรวจคลื่นไหวสะเทือนแบบหักเห และเป็นวิธีที่นักแผ่นดินไหววิทยาค้นพบโมโฮ (Moho) ซึ่งเป็นรอยต่อระหว่างเปลือกโลกกับเนื้อโลก

**ความหมายของคำตอบ** รังสีที่เร็วที่สุดจะหักเห **ออกจาก** แนวดิ่งเมื่อเข้าสู่ชั้นที่เร็วกว่า ($\theta_2>\theta_1$) เวลาต่ำสุดประมาณ 1.91 s คือสิ่งที่เครื่องวัดแผ่นดินไหวบันทึกเป็นคลื่นที่มาถึงก่อน การสำรวจแบบหักเหคลื่นไหวสะเทือน (seismic refraction) ใช้การหักเหนี้วัดความเร็วและความลึกของชั้นหิน และเป็นวิธีที่ค้นพบแนวไม่ต่อเนื่องโมโฮ (Moho) ตรวจสอบความสมเหตุสมผล: จุดตัดอยู่ใกล้แหล่งกำเนิด คลื่นจึงเดินทางในชั้นที่ช้าเป็นระยะสั้น ซึ่งสมเหตุสมผล

**ข้อผิดพลาดที่พบบ่อย**

- ลืมอนุพันธ์ภายใน $-1$ ของ $(6-x)$ ทำให้พจน์ที่สองของ $T'(x)$ มีเครื่องหมายผิด
- วัดมุมจากรอยต่อชั้นหินแทนที่จะวัดจากแนวดิ่ง (แนวตั้งฉาก) กฎของสเนลล์ใช้มุมที่วัดจากแนวตั้งฉาก
- หยุดแค่ $T'(x)=0$ โดยไม่ตรวจว่าเป็นค่าต่ำสุด (ด้วยอนุพันธ์อันดับสองหรือเทียบกับค่าที่ปลายช่วง)

**ลองทำดู** สลับความเร็ว ($v_1 = 5$, $v_2 = 3$) รังสีคลื่นหักเหเข้าหาหรือออกจากแนวดิ่ง

## ตัวอย่างที่ 25 ⭐⭐: ทำไมดวงอาทิตย์เปล่งแสงช่วงที่ตามองเห็น แต่โลกเปล่งรังสีอินฟราเรด (การหาค่าเหมาะที่สุด → กฎของวีน)

**บริบท** วัตถุที่อุณหภูมิ $T$ เปล่งรังสีตาม **กฎของพลังค์ (Planck's law)**

$$B(\lambda,T) = \frac{2hc^2}{\lambda^5}\,\frac{1}{e^{hc/(\lambda k_BT)}-1}.$$

ที่ความยาวคลื่น $\lambda_{\max}$ ใดที่การเปล่งรังสีมีค่ามากที่สุด

**แคลคูลัส** ให้ $\partial B/\partial\lambda = 0$ เมื่อแทน $x = hc/(\lambda k_BT)$ สมการจะลดรูปเป็น $5(1-e^{-x}) = x$ ซึ่งมีรากคือ $x\approx4.965$ และได้ **กฎของวีน (Wien's law)** $\lambda_{\max} = \dfrac{hc}{4.965\,k_B}\dfrac1T \approx \dfrac{2898\ \mu\text{m·K}}{T}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ความยาวคลื่นที่แผ่รังสีมากที่สุดคือจุดสูงสุดของเส้นโค้งพลังค์ เราจึงตั้ง $\partial B/\partial\lambda=0$ **การหาค่าเหมาะที่สุด (optimisation)** นี้ลดรูปเป็นสมการ $5(1-e^{-x})=x$ ซึ่งต้องแก้เชิงตัวเลข และให้กฎของวีน (Wien's law) $\lambda_{\max}=b/T$

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ทำให้ง่าย** เขียน $B=C\lambda^{-5}\left(e^{a/\lambda}-1\right)^{-1}$ โดย $C=2hc^2$ และ $a=hc/(k_BT)$
2. **หาอนุพันธ์ (กฎผลคูณ + หลักเกณฑ์ลูกโซ่)** $$\frac{\partial B}{\partial\lambda}=C\left[-5\lambda^{-6}\left(e^{a/\lambda}-1\right)^{-1}+\lambda^{-5}\left(e^{a/\lambda}-1\right)^{-2}e^{a/\lambda}\frac{a}{\lambda^2}\right]$$
3. **ตั้งให้เท่ากับศูนย์** คูณด้วย $\lambda^6(e^{a/\lambda}-1)^2/C$ และให้ $x=a/\lambda$ ได้ $5(e^x-1)=xe^x$ คือ $5(1-e^{-x})=x$
4. **แก้เชิงตัวเลข (วิธีของนิวตัน)** ให้ $F(x)=5-5e^{-x}-x$, $F'(x)=5e^{-x}-1$ เริ่มที่ $x_0=5$: $F(5)=-0.0337$, $F'(5)=-0.966$ ดังนั้น $x_1=5-0.0349=4.9651$ รากคือ $x\approx4.96511$
5. **หาค่าคงตัวของวีน** $\lambda_{\max}=\dfrac{hc}{4.965\,k_BT}=\dfrac bT$ โดย $b=\dfrac{6.626\times10^{-34}\times2.998\times10^{8}}{4.965\times1.381\times10^{-23}}\approx2.898\times10^{-3}$ m·K $=2897.8$ μm·K
6. **นำไปใช้** ดวงอาทิตย์ ($T=5778$ K): $\lambda_{\max}=2897.8/5778\approx0.50$ μm โลก ($288$ K): $\approx10.06$ μm ลาวา ($1373$ K): $\approx2.11$ μm

```python
lam, Tk, xs = sp.symbols('lambda T x', positive=True)
h_, c_, kB = 6.62607015e-34, 2.99792458e8, 1.380649e-23
eq = sp.Eq(5*(1 - sp.exp(-xs)), xs)
x_root = sp.nsolve(5*(1 - sp.exp(-xs)) - xs, xs, 5)
print(f"root of 5(1 − e^(−x)) = x : x = {float(x_root):.5f}")
b_wien = h_*c_/(float(x_root)*kB)
print(f"Wien constant b = hc/(x k_B) = {b_wien*1e6:.1f} μm·K")
for name, T in [("Sun (photosphere)", 5778), ("lava (1100 °C)", 1373), ("Earth surface", 288)]:
    print(f"{name:20s} T = {T:5d} K -> λ_max = {b_wien/T*1e6:6.2f} μm")
```

```python
B = lambda l, T: 2*h_*c_**2/l**5/np.expm1(h_*c_/(l*kB*T))
L = np.logspace(-7, -4, 600)
fig, ax = plt.subplots()
for T, col in [(5778, "C1"), (1373, "C3"), (288, "C0")]:
    y = B(L, T); ax.semilogx(L*1e6, y/y.max(), c=col, lw=2, label=f"T = {T} K")
    ax.axvline(b_wien/T*1e6, c=col, ls=":")
ax.axvspan(0.38, 0.75, color="gold", alpha=.2, label="visible light")
ax.set(xlabel="wavelength (μm)", ylabel="B / B_max", title="Planck curves and their maxima (Wien's law)"); ax.legend(); plt.show()
```

**การตีความ** ดวงอาทิตย์เปล่งรังสีสูงสุดที่ประมาณ **0.50 μm** (แสงสีเขียวที่ตามองเห็น) และโลกที่ประมาณ **10.1 μm** (อินฟราเรดความร้อน)
นี่คือเหตุผลที่แก๊สเรือนกระจกซึ่งดูดกลืนอินฟราเรดสามารถกักเก็บความร้อนที่โลกปล่อยออกไปได้ ขณะที่ยอมให้แสงอาทิตย์ผ่านเข้ามา เซนเซอร์ความร้อนบนดาวเทียม (เช่น Landsat แบนด์ 10 ที่ 10.9 μm) จึงถูกปรับให้ตรงกับช่วงสูงสุดของโลก ลาวาที่ 1373 K มีค่าสูงสุดที่ 2.1 μm ธารลาวาร้อนจึงเห็นได้เด่นชัดในช่วงอินฟราเรดคลื่นสั้น

**ความหมายของคำตอบ** การแผ่รังสีของดวงอาทิตย์สูงสุดที่แสงสีเขียวในช่วงที่ตามองเห็น ประมาณ 0.50 μm ส่วนโลกสูงสุดที่อินฟราเรดความร้อน ประมาณ 10 μm แก๊สเรือนกระจกดูดกลืนอินฟราเรด จึงปล่อยแสงอาทิตย์เข้ามาแต่กักความร้อนขาออกไว้ เซนเซอร์ความร้อนของดาวเทียม (เช่น Landsat แบนด์ 10) จึงตั้งไว้ใกล้ 10 μm และลาวาร้อนจะเด่นในอินฟราเรดคลื่นสั้นใกล้ 2 μm

**ข้อผิดพลาดที่พบบ่อย**

- รับรากชัด $x=0$ ซึ่งสอดคล้องกับ $\lambda\to\infty$ ไม่ใช่จุดสูงสุด รากที่มีความหมายทางฟิสิกส์คือ $x\approx4.965$
- ใช้ °C แทนเคลวินสำหรับ $T$ กฎของวีนต้องใช้อุณหภูมิสัมบูรณ์
- ลืมว่า $e^{a/\lambda}$ มีอนุพันธ์ภายใน $-a/\lambda^2$ ซึ่งเปลี่ยนเครื่องหมายของพจน์ที่สอง

**ลองทำดู** ไฟป่าลุกไหม้ที่ประมาณ 1000 K ดาวเทียมตรวจจับไฟควรวัดที่ความยาวคลื่นใด

## ตัวอย่างที่ 26 ⭐⭐: ระเบิดภูเขาไฟถูกพ่นออกมาเร็วเพียงใด (การหาค่าเหมาะที่สุด)

**บริบท** พบระเบิดภูเขาไฟ (volcanic bomb) ไกลจากปล่องได้ถึง $R = 2$ km หากไม่คิดแรงต้านอากาศ ระเบิดที่ถูกยิงด้วยอัตราเร็ว $v$ และมุม $\theta$ บนพื้นราบจะตกที่ระยะ

$$R(\theta) = \frac{v^2\sin 2\theta}{g}.$$

**คำถาม** (1) มุมใดให้ระยะไกลที่สุด (2) อัตราเร็วการพ่น **ต่ำสุด** ที่อธิบายระเบิดที่ระยะ 2 km ได้คือเท่าใด
(3) หากปล่องอยู่บนขอบปากปล่องสูงกว่าที่ราบ $H = 300$ m ระยะจะเป็น $R(\theta) = \dfrac{v\cos\theta}{g}\left(v\sin\theta+\sqrt{v^2\sin^2\theta+2gH}\right)$ ตอนนี้มุมใดดีที่สุด

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ระยะตกของบอมบ์ภูเขาไฟ (volcanic bomb) เป็นฟังก์ชันของมุมยิง $R(\theta)$ มุมที่ทำให้ไปไกลที่สุดคือจุดที่กราฟแบนราบที่ยอด เราจึงแก้สมการ $R'(\theta)=0$ (การหาค่าสูงสุด (optimisation)) และเมื่อมองกลับด้าน บอมบ์ที่ตกไกลที่สุดในแผนที่ภาคสนามจะบอก**อัตราเร็วต่ำสุด**ที่ภูเขาไฟต้องพ่นออกมา

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **พื้นราบ ($H=0$)** ถ้าไม่คิดแรงต้านอากาศ ระยะตกคือ $R(\theta)=\dfrac{v^2\sin2\theta}{g}$ หาอนุพันธ์ด้วยหลักเกณฑ์ลูกโซ่ (chain rule): $R'(\theta)=\dfrac{v^2}{g}\cdot\cos2\theta\cdot2=\dfrac{2v^2\cos2\theta}{g}$
2. **จุดวิกฤต** $R'(\theta)=0\Rightarrow\cos2\theta=0\Rightarrow2\theta=90^\circ\Rightarrow\theta=45^\circ$ (ในช่วง $0<\theta<90^\circ$) อนุพันธ์อันดับสอง $R''=-\dfrac{4v^2\sin2\theta}{g}<0$ ที่จุดนี้ จึงเป็นค่าสูงสุด และ $R_{\max}=\dfrac{v^2}{g}$
3. **อัตราเร็วพ่นต่ำสุด** บอมบ์ที่พบห่างออกไป 2 km ต้องมี $\dfrac{v^2}{g}\ge2000$ m ดังนั้น $v\ge\sqrt{9.81\times2000}=\sqrt{19620}\approx140.1$ m/s $\approx504$ km/h มุมอื่นต้องใช้อัตราเร็วมากกว่านี้ ค่านี้จึงเป็นขอบเขตล่าง
4. **ยิงจากขอบปล่อง สูง $H=300$ m** ตอนนี้ $R(\theta)=\dfrac{v\cos\theta}{g}\left(v\sin\theta+\sqrt{v^2\sin^2\theta+2gH}\right)$ เมื่อตั้ง $R'(\theta)=0$ แล้วจัดรูปจะได้ $\tan\theta^*=\dfrac{v}{\sqrt{v^2+2gH}}$
5. **แทนค่า $v=140$ m/s** $\sqrt{v^2+2gH}=\sqrt{19600+5886}=\sqrt{25486}\approx159.6$ ดังนั้น $\tan\theta^*\approx0.877$ และ $\theta^*\approx41.25^\circ$ ระยะตกที่มุมนี้คือ $R\approx2278$ m มากกว่า $2263$ m ที่มุม $45^\circ$ เล็กน้อย

```python
th, v, g_ = sp.symbols('theta v g', positive=True)
R = v**2*sp.sin(2*th)/g_
print("R'(θ) = 0 ->", sp.solve(sp.diff(R, th), th), "(θ = π/4 = 45°)")
gv = 9.81
v_min = np.sqrt(gv*2000)
print(f"minimum ejection speed for a 2 km bomb: √(gR) = {v_min:.1f} m/s = {v_min*3.6:.0f} km/h")

Hv, vv = 300, 140.0
Rr = lambda t: vv*np.cos(t)/gv*(vv*np.sin(t) + np.sqrt(vv**2*np.sin(t)**2 + 2*gv*Hv))
res = optimize.minimize_scalar(lambda t: -Rr(t), bounds=(0.1, 1.5), method="bounded")
print(f"from a {Hv} m rim at {vv} m/s: best angle = {np.degrees(res.x):.2f}°, range = {Rr(res.x):.0f} m "
      f"(vs {Rr(np.pi/4):.0f} m at 45°)")
print(f"closed form: tan θ* = v/√(v² + 2gH) -> θ* = {np.degrees(np.arctan(vv/np.sqrt(vv**2 + 2*gv*Hv))):.2f}°")
```

```python
TH = np.radians(np.linspace(5, 85, 300))
fig, ax = plt.subplots()
ax.plot(np.degrees(TH), vv**2*np.sin(2*TH)/gv, lw=2, label="flat ground")
ax.plot(np.degrees(TH), Rr(TH), "C3", lw=2, label=f"from {Hv} m crater rim")
ax.plot(np.degrees(res.x), Rr(res.x), "C3o"); ax.axvline(45, c="gray", ls=":")
ax.set(xlabel="launch angle θ (°)", ylabel="range (m)", title=f"Ballistic range at v = {vv:.0f} m/s"); ax.legend(); plt.show()
```

**การตีความ** บนพื้นราบมุมที่ดีที่สุดคือ 45° ดังนั้นระเบิดที่ระยะ 2 km ต้องมีอัตราเร็วอย่างน้อย **140 m/s** (ประมาณ 500 km/h) อัตราเร็วจริงสูงกว่านี้ เพราะแรงต้านอากาศทำให้ระยะสั้นลง
เมื่อยิงจากขอบปากปล่องสูง 300 m มุมเหมาะที่สุดลดลงเหลือประมาณ **41°** และระยะเพิ่มเป็นประมาณ 2.28 km นักภูเขาไฟวิทยาใช้การคำนวณนี้กำหนดเขตอันตรายจากวัตถุพุ่ง (ballistic hazard zone) รอบปากปล่อง

**ความหมายของคำตอบ** การยกจุดยิงให้สูงขึ้นทำให้มุมที่ดีที่สุดต่ำกว่า $45^\circ$ เพราะการยิงที่แบนกว่าใช้เวลาตกที่เพิ่มขึ้นไปกับการเคลื่อนที่ในแนวราบ ค่า 140 m/s เป็นอัตราเร็ว**ต่ำสุด** เพราะบอมบ์จริงถูกอากาศหน่วงไว้ อัตราเร็วพ่นจริงจึงสูงกว่านี้ แผนที่ภัยพิบัติใช้การคำนวณระยะไกลสุดแบบนี้กำหนดเขตห้ามเข้า

**ข้อผิดพลาดที่พบบ่อย**

- ลืมตัวคูณ 2 จากหลักเกณฑ์ลูกโซ่ใน $\dfrac{d}{d\theta}\sin2\theta=2\cos2\theta$ (ไม่เปลี่ยนมุมในข้อนี้ แต่ทำให้ $R''$ และคำตอบอื่นผิด)
- ใช้หน่วยองศาขณะหาอนุพันธ์ สูตรอย่าง $\dfrac{d}{dx}\sin x=\cos x$ ใช้ได้กับเรเดียนเท่านั้น ให้แปลงเป็นองศาตอนท้าย
- คิดว่ามุม $45^\circ$ ดีที่สุดเสมอ ซึ่งจริงเฉพาะเมื่อจุดยิงและจุดตกอยู่ระดับเดียวกัน

**ลองทำดู** จงแสดงโดยการหาอนุพันธ์ว่าบนพื้นราบ ระยะไกลสุดคือ $v^2/g$

## ตัวอย่างที่ 27 ⭐⭐: อนุกรมการสลายตัวกัมมันตรังสีเมื่อ $\lambda_1 = \lambda_2$ (กฎของโลปีตาล)

**บริบท** ในอนุกรมการสลายตัว ธาตุแม่ → ธาตุลูก → ธาตุหลาน (เช่น ในอนุกรมยูเรเนียม) ปริมาณธาตุลูกหาได้จากสมการเบตแมน (Bateman equation)

$$N_2(t) = N_0\,\frac{\lambda_1}{\lambda_2-\lambda_1}\left(e^{-\lambda_1 t}-e^{-\lambda_2 t}\right).$$

**คำถาม** จะเกิดอะไรขึ้นเมื่อ $\lambda_2 \to \lambda_1 = \lambda$ สูตรจะกลายเป็น $\tfrac00$ ซึ่งเป็นรูปแบบยังไม่กำหนด (indeterminate form)

**แคลคูลัส** กฎของโลปีตาล (L'Hôpital): หาอนุพันธ์ของตัวเศษและตัวส่วน **เทียบกับ $\lambda_2$**:

$$\lim_{\lambda_2\to\lambda}\frac{\lambda\left(e^{-\lambda t}-e^{-\lambda_2 t}\right)}{\lambda_2-\lambda} = \lim_{\lambda_2\to\lambda}\frac{\lambda\,t\,e^{-\lambda_2 t}}{1} = \lambda t\,e^{-\lambda t}.$$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** สมการเบตแมน (Bateman equation) ของไอโซโทปลูกมีตัวประกอบ $\dfrac{1}{\lambda_2-\lambda_1}$ เมื่อแทนค่าคงที่การสลายตัวที่เท่ากันจึงได้ **รูปแบบไม่กำหนด (indeterminate form)** $\tfrac00$ **หลักเกณฑ์โลปีตาล (L'Hôpital's rule)** บอกว่าลิมิตของผลหารรูป $\tfrac00$ เท่ากับลิมิตของผลหารของอนุพันธ์ ซึ่งให้สูตรที่มีค่าจำกัดสำหรับกรณีพิเศษนี้

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **สังเกตรูป $\tfrac00$** จาก $N_2(t)=N_0\dfrac{\lambda_1\left(e^{-\lambda_1t}-e^{-\lambda_2t}\right)}{\lambda_2-\lambda_1}$ เมื่อ $\lambda_2=\lambda_1=\lambda$ ตัวเศษเป็น $\lambda(e^{-\lambda t}-e^{-\lambda t})=0$ และตัวส่วนเป็น $0$ สูตรจึงหาค่าไม่ได้ แต่ปรากฏการณ์ทางฟิสิกส์ยังมีความหมาย จึงหาลิมิตเมื่อ $\lambda_2\to\lambda$
2. **หาอนุพันธ์ทั้งเศษและส่วนเทียบกับ $\lambda_2$** (ตัวแปรที่กำลังเปลี่ยน โดยถือ $t$ และ $\lambda$ เป็นค่าคงที่) $\dfrac{d}{d\lambda_2}\left[\lambda\left(e^{-\lambda t}-e^{-\lambda_2 t}\right)\right]=\lambda t\,e^{-\lambda_2 t}$ (หลักเกณฑ์ลูกโซ่: อนุพันธ์ของ $-e^{-\lambda_2 t}$ คือ $+t e^{-\lambda_2 t}$) และ $\dfrac{d}{d\lambda_2}(\lambda_2-\lambda)=1$
3. **หาลิมิต** $$\lim_{\lambda_2\to\lambda}\frac{N_2}{N_0}=\lim_{\lambda_2\to\lambda}\frac{\lambda t e^{-\lambda_2 t}}{1}=\lambda t\,e^{-\lambda t}.$$ ไอโซโทปลูกเพิ่มขึ้นแบบเชิงเส้นในตอนแรก (ตัวประกอบ $t$) แล้วจึงสลายตัวลง (ตัวประกอบ $e^{-\lambda t}$)
4. **หาค่าสูงสุด (กรณีค่าเริ่มต้น $\lambda_1=0.1$, $\lambda_2=0.15$)** ตั้ง $\dfrac{dN_2}{dt}=0$: $\lambda_1e^{-\lambda_1t}=\lambda_2e^{-\lambda_2t}$ จึงได้ $t_{\max}=\dfrac{\ln(\lambda_2/\lambda_1)}{\lambda_2-\lambda_1}=\dfrac{\ln1.5}{0.05}\approx8.11$ และ $N_2/N_0=\dfrac{0.1}{0.05}\left(e^{-0.811}-e^{-1.216}\right)\approx0.2963$
5. **กรณีค่าคงที่เท่ากัน** สำหรับ $N_2/N_0=\lambda te^{-\lambda t}$ กฎผลคูณให้ $\dfrac{d}{dt}=\lambda e^{-\lambda t}(1-\lambda t)=0$ จึงได้ $t_{\max}=1/\lambda=10$ และค่าสูงสุดคือ $e^{-1}\approx0.3679$ (สังเกตว่า $t_{\max}=\ln(\lambda_2/\lambda_1)/(\lambda_2-\lambda_1)$ เองก็เป็นรูป $\tfrac00$ เมื่อค่าคงที่เท่ากัน ใช้โลปีตาลอีกครั้งก็ได้ $1/\lambda$)
6. **เหตุที่คอมพิวเตอร์ต้องใช้สูตรนี้** เมื่อ $\lambda=0.1$, $t=10$ และ $\lambda_2-\lambda_1=10^{-13}$ สูตรเดิมให้ $0.36793$ แทนค่าจริง $0.36788$ เพราะการลบเลขชี้กำลังสองค่าที่เกือบเท่ากันทำให้เลขนัยสำคัญหายไป (catastrophic cancellation) สูตรจากโลปีตาลไม่มีการลบลักษณะนี้จึงแม่นยำ

```python
l1, l2, t, N0 = sp.symbols('lambda_1 lambda_2 t N_0', positive=True)
num = N0*l1*(sp.exp(-l1*t) - sp.exp(-l2*t)); den = l2 - l1
print("at λ2 = λ1:  numerator =", num.subs(l2, l1), ", denominator =", den.subs(l2, l1), " -> 0/0")
lhop = sp.diff(num, l2)/sp.diff(den, l2)
show(r"\text{L'Hôpital}", sp.simplify(lhop.subs(l2, l1)))
show(r"\text{SymPy limit}", sp.limit(num/den, l2, l1))

# Why it matters numerically: the original formula loses accuracy when λ2 ≈ λ1
lam, tv = 0.1, 10.0
for eps in [1e-2, 1e-6, 1e-10, 1e-13]:
    bateman = lam/eps*(np.exp(-lam*tv) - np.exp(-(lam+eps)*tv))
    print(f"λ2-λ1 = {eps:.0e}: Bateman formula = {bateman:.10f}   limit λt e^(-λt) = {lam*tv*np.exp(-lam*tv):.10f}")
```

```python
tt = np.linspace(0, 60, 400); lam = 0.1
fig, ax = plt.subplots()
for l2v in [0.3, 0.15, 0.11]:
    ax.plot(tt, lam/(l2v-lam)*(np.exp(-lam*tt) - np.exp(-l2v*tt)), label=f"λ₂ = {l2v}")
ax.plot(tt, lam*tt*np.exp(-lam*tt), "k--", lw=2, label="limit λ₂ → λ₁ = 0.1 : λt e^(−λt)")
ax.set(xlabel="time", ylabel="N₂ / N₀", title="Daughter abundance; the curves approach the L'Hôpital limit"); ax.legend(); plt.show()
```

**การตีความ** ลิมิต $N_2 = N_0\lambda t e^{-\lambda t}$ มีค่าจำกัดอย่างสมบูรณ์ และมีค่าสูงสุดที่ $t = 1/\lambda$
สูตรโดยตรงประสบปัญหา **การหักล้างอย่างรุนแรง (catastrophic cancellation)** เมื่อ $\lambda_2\approx\lambda_1$ (ดูตัวเลขด้านบน) รูปแบบจากโลปีตาลจึงเป็นสูตรที่ดีกว่าสำหรับใช้ในโปรแกรมคอมพิวเตอร์ด้วย

**ความหมายของคำตอบ** แม้ไอโซโทปแม่และลูกสลายตัวด้วยอัตราเท่ากัน ปริมาณไอโซโทปลูกยังมีค่าจำกัด โดยเพิ่มขึ้นจนสูงสุดที่ $t=1/\lambda$ (หนึ่งช่วงชีวิตเฉลี่ย) ที่ $N_0/e\approx37\%$ ของไอโซโทปแม่ตั้งต้น แล้วจึงลดลง นักธรณีกาลวิทยาที่ศึกษาอนุกรมยูเรเนียมพบค่าคงที่ที่ใกล้กันมาก และสูตรลิมิตคือสูตรที่เชื่อถือได้ ตรวจสอบ: เมื่อเลื่อน $\lambda_2$ เข้าหา $\lambda_1$ ในแอป กราฟจะเคลื่อนเข้าหาเส้นประของลิมิตอย่างราบเรียบ ไม่พุ่งขึ้นเป็นอนันต์

**ข้อผิดพลาดที่พบบ่อย**

- หาอนุพันธ์เทียบกับ $t$ แทน $\lambda_2$ ลิมิตนี้เป็นลิมิตใน $\lambda_2$ จึงต้องหาอนุพันธ์เทียบกับ $\lambda_2$
- ใช้กฎผลหารกับ $\dfrac{f}{g}$ ซึ่งผิด หลักเกณฑ์โลปีตาลหาอนุพันธ์ของเศษและส่วน **แยกกัน** คือ $\dfrac{f'}{g'}$
- ใช้โลปีตาลโดยไม่ตรวจก่อนว่าเป็นรูป $\tfrac00$ หรือ $\tfrac{\infty}{\infty}$ จริง

**ลองทำดู** ใช้อนุพันธ์พิสูจน์ว่า $N_0\lambda t e^{-\lambda t}$ มีค่ามากที่สุดที่ $t = 1/\lambda$ และหาค่าสูงสุดนั้น

## ตัวอย่างที่ 28 ⭐⭐: การชะล้างทะเลสาบที่ปนเปื้อน (กฎของโลปีตาลกับรูปแบบ $1^\infty$)

**บริบท** ในแต่ละปี ทะเลสาบชะล้างสารมลพิษที่ละลายอยู่ออกไปเป็นสัดส่วน $k = 0.3$ หากการชะล้างเกิดเป็น $n$ ขั้นเท่า ๆ กันต่อปี สัดส่วนที่เหลือหลัง $t$ ปีคือ $\left(1-\dfrac kn\right)^{nt}$
เมื่อขั้นตอนกลายเป็นแบบต่อเนื่อง ($n\to\infty$) นิพจน์นี้อยู่ในรูปแบบยังไม่กำหนด (indeterminate form) $1^\infty$

**แคลคูลัส** ใส่ลอการิทึม: $\ln y = t\,\dfrac{\ln(1-k/n)}{1/n}\to\dfrac00$ ให้ $u = 1/n$ แล้วใช้กฎของโลปีตาล (L'Hôpital's rule):

$$\lim_{u\to0^+}\frac{\ln(1-ku)}{u} = \lim_{u\to0^+}\frac{-k/(1-ku)}{1} = -k\quad\Rightarrow\quad\lim_{n\to\infty}\left(1-\frac kn\right)^{nt} = e^{-kt}.$$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** การชะล้างทะเลสาบแบบเป็นขั้น $n$ ครั้งต่อปีจะเหลือสารมลพิษ $\left(1-\dfrac kn\right)^{nt}$ เมื่อให้ $n\to\infty$ จะได้รูปแบบไม่กำหนด (indeterminate form) $1^\infty$ ฐานเข้าใกล้ 1 แต่เลขชี้กำลังเข้าใกล้อนันต์ การใส่ลอการิทึมเปลี่ยนเป็นรูป $\tfrac00$ ซึ่งใช้หลักเกณฑ์โลปีตาล (L'Hôpital's rule) ได้ และคำตอบอธิบายว่าทำไมกฎการลดลงแบบต่อเนื่องจึงใช้ $e$

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ตั้งชื่อลิมิต** ให้ $y=\left(1-\dfrac kn\right)^{nt}$ เมื่อ $n\to\infty$ จะได้ $1-\dfrac kn\to1$ และ $nt\to\infty$ การแทนค่าตรง ๆ ให้ $1^\infty$ ซึ่ง**ไม่**จำเป็นต้องเท่ากับ 1
2. **ใส่ลอการิทึม** เพื่อดึงเลขชี้กำลังลงมา: $\ln y=nt\ln\!\left(1-\dfrac kn\right)=t\,\dfrac{\ln(1-k/n)}{1/n}$ ตอนนี้ตัวเศษ $\to\ln1=0$ และตัวส่วน $\to0$ เป็นรูป $\tfrac00$
3. **แทน $u=1/n$** ให้ $u\to0^+$: $\displaystyle\lim_{u\to0^+}\frac{\ln(1-ku)}{u}$ ใช้หลักเกณฑ์โลปีตาล หาอนุพันธ์ตัวเศษและตัวส่วนแยกกัน $\dfrac{d}{du}\ln(1-ku)=\dfrac{-k}{1-ku}$ และ $\dfrac{d}{du}u=1$
4. **หาค่า** $\displaystyle\lim_{u\to0^+}\frac{-k/(1-ku)}{1}=-k$ ดังนั้น $\ln y\to-kt$ และเนื่องจาก $e^x$ ต่อเนื่อง จึงได้ $y\to e^{-kt}$
5. **ตัวเลขเมื่อ $k=0.3$, $t=5$ ปี** รายปี ($n=1$): $0.7^5=0.16807$ รายเดือน ($n=12$): $0.2189$ รายวัน ($n=365$): $0.2230$ ลิมิต: $e^{-1.5}=0.22313$ ผลต่างระหว่างแบบรายปีกับลิมิตประมาณ $0.055$

```python
u, kk, t = sp.symbols('u k t', positive=True)
num, den = sp.log(1 - kk*u), u
print("at u = 0:", num.subs(u, 0), "/", den.subs(u, 0), " -> 0/0")
show(r"\lim_{u\to0}\frac{\ln(1-ku)}{u}", sp.limit(sp.diff(num, u)/sp.diff(den, u), u, 0))
n = sp.symbols('n', positive=True)
show(r"\lim_{n\to\infty}\left(1-\frac kn\right)^{nt}", sp.limit((1 - kk/n)**(n*t), n, sp.oo))

kv, tv = 0.3, 5
for nv in (1, 2, 12, 365):
    print(f"n = {nv:>4} steps/yr: fraction left after {tv} yr = {(1 - kv/nv)**(nv*tv):.4f}")
print(f"continuous limit e^(-kt)            = {np.exp(-kv*tv):.4f}")
```

```python
T = np.linspace(0, 15, 300)
fig, ax = plt.subplots()
for nv in (1, 2, 12):
    tt_ = np.arange(0, 15 + 1e-9, 1/nv)
    ax.step(tt_, (1 - kv/nv)**(nv*tt_), where="post", lw=1.5, label=f"n = {nv} steps/yr")
ax.plot(T, np.exp(-kv*T), "k", lw=2, label="limit e^(−kt)")
ax.set(xlabel="time (yr)", ylabel="fraction of pollutant left", title="Discrete flushing approaches exponential decay"); ax.legend(); plt.show()
```

**การตีความ** หากชะล้างปีละครั้ง หลัง 5 ปีจะเหลือสารมลพิษ 16.8 % เมื่อเพิ่มจำนวนขั้น คำตอบจะเข้าใกล้ค่าแบบต่อเนื่อง $e^{-1.5} = 22.3\,\%$
ลิมิต $(1-k/n)^n\to e^{-k}$ คือเหตุผลที่ฟังก์ชันเลขชี้กำลังปรากฏอยู่ทั่วไปในธรณีศาสตร์ ทั้งการสลายตัวของสารกัมมันตรังสี การลดลงของน้ำบาดาล และการเย็นตัว ล้วนเกิดจากการสูญเสียตามสัดส่วนเล็ก ๆ จำนวนมาก

**ความหมายของคำตอบ** การกำจัด 30% ครั้งเดียวต่อปีได้ผลมากกว่าการกำจัดแบบต่อเนื่อง หลัง 5 ปีเหลือ 16.8% แทนที่จะเป็น 22.3% เพราะแต่ละขั้นรายปีกระทำกับปริมาณทั้งหมด เมื่อขั้นเล็กลงเรื่อย ๆ กราฟขั้นบันไดจะเข้าใกล้กราฟเอกซ์โพเนนเชียลเรียบ $e^{-kt}$ ซึ่งเป็นแบบจำลองของการชะล้าง การสลายกัมมันตรังสี และกระบวนการอันดับหนึ่งอื่น ๆ

**ข้อผิดพลาดที่พบบ่อย**

- สรุปว่า $1^\infty=1$ ฐานและเลขชี้กำลังแข่งกันอยู่ ต้องใส่ลอการิทึมก่อน
- ใช้กฎผลหารกับ $\dfrac{\ln(1-ku)}{u}$ แทนโลปีตาล หลักเกณฑ์โลปีตาลหาอนุพันธ์ตัวเศษและตัวส่วน**แยกกัน**
- หยุดที่ $\ln y\to-kt$ แล้วลืมยกกำลัง $e$ กลับเป็น $y=e^{-kt}$

**ลองทำดู** ในแบบจำลองต่อเนื่อง ต้องใช้เวลากี่ปีจนเหลือสารมลพิษเพียง 5 %

# ส่วน E: ฟังก์ชันหลายตัวแปร (Functions of Several Variables)

⭐⭐⭐ ยาก · ตัวอย่างที่ 29–33 · สัปดาห์ที่ 11 ในประมวลรายวิชา

## ตัวอย่างที่ 29 ⭐⭐⭐: อะไรทำให้น้ำทะเลจม (อนุพันธ์ย่อย)

**บริบท** ความหนาแน่นของน้ำทะเลขึ้นกับอุณหภูมิ $T$ (°C) และความเค็ม $S$ (g/kg) สมการสถานะอย่างง่ายคือ

$$\rho(T,S) = 1027\left[1-\alpha_0(T-10)-\frac\gamma2(T-10)^2+\beta_0(S-35)\right],$$

โดย $\alpha_0 = 1.7\times10^{-4}$, $\gamma = 1\times10^{-5}$, $\beta_0 = 7.6\times10^{-4}$

**แคลคูลัส** อนุพันธ์ย่อย (partial derivative) $\dfrac{\partial\rho}{\partial T}$ (ให้ $S$ คงที่) และ $\dfrac{\partial\rho}{\partial S}$ (ให้ $T$ คงที่) บอกว่าแต่ละสมบัติควบคุมความหนาแน่นมากเพียงใด สัมประสิทธิ์การขยายตัวเชิงความร้อนคือ $\alpha = -\dfrac1\rho\dfrac{\partial\rho}{\partial T}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ความหนาแน่นของน้ำทะเลขึ้นกับสองตัวแปร คืออุณหภูมิ $T$ และความเค็ม $S$ อนุพันธ์ย่อย (partial derivative) เปลี่ยนตัวแปรหนึ่งโดยตรึงอีกตัวไว้ ดังนั้น $\partial\rho/\partial T$ และ $\partial\rho/\partial S$ จึงวัดแยกกันว่าความร้อนและเกลือควบคุมความหนาแน่นมากเพียงใด ซึ่งเป็นตัวขับเคลื่อนการไหลเวียนของมหาสมุทรลึก

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **แบบจำลอง** $\rho(T,S)=1027\left[1-\alpha_0(T-10)-\dfrac\gamma2(T-10)^2+\beta_0(S-35)\right]$ โดย $\alpha_0=1.7\times10^{-4}$, $\gamma=10^{-5}$, $\beta_0=7.6\times10^{-4}$
2. **อนุพันธ์ย่อยเทียบ $T$** (ถือ $S$ เป็นค่าคงตัว พจน์ความเค็มจึงมีอนุพันธ์เป็น 0): $\dfrac{\partial\rho}{\partial T}=1027\left[-\alpha_0-\gamma(T-10)\right]=-1027\left[\alpha_0+\gamma(T-10)\right]$
3. **อนุพันธ์ย่อยเทียบ $S$** (ถือ $T$ เป็นค่าคงตัว): $\dfrac{\partial\rho}{\partial S}=1027\,\beta_0=0.7805$ kg/m³ ต่อ g/kg มีค่าเท่ากันทุกจุดเพราะแบบจำลองเป็นเชิงเส้นใน $S$
4. **น้ำผิวหน้าเขตร้อน $T=25$ °C, $S=34$ g/kg** $T-10=15$: $\rho=1027[1-0.00255-0.001125-0.00076]=1022.445$ kg/m³ และ $\dfrac{\partial\rho}{\partial T}=-1027(1.7\times10^{-4}+1.5\times10^{-4})=-0.3286$ kg/m³ ต่อ °C
5. **เปรียบเทียบสองผล** สัมประสิทธิ์การขยายตัวทางความร้อน $\alpha=-\dfrac1\rho\dfrac{\partial\rho}{\partial T}=\dfrac{0.3286}{1022.445}\approx3.21\times10^{-4}$ /°C เกลือ 1 g/kg เปลี่ยนความหนาแน่นได้เท่ากับอุณหภูมิประมาณ $\dfrac{0.7805}{0.3286}\approx2.38$ °C
6. **น้ำขั้วโลก $T=2$ °C** $\dfrac{\partial\rho}{\partial T}=-1027(1.7\times10^{-4}-8\times10^{-5})\approx-0.092$ kg/m³ ต่อ °C น้อยกว่าหนึ่งในสามของค่าในเขตร้อน ขณะที่ $\partial\rho/\partial S$ ไม่เปลี่ยน

```python
T, S = sp.symbols('T S', real=True)
a0, gm, b0 = 1.7e-4, 1e-5, 7.6e-4
rho = 1027*(1 - a0*(T - 10) - gm/2*(T - 10)**2 + b0*(S - 35))
rT, rS = sp.diff(rho, T), sp.diff(rho, S)
show(r"\frac{\partial\rho}{\partial T}", sp.expand(rT)); show(r"\frac{\partial\rho}{\partial S}", rS)
for name, Tv, Sv in [("tropical surface (Andaman Sea)", 25, 34.0), ("polar water", 2, 34.7)]:
    P = {T: Tv, S: Sv}
    print(f"{name:32s} ρ = {float(rho.subs(P)):.2f} kg/m³, ∂ρ/∂T = {float(rT.subs(P)):.4f}, "
          f"∂ρ/∂S = {float(rS.subs(P)):.4f}, α = {-float(rT.subs(P)/rho.subs(P)):.2e} /K")
```

```python
TT, SS = np.meshgrid(np.linspace(-2, 30, 200), np.linspace(32, 37, 200))
R = sp.lambdify((T, S), rho)(TT, SS)
fig, ax = plt.subplots(figsize=(7, 5))
cs = ax.contour(SS, TT, R, levels=np.arange(1018, 1032, 1), cmap="viridis"); ax.clabel(cs, fmt="%d", fontsize=8)
ax.plot([34.0, 34.7], [25, 2], "r*", ms=12)
ax.set(xlabel="salinity S (g/kg)", ylabel="temperature T (°C)", title="Density contours ρ(T, S): curved isopycnals"); plt.show()
```

**การตีความ** ในน้ำอุ่นเขตร้อน การเปลี่ยนแปลงอุณหภูมิ 1 °C เปลี่ยนความหนาแน่นมากกว่าในน้ำเขตขั้วโลกประมาณ 3.5 เท่า ($\alpha$ เท่ากับ $3.2\times10^{-4}$ เทียบกับ $9\times10^{-5}$ ต่อ K) ใกล้จุดเยือกแข็ง **ความเค็ม** เป็นตัวควบคุมความหนาแน่น
นี่คือเหตุผลที่น้ำลึกก่อตัวในทะเลขั้วโลกเมื่อน้ำแข็งทะเลขับเกลือออกมา ซึ่งขับเคลื่อนการไหลเวียนแบบ "สายพานลำเลียง" ของมหาสมุทรโลก เส้นชั้นความสูงที่โค้งเกิดจากพจน์ $(T-10)^2$

**ความหมายของคำตอบ** น้ำอุ่นเบากว่า ($\partial\rho/\partial T$ เป็นลบ) และน้ำเค็มหนักกว่า ($\partial\rho/\partial S$ เป็นบวก) ในทะเลขั้วโลกที่เย็น อุณหภูมิแทบไม่มีผล ความเค็มจึงเป็นตัวตัดสินว่าน้ำจะจมหรือไม่ เมื่อน้ำแข็งทะเลก่อตัวจะทิ้งเกลือไว้ น้ำจึงหนาแน่นขึ้นและจมลง เป็นต้นกำเนิดของสายพานน้ำลึกทั่วโลก ตรวจสอบความสมเหตุสมผล: ความหนาแน่นทั้งหมดอยู่ราว 1020–1030 kg/m³ เหมือนน้ำทะเลจริง

**ข้อผิดพลาดที่พบบ่อย**

- หาอนุพันธ์ของพจน์ความเค็มเทียบกับ $T$ เมื่อหา $\partial/\partial T$ ทุกพจน์ที่ไม่มี $T$ เป็นค่าคงตัวและมีอนุพันธ์เป็น 0
- ทำเครื่องหมายลบหาย ความหนาแน่น**ลดลง**เมื่ออุณหภูมิสูงขึ้น ดังนั้น $\partial\rho/\partial T<0$
- เปรียบเทียบ $\partial\rho/\partial T$ กับ $\partial\rho/\partial S$ ตรง ๆ โดยไม่ดูหน่วย ทั้งสองมีหน่วยต่อ °C และต่อ g/kg จึงควรเทียบอัตราส่วน (กี่ °C เทียบเท่า 1 g/kg)

**ลองทำดู** ฝนทำให้ความเค็มของน้ำเขตร้อนลดลง 1 g/kg อุณหภูมิต้องสูงขึ้นเท่าใดจึงจะทำให้ความหนาแน่นเปลี่ยนเท่ากัน

## ตัวอย่างที่ 30 ⭐⭐⭐: ความชัน ทิศด้านลาด และทิศทางการไหลจากแบบจำลองระดับสูงเชิงเลข (อนุพันธ์ย่อย)

**บริบท** แบบจำลองระดับสูงเชิงเลข (digital elevation model, DEM) คือฟังก์ชัน $z(x,y)$ แผนที่ความชันทุกแผ่นใน GIS สร้างจากอนุพันธ์ย่อย (partial derivatives) ของฟังก์ชันนี้:

* เกรเดียนต์ $\nabla z = \left(\dfrac{\partial z}{\partial x},\ \dfrac{\partial z}{\partial y}\right)$ ชี้ขึ้นเนิน
* มุมความชัน $\theta = \arctan\lVert\nabla z\rVert$
* น้ำไหลไปในทิศทางของ $-\nabla z$ (ทิศที่ลาดลงชันที่สุด)

ภูมิประเทศสังเคราะห์ ($x, y$ หน่วย km, $z$ หน่วย m): เนินสองลูกบนพื้นที่ลาดเอียงระดับภูมิภาค
$z = 600e^{-[(x-1)^2+(y-0.5)^2]/1.5} + 350e^{-[(x+1.5)^2+(y+1)^2]/0.8} + 40x$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** แบบจำลองความสูงเชิงเลข (DEM) คือฟังก์ชันสองตัวแปร $z(x,y)$ **อนุพันธ์ย่อย (partial derivatives)** $\partial z/\partial x$ และ $\partial z/\partial y$ คืออัตราการไต่ขึ้นเมื่อเดินไปทางตะวันออกและทางเหนือ เมื่อรวมกันเป็น **เกรเดียนต์ (gradient)** $\nabla z$ ซึ่งชี้ขึ้นเนินโดยตรง ความยาวของเกรเดียนต์ให้ความลาดชัน และ $-\nabla z$ คือทิศทางที่น้ำไหล

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ตั้งชื่อส่วนประกอบ** เขียน $z=g_1+g_2+40x$ โดย $g_1=600e^{-[(x-1)^2+(y-0.5)^2]/1.5}$ (เนินหลัก) และ $g_2=350e^{-[(x+1.5)^2+(y+1)^2]/0.8}$ (เนินเล็ก) ที่จุดวัด $(0,-0.5)$: $g_1=600e^{-4/3}=158.16$ m และ $g_2=350e^{-3.125}=15.38$ m ดังนั้น $z=158.16+15.38+0=173.5$ m
2. **อนุพันธ์ย่อยเทียบกับ $x$** (ถือ $y$ เป็นค่าคงที่) จากหลักเกณฑ์ลูกโซ่ $\dfrac{\partial}{\partial x}e^{-u}=-e^{-u}\dfrac{\partial u}{\partial x}$ จึงได้ $$\frac{\partial z}{\partial x}=g_1\cdot\frac{-2(x-1)}{1.5}+g_2\cdot\frac{-2(x+1.5)}{0.8}+40.$$ ที่จุดวัด: $158.16(1.333)+15.38(-3.75)+40=210.88-57.67+40=193.2$ m/km
3. **อนุพันธ์ย่อยเทียบกับ $y$** (ถือ $x$ เป็นค่าคงที่ พจน์ความเอียง $40x$ ไม่มี $y$ จึงหายไป): $$\frac{\partial z}{\partial y}=g_1\cdot\frac{-2(y-0.5)}{1.5}+g_2\cdot\frac{-2(y+1)}{0.8}=210.88-19.22=191.7\ \text{m/km}.$$
4. **มุมความลาดชัน** $\lVert\nabla z\rVert=\sqrt{193.2^2+191.7^2}=272.1$ m/km $=0.2721$ m/m (หารด้วย 1000 เพื่อให้ความยาวทั้งสองเป็นเมตร) จากนั้น $\theta=\arctan(0.2721)\approx15.2^\circ$
5. **ทิศทางการไหล** น้ำเคลื่อนตาม $-\nabla z=(-193.2,-191.7)$ คือไปทางตะวันตกและทางใต้ในปริมาณเกือบเท่ากัน เมื่อเขียนเป็นมุมทิศ (0° = เหนือ, 90° = ตะวันออก) ได้ประมาณ $225^\circ$ คือไปทางตะวันตกเฉียงใต้

```python
x, y = sp.symbols('x y', real=True)
zexpr = 600*sp.exp(-((x-1)**2 + (y-0.5)**2)/1.5) + 350*sp.exp(-((x+1.5)**2 + (y+1)**2)/0.8) + 40*x
zx, zy = sp.diff(zexpr, x), sp.diff(zexpr, y)
show(r"\frac{\partial z}{\partial x}", zx)

P = {x: 0.0, y: -0.5}                               # a point between the hills
gx, gy = float(zx.subs(P)), float(zy.subs(P))       # m per km
grad_mag = np.hypot(gx, gy)/1000                    # dimensionless (m per m)
print(f"at (x, y) = (0, -0.5) km: ∂z/∂x = {gx:.1f} m/km, ∂z/∂y = {gy:.1f} m/km")
print(f"slope angle = arctan|∇z| = {np.degrees(np.arctan(grad_mag)):.1f}°")
flow = np.degrees(np.arctan2(-gx, -gy)) % 360       # compass bearing of -∇z (0° = North = +y)
print(f"water flows toward bearing {flow:.0f}° (0° = north, 90° = east)")
```

```python
X, Y = np.meshgrid(np.linspace(-4, 4, 161), np.linspace(-3.5, 3.5, 141))
Z = sp.lambdify((x, y), zexpr)(X, Y)
GX, GY = sp.lambdify((x, y), zx)(X, Y)/1000, sp.lambdify((x, y), zy)(X, Y)/1000
SLOPE = np.degrees(np.arctan(np.hypot(GX, GY)))
fig, (a1, a2) = plt.subplots(1, 2, figsize=(12, 4.8))
cs = a1.contourf(X, Y, Z, 20, cmap="terrain"); a1.contour(X, Y, Z, 20, colors="k", linewidths=.3)
s = (slice(None, None, 10), slice(None, None, 10))
a1.quiver(X[s], Y[s], -GX[s], -GY[s], color="navy", width=.003)
a1.plot(0, -0.5, "r*", ms=12); fig.colorbar(cs, ax=a1, label="elevation (m)")
a1.set(title="elevation + flow direction −∇z", xlabel="x (km)", ylabel="y (km)", aspect="equal")
im = a2.pcolormesh(X, Y, SLOPE, cmap="magma_r", shading="auto"); fig.colorbar(im, ax=a2, label="slope (°)")
a2.set(title="slope map θ = arctan|∇z|", xlabel="x (km)", aspect="equal"); plt.tight_layout(); plt.show()
```

**การตีความ** ลูกศรตั้งฉากกับเส้นชั้นความสูงและชี้ลงเนิน ตรงกับทิศที่น้ำไหลจริง ความชันเป็นศูนย์บนยอดเนิน (ที่ $\nabla z = 0$) และมากที่สุดบนไหล่เนิน
แผนที่ความเสี่ยงดินถล่ม ขอบเขตลุ่มน้ำ และโครงข่ายทางน้ำ ล้วนสร้างจากอนุพันธ์ย่อยเหล่านี้

**ความหมายของคำตอบ** จุดวัดอยู่บนไหล่ด้านตะวันตกเฉียงใต้ของเนินหลัก มีความลาดชันปานกลาง 15° และน้ำฝนที่ตกตรงนั้นจะไหลไปทางตะวันตกเฉียงใต้ ตั้งฉากกับเส้นชั้นความสูง ซอฟต์แวร์ GIS คำนวณอนุพันธ์ย่อยสองค่านี้ทุกช่องกริดของ DEM (ด้วยผลต่างจำกัด) เพื่อทำแผนที่ความลาดชัน ทิศทางการไหล ขอบเขตลุ่มน้ำ และแผนที่ความเสี่ยงดินถล่ม ตรวจสอบ: บนยอดเนินอนุพันธ์ย่อยทั้งสองเป็นศูนย์ ความลาดชันจึงเป็น 0° และน้ำไม่มีทิศทางที่ชัดเจน

**ข้อผิดพลาดที่พบบ่อย**

- ใช้หน่วยปนกัน เกรเดียนต์มีหน่วย m/km จึงต้องหารด้วย 1000 ก่อนหา $\arctan$ มิฉะนั้นจะได้มุมเกือบ $90^\circ$
- ลืมว่าน้ำไหลตาม $-\nabla z$ (ลงเนิน) ไม่ใช่ $+\nabla z$ (ขึ้นเนิน)
- ตกหล่นค่าคงที่ $40$ ใน $\partial z/\partial x$ หรือใส่ไว้ใน $\partial z/\partial y$ ซึ่งไม่ควรมี

**ลองทำดู** หาจุดวิกฤตของ $z$ (ที่ $\nabla z = 0$) เชิงตัวเลขโดยใช้ `optimize.minimize` กับ $-z$ มีจุดอานม้า (saddle) ระหว่างเนินทั้งสองหรือไม่

## ตัวอย่างที่ 31 ⭐⭐⭐: ปัญหาสามจุด: น้ำบาดาลไหลไปทางไหน (เกรเดียนต์และอนุพันธ์ระบุทิศทาง)

**บริบท** วัดระดับพลังงานน้ำ (hydraulic head) ในบ่อสามบ่อ ($x$ ทิศตะวันออก, $y$ ทิศเหนือ, หน่วยเมตร):

| บ่อ | $x$ | $y$ | ระดับพลังงานน้ำ $h$ (m) |
|---|---|---|---|
| A | 0 | 0 | 52.0 |
| B | 400 | 0 | 50.4 |
| C | 100 | 300 | 50.9 |

สมมติว่าระดับน้ำใต้ดินเป็นระนาบ $h = h_0 + ax + by$

**แคลคูลัส** เกรเดียนต์ (gradient) $\nabla h = (a, b) = \left(\dfrac{\partial h}{\partial x}, \dfrac{\partial h}{\partial y}\right)$ น้ำบาดาลไหลไปตามทิศ $-\nabla h$ อนุพันธ์ระบุทิศทาง (directional derivative) ในทิศเวกเตอร์หนึ่งหน่วย $\mathbf u$ คือ $D_{\mathbf u}h = \nabla h\cdot\mathbf u$
ความเร็วการซึมตามกฎของดาร์ซี: $v = K\lVert\nabla h\rVert/n$ โดย $K = 10$ m/day และความพรุน $n = 0.25$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** บ่อน้ำสามบ่อกำหนดระนาบของระดับน้ำใต้ดิน $h(x,y)=h_0+ax+by$ อนุพันธ์ย่อยของระนาบรวมกันเป็นเกรเดียนต์ (gradient) $\nabla h=(a,b)$ ซึ่งชี้ขึ้นเนิน น้ำใต้ดินไหลไปทางตรงข้าม คือตาม $-\nabla h$ และความเร็วหาได้จากกฎของดาร์ซี วิธีสามจุดแบบเดียวกันใช้หาแนวระดับและมุมเท (strike and dip) จากหลุมเจาะสามหลุมได้

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ใช้บ่อ A ที่จุดกำเนิด** $h(0,0)=52.0$ ดังนั้น $h_0=52.0$
2. **บ่อ B ที่ $(400,0)$:** $52.0+400a=50.4\Rightarrow a=\dfrac{-1.6}{400}=-0.004$ ค่านี้คือ $\dfrac{\partial h}{\partial x}$ ความชันในทิศตะวันออก
3. **บ่อ C ที่ $(100,300)$:** $52.0+100(-0.004)+300b=50.9\Rightarrow300b=-0.7\Rightarrow b\approx-0.002333=\dfrac{\partial h}{\partial y}$
4. **ขนาดของเกรเดียนต์** $\lVert\nabla h\rVert=\sqrt{0.004^2+0.002333^2}\approx0.004631$ คือระดับน้ำลดลงประมาณ 4.6 m ต่อ km ในทิศที่ชันที่สุด
5. **ทิศการไหล** $-\nabla h=(0.004,\ 0.002333)$ ชี้ไปทางตะวันออกและเหนือ มุมทิศ (bearing) จากทิศเหนือคือ $\arctan\dfrac{0.004}{0.002333}\approx59.7^\circ$ (ไปทางตะวันออกเฉียงเหนือ)
6. **ความเร็วและอนุพันธ์ระบุทิศทาง** ความเร็วซึม $v=\dfrac{K\lVert\nabla h\rVert}{n}=\dfrac{10\times0.004631}{0.25}\approx0.185$ m/วัน ($\approx68$ m/ปี) ในทิศตะวันออก $D_{\mathbf u}h=\nabla h\cdot(1,0)=-0.004$ ตามทิศการไหล $D_{\mathbf u}h=-\lVert\nabla h\rVert=-0.004631$ ซึ่งเป็นการลดลงที่ชันที่สุดที่เป็นไปได้

```python
A = np.array([[1, 0, 0], [1, 400, 0], [1, 100, 300]], float)
hv = np.array([52.0, 50.4, 50.9])
h0, a, b = np.linalg.solve(A, hv)
grad = np.array([a, b]); mag = np.linalg.norm(grad)
bearing = np.degrees(np.arctan2(-a, -b)) % 360
print(f"plane: h = {h0:.2f} + ({a:.6f}) x + ({b:.6f}) y")
print(f"∂h/∂x = {a:.6f}, ∂h/∂y = {b:.6f}, |∇h| = {mag:.6f}")
print(f"flow direction (−∇h) bearing = {bearing:.1f}° (0° = north, 90° = east)")
print(f"seepage velocity = K|∇h|/n = {10*mag/0.25:.3f} m/day = {10*mag/0.25*365:.0f} m/yr")
for name, ang in [("due east", 90), ("due north", 0), ("along the flow", bearing)]:
    u = np.array([np.sin(np.radians(ang)), np.cos(np.radians(ang))])
    print(f"directional derivative {name:15s}: D_u h = {grad @ u:+.6f} m/m")
```

```python
X, Y = np.meshgrid(np.linspace(-50, 450, 100), np.linspace(-50, 350, 100))
Hh = h0 + a*X + b*Y
fig, ax = plt.subplots(figsize=(7, 5.5))
cs = ax.contour(X, Y, Hh, levels=np.arange(49.8, 52.4, 0.2), cmap="Blues_r"); ax.clabel(cs, fmt="%.1f", fontsize=8)
for (x_, y_), hh, nm in zip([(0, 0), (400, 0), (100, 300)], hv, "ABC"):
    ax.plot(x_, y_, "ks"); ax.annotate(f"{nm}: {hh} m", (x_, y_), xytext=(6, 6), textcoords="offset points")
ax.annotate("", xy=(200 - a/mag*120, 120 - b/mag*120), xytext=(200, 120), arrowprops=dict(arrowstyle="->", color="C3", lw=2.5))
ax.text(205, 100, "flow", color="C3")
ax.set(xlabel="x, east (m)", ylabel="y, north (m)", aspect="equal", title="Head contours and flow direction from three wells"); plt.show()
```

**การตีความ** ระดับน้ำใต้ดินลาดลง 0.0046 m/m และน้ำบาดาลไหลไปทางแบริ่งประมาณ **60°** (ตะวันออกเฉียงเหนือ) ด้วยอัตราประมาณ 0.19 m/day (ประมาณ 68 m/yr)
อนุพันธ์ระบุทิศทางมีค่าเป็นลบมากที่สุดตามทิศการไหล (−|∇h|) และเป็นศูนย์ตามแนวเส้นชั้นความสูง นักอุทกธรณีวิทยาใช้ "ปัญหาสามจุด" นี้ทำนายว่ากลุ่มสารปนเปื้อนจะเคลื่อนไปทางใด

**ความหมายของคำตอบ** สารปนเปื้อนที่หกใกล้บ่อ A จะเคลื่อนไปทางตะวันออกเฉียงเหนือด้วยความเร็วประมาณ 68 m ต่อปี จึงควรวางบ่อเฝ้าระวังในทิศนั้น อนุพันธ์ระบุทิศทางมีขนาดมากที่สุดตามทิศ $-\nabla h$ และเป็นศูนย์ตามเส้นชั้นระดับน้ำ นี่คือเหตุผลที่น้ำใต้ดินไหลตัดเส้นชั้นความสูงระดับน้ำเป็นมุมฉาก

**ข้อผิดพลาดที่พบบ่อย**

- บอกว่าน้ำไหลตาม $+\nabla h$ เกรเดียนต์ชี้**ขึ้นเนิน** น้ำไหลลงเนิน คือตาม $-\nabla h$
- หา $b$ จากบ่อ C โดยลืมพจน์ $100a$ บ่อ C ไม่ได้อยู่ทางเหนือของ A พอดี พิกัด $x$ ของมันจึงทำให้ $h$ เปลี่ยนด้วย
- สับสนมุมทิศ มุมทิศบนแผนที่วัดตามเข็มนาฬิกาจาก**ทิศเหนือ** ส่วน $\arctan(y/x)$ วัดจากทิศตะวันออก

**ลองทำดู** เกิดน้ำมันเชื้อเพลิงรั่วที่บ่อ A การปนเปื้อนจะใช้เวลานานเท่าใดในการเคลื่อนที่ 500 m ตามทิศการไหล

## ตัวอย่างที่ 32 ⭐⭐⭐: การแพร่ความคลาดเคลื่อนในกฎของดาร์ซี และการเย็นตัวของหินที่ถูกยกขึ้น (ผลต่างเชิงอนุพันธ์รวมและอนุพันธ์รวม)

**ส่วน (a) ผลต่างเชิงอนุพันธ์รวม (total differential)** อัตราการไหลของน้ำใต้ดินผ่านคอลัมน์ทรายหาได้จากกฎของดาร์ซี (Darcy's law) $Q = \dfrac{KA\,\Delta h}{L}$ แต่ละปริมาณวัดได้โดยมีความคลาดเคลื่อน
ผลต่างเชิงอนุพันธ์รวม

$$dQ = \frac{\partial Q}{\partial K}dK + \frac{\partial Q}{\partial A}dA + \frac{\partial Q}{\partial (\Delta h)}d(\Delta h) + \frac{\partial Q}{\partial L}dL$$

ใช้ประมาณความคลาดเคลื่อนของ $Q$

**ส่วน (b) อนุพันธ์รวม (total derivative)** หินก้อนหนึ่งถูกยกขึ้นสู่ผิวโลก ($z(t) = z_0 - u t$) ขณะที่เส้นอุณหภูมิตามความลึก (geotherm) ของเปลือกโลกกำลังเย็นลง $T(z,t) = 10 + 30\,z\,e^{-t/50}$ (°C, $z$ หน่วย km, $t$ หน่วย Myr)
อุณหภูมิที่ *หิน* ได้รับเปลี่ยนแปลงด้วยอัตรา $\dfrac{dT}{dt} = \dfrac{\partial T}{\partial z}\dfrac{dz}{dt} + \dfrac{\partial T}{\partial t}$ (กฎลูกโซ่)

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** เมื่อผลลัพธ์ขึ้นกับปริมาณที่วัดได้หลายค่า **ผลต่างเชิงอนุพันธ์รวม (total differential)** $dQ=\sum\dfrac{\partial Q}{\partial x_i}dx_i$ ใช้ประมาณว่าความคลาดเคลื่อนเล็ก ๆ ของแต่ละตัวแปรรวมกันเป็นความคลาดเคลื่อนของผลลัพธ์อย่างไร สำหรับผลคูณ/ผลหารแบบกฎของดาร์ซี สูตรนี้ลดรูปเป็น **การบวกความคลาดเคลื่อนสัมพัทธ์** ซึ่งชี้ได้ทันทีว่าการวัดใดสำคัญที่สุด แนวคิดหลักเกณฑ์ลูกโซ่เดียวกันเมื่อใช้ตามเส้นทางที่เคลื่อนที่ จะได้ **อนุพันธ์รวม (total derivative)** ของอุณหภูมิที่หินซึ่งถูกยกตัวขึ้นได้รับ

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **คำนวณ $Q$** เมื่อ $K=2\times10^{-4}$ m/s, $A=0.05$ m², $\Delta h=0.40$ m, $L=1.20$ m: $$Q=\frac{KA\,\Delta h}{L}=\frac{(2\times10^{-4})(0.05)(0.40)}{1.20}=3.333\times10^{-6}\ \text{m}^3/\text{s}.$$
2. **อนุพันธ์ย่อย** $\dfrac{\partial Q}{\partial K}=\dfrac{A\Delta h}{L}=\dfrac{Q}{K}$, $\dfrac{\partial Q}{\partial A}=\dfrac{Q}{A}$, $\dfrac{\partial Q}{\partial \Delta h}=\dfrac{Q}{\Delta h}$ และ $\dfrac{\partial Q}{\partial L}=-\dfrac{KA\Delta h}{L^2}=-\dfrac{Q}{L}$ แต่ละค่าหาได้โดยถือตัวแปรอีกสามตัวเป็นค่าคงที่
3. **หารด้วย $Q$ เพื่อให้ได้ความคลาดเคลื่อนสัมพัทธ์** $$\frac{dQ}{Q}=\frac{dK}{K}+\frac{dA}{A}+\frac{d\Delta h}{\Delta h}-\frac{dL}{L}.$$ เมื่อ $dK=2\times10^{-5}$, $dA=0.001$, $d\Delta h=0.01$, $dL=0.01$ สี่พจน์มีค่า $10\%$, $2\%$, $2.5\%$ และ $0.8\%$ ของ $Q$
4. **กรณีเลวร้ายที่สุดและกรณีทั่วไป** ในกรณีเลวร้ายที่สุด ความคลาดเคลื่อนทั้งหมดไปทางเดียวกัน จึงบวกค่าสัมบูรณ์: $10+2+2.5+0.8=15.3\%$ คือ $|dQ|\approx5.11\times10^{-7}$ m³/s ถ้าความคลาดเคลื่อนเป็นอิสระและสุ่ม จะหักล้างกันบางส่วน ค่าประมาณที่นิยมคือ $\sqrt{10^2+2^2+2.5^2+0.8^2}\approx10.53\%$
5. **อนุพันธ์รวมตามเส้นทาง** หินถูกยกตัวขึ้นตาม $z=10-0.15t$ (km, ล้านปี) ผ่านเส้นอุณหภูมิตามความลึกที่กำลังเย็นตัว $T=10+30ze^{-t/50}$ จากหลักเกณฑ์ลูกโซ่ $$\frac{dT}{dt}=\frac{\partial T}{\partial z}\frac{dz}{dt}+\frac{\partial T}{\partial t}=30e^{-t/50}(-0.15)-\frac{30z}{50}e^{-t/50}.$$ ที่ $t=0$, $z=10$: $-4.5-6.0=-10.5$ °C ต่อล้านปี

```python
K, A, Dh, L = sp.symbols('K A Delta_h L', positive=True)
Q = K*A*Dh/L
vals = {K: 2e-4, A: 0.05, Dh: 0.40, L: 1.20}         # m/s, m², m, m
errs = {K: 2e-5, A: 0.001, Dh: 0.01, L: 0.01}        # measurement uncertainties
Q0 = float(Q.subs(vals))
print(f"Q = {Q0:.3e} m³/s")
dQ = 0
for var in (K, A, Dh, L):
    part = float(sp.diff(Q, var).subs(vals))*errs[var]
    dQ += abs(part)
    print(f"  ∂Q/∂{str(var):7s}·d{str(var):7s} = {part:+.3e}  ({abs(part)/Q0*100:4.1f}% of Q)")
print(f"maximum error |dQ| ≈ {dQ:.2e} m³/s = {dQ/Q0*100:.1f}% of Q  -> K dominates; measure K better!")
```

```python
t, z = sp.symbols('t z', nonnegative=True)
z0, u = 10, 0.15                                     # start at 10 km depth, uplift 0.15 km/Myr
Tzt = 10 + 30*z*sp.exp(-t/50)
zt = z0 - u*t
dTdt_chain = sp.diff(Tzt, z)*sp.diff(zt, t) + sp.diff(Tzt, t)
dTdt_chain = dTdt_chain.subs(z, zt)
dTdt_direct = sp.diff(Tzt.subs(z, zt), t)            # substitute first, then differentiate
print("chain rule result equals direct differentiation:", sp.simplify(dTdt_chain - dTdt_direct) == 0)
for tv in (0, 20, 40):
    print(f"t = {tv:>2} Myr: depth {float(zt.subs(t, tv)):.1f} km, rock T = {float(Tzt.subs({z: zt, t: t}).subs(t, tv)):6.1f} °C, "
          f"dT/dt = {float(dTdt_chain.subs(t, tv)):.2f} °C/Myr")
```

```python
tt = np.linspace(0, 60, 300)
Tpath = sp.lambdify(t, Tzt.subs(z, zt))(tt)
fig, ax = plt.subplots()
ax.plot(tt, Tpath, lw=2, label="T experienced by the rock")
ax.axhline(180, c="C3", ls="--"); ax.text(35, 186, "zircon (U-Th)/He closure ≈ 180 °C", color="C3")
ax.axhline(110, c="C2", ls="--"); ax.text(35, 116, "apatite fission-track closure ≈ 110 °C", color="C2")
ax.set(xlabel="time (Myr)", ylabel="temperature (°C)", title="Cooling path = uplift effect + geotherm-cooling effect"); ax.legend(); plt.show()
```

**การตีความ** (a) ความไม่แน่นอน 10 % ในสภาพนำน้ำ (hydraulic conductivity) $K$ ครอบงำงบความคลาดเคลื่อน มากกว่าความคลาดเคลื่อนอื่นทั้งหมดรวมกัน ผลต่างเชิงอนุพันธ์รวมบอกว่าควรลงทุนวัดค่าใดให้แม่นยำขึ้น
(b) หินเย็นลงด้วยสองสาเหตุพร้อมกัน: หินเคลื่อนขึ้นตามเส้นอุณหภูมิตามความลึก ($\partial T/\partial z\cdot dz/dt$) **และ** เส้นอุณหภูมินั้นเองก็เย็นลง ($\partial T/\partial t$)
นักเทอร์โมโครโนโลยีใช้เหตุผลนี้แปลงอายุการเย็นตัวของแร่ให้เป็นอัตราการยกตัว

**ความหมายของคำตอบ** ความคลาดเคลื่อน 10% ของสัมประสิทธิ์การซึมผ่านได้ $K$ เพียงค่าเดียวมากกว่าความคลาดเคลื่อนอื่นรวมกันทั้งหมด วิธีที่ดีที่สุดในการปรับปรุงค่าประมาณอัตราการไหลจึงเป็นการทดสอบหาค่า $K$ ให้ดีขึ้น (เช่นการทดสอบสูบน้ำ) ไม่ใช่การใช้ไม้บรรทัดที่ดีขึ้น ในปัญหาการเย็นตัว หินเย็นลงด้วยอัตรา 10.5 °C ต่อล้านปีในตอนแรก โดย 4.5 มาจากการเคลื่อนขึ้นตามเส้นอุณหภูมิ และ 6.0 มาจากการที่เส้นอุณหภูมิเองเย็นลง นักเทอร์โมโครโนโลยีใช้การแยกส่วนนี้แปลงอายุการเย็นตัวของแร่เป็นอัตราการยกตัว

**ข้อผิดพลาดที่พบบ่อย**

- ปล่อยให้พจน์ลบ $-dL/L$ หักล้างพจน์บวกในการประมาณกรณีเลวร้ายที่สุด ต้องใช้ค่าสัมบูรณ์
- บวกความคลาดเคลื่อนสัมบูรณ์ที่มีหน่วยต่างกัน (m/s, m², m) แทนที่จะบวกความคลาดเคลื่อนสัมพัทธ์ (%)
- ในอนุพันธ์รวม คิดแค่ $\partial T/\partial t$ แล้วลืมว่าความลึกของหินก็เปลี่ยนตามเวลาด้วย

**ลองทำดู** ใน (a) ความแม่นยำของ $K$ ต้องเป็นเท่าใด ความคลาดเคลื่อนส่วนนั้นจึงจะเท่ากับความคลาดเคลื่อนที่มาจาก $L$

## ตัวอย่างที่ 33 ⭐⭐⭐: ทำไมแมกมาที่มีฟองแก๊สจึงระเบิดใกล้ผิวโลก (อนุพันธ์รวม)

**บริบท** ฟองแก๊สในแมกมาที่กำลังลอยตัวขึ้นเป็นไปตามกฎแก๊สอุดมคติ $V = \dfrac{nRT}{P}$ โดยความดันแมกมา $P = \rho g z$ ($\rho = 2600$ kg/m³, $z$ = ความลึก)

**แคลคูลัส** ผลต่างเชิงอนุพันธ์รวม (total differential) ของ $\ln V = \ln(nR) + \ln T - \ln P$ ให้

$$\frac{dV}{V} = \frac{dT}{T} - \frac{dP}{P}.$$

เปรียบเทียบการลอยขึ้น 100 m (พร้อมการเย็นตัว $dT = -0.5$ K, $T = 1273$ K) เมื่อเริ่มที่ความลึก 2 km และที่ความลึก 200 m

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ปริมาตรฟอง $V=\dfrac{nRT}{P}$ ขึ้นกับสองปริมาณที่เปลี่ยนไป คืออุณหภูมิและความดัน ผลต่างเชิงอนุพันธ์รวม (total differential) รวมการเปลี่ยนแปลงเล็ก ๆ จากแต่ละตัว $dV=\dfrac{\partial V}{\partial T}dT+\dfrac{\partial V}{\partial P}dP$ เราจึงเห็นได้ว่าผลใด ระหว่างการเย็นตัวกับการลดความดัน ที่ควบคุมการขยายตัวของฟองจริง ๆ เมื่อแมกมาลอยขึ้น

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **อนุพันธ์ย่อย** $\dfrac{\partial V}{\partial T}=\dfrac{nR}{P}=\dfrac VT$ และ $\dfrac{\partial V}{\partial P}=-\dfrac{nRT}{P^2}=-\dfrac VP$
2. **ผลต่างเชิงอนุพันธ์รวม หารด้วย $V$:** $\dfrac{dV}{V}=\dfrac{dT}{T}-\dfrac{dP}{P}$ (หรือหา $d$ ของ $\ln V=\ln(nR)+\ln T-\ln P$ ก็ได้) การเปลี่ยนแปลงสัมพัทธ์จึงเป็นผลรวมของพจน์การเย็นตัวและพจน์การลดความดัน
3. **ความดันที่ความลึก** $P=\rho gz=2600\times9.81\times2000\approx51.01$ MPa การลอยขึ้น $100$ m ให้ $dP=-\rho g\,(100)$ ดังนั้น $-\dfrac{dP}{P}=\dfrac{100}{2000}=5\%$
4. **พจน์การเย็นตัว** $dT=-0.5$ K ที่ $T=1273$ K: $\dfrac{dT}{T}=-\dfrac{0.5}{1273}\approx-0.039\%$ เล็กกว่าพจน์ความดันกว่า 100 เท่า
5. **รวมกัน** $\dfrac{dV}{V}\approx5\%-0.039\%\approx4.96\%$ ค่าจริงคือ $\dfrac{1272.5}{1273}\cdot\dfrac{2000}{1900}-1\approx5.22\%$ ค่าประมาณเชิงเส้นจึงใกล้เคียง
6. **ขั้นเดียวกันใกล้ผิวโลก เริ่มที่ 200 m** $-\dfrac{dP}{P}=\dfrac{100}{200}=50\%$ (ผลต่างเชิงอนุพันธ์ $\approx49.96\%$) แต่ค่าจริงประมาณ $99.9\%$ ฟองเกือบขยายเป็นสองเท่า ผลต่างเชิงอนุพันธ์แม่นยำเฉพาะเมื่อการเปลี่ยนแปลงเล็กเมื่อเทียบกับ $P$

```python
nR, Tt, Pp = sp.symbols('nR T P', positive=True)
Vb = nR*Tt/Pp
dV_over_V = sp.simplify((sp.diff(Vb, Tt)*sp.Symbol('dT') + sp.diff(Vb, Pp)*sp.Symbol('dP'))/Vb)
show(r"\frac{dV}{V}", sp.expand(dV_over_V))
rho_m, gv, Tm, dT = 2600, 9.81, 1273, -0.5
for z0 in (2000, 200):
    P0 = rho_m*gv*z0; dP = -rho_m*gv*100
    lin = dT/Tm - dP/P0
    exact = (Tm + dT)/Tm * P0/(P0 + dP) - 1
    print(f"start depth {z0:>5} m: P = {P0/1e6:6.2f} MPa, dV/V (differential) = {lin*100:+6.2f}%, exact = {exact*100:+6.2f}%")
```

```python
Z = np.linspace(150, 5000, 400)
fig, ax = plt.subplots()
ax.plot(Z, (dT/Tm + 100/Z)*100, lw=2, label="differential dV/V")
ax.plot(Z, ((Tm + dT)/Tm*Z/(Z - 100) - 1)*100, "C3--", lw=1.5, label="exact")
ax.set_yscale("log"); ax.invert_xaxis()
ax.set(xlabel="depth at start of 100 m rise (m)", ylabel="bubble volume increase (%)", title="Bubble growth per 100 m of ascent explodes near the surface"); ax.legend(); plt.show()
```

**การตีความ** ที่ความลึก 2 km การลอยขึ้น 100 m ทำให้ฟองโตขึ้นเพียงประมาณ **5 %** แต่ที่ความลึก 200 m ผลต่างเชิงอนุพันธ์ทำนายได้ **50 %** (ค่าจริงประมาณ 100 % เพราะการเปลี่ยนแปลงสัมพัทธ์ที่มากทำให้การประมาณเชิงเส้นใช้ไม่ได้)
การโตอย่างรวดเร็วของฟองในช่วงไม่กี่ร้อยเมตรสุดท้ายทำให้แมกมาแตกเป็นชิ้น นี่คือเหตุผลที่ภูเขาไฟที่มีแก๊สมากปะทุแบบระเบิด พจน์การเย็นตัว $dT/T$ มีค่าน้อยมากเมื่อเทียบกับพจน์ความดัน

**ความหมายของคำตอบ** เนื่องจาก $P$ แปรผันตรงกับความลึก การลอยขึ้น 100 m เท่ากันจึงทำให้ความดันลดลงเชิงสัมพัทธ์ที่ 200 m มากกว่าที่ 2 km ถึงสิบเท่า ฟองจึงโตช้าที่ระดับลึกแล้วโตอย่างรุนแรงในช่วงไม่กี่ร้อยเมตรสุดท้าย เมื่อสัดส่วนแก๊สสูงพอ แมกมาจะแตกเป็นชิ้น และการไหลของลาวาอย่างเงียบ ๆ จะกลายเป็นการปะทุแบบระเบิด

**ข้อผิดพลาดที่พบบ่อย**

- เครื่องหมายของ $dP$ ผิด การเคลื่อนที่**ขึ้น**ทำให้ความดันลดลง $dP<0$ และ $-dP/P>0$ (ฟองขยาย)
- ใช้ °C แทนเคลวินใน $dT/T$ กฎแก๊สอุดมคติต้องใช้อุณหภูมิสัมบูรณ์
- เชื่อค่าประมาณเชิงเส้นแม้ขั้นจะใหญ่ ให้เทียบ $dV/V$ กับค่าจริงเพื่อดูว่าเมื่อใดใช้ไม่ได้

**ลองทำดู** ตามผลต่างเชิงอนุพันธ์ ที่ความลึกเท่าใดการลอยขึ้น 100 m จะทำให้ปริมาตรเพิ่มขึ้น 10 %

**E4** (ตัวอย่างที่ 31) ระดับพลังน้ำคือ $h(x,y)$ อัตราการเปลี่ยนของระดับพลังน้ำในทิศของเวกเตอร์หนึ่งหน่วย $\mathbf u$ คือ…

- (a) $\partial h/\partial x$ เท่านั้น
- (b) $\nabla h\times\mathbf u$
- (c) $|\nabla h|$ ในทุกทิศทาง
- (d) $\nabla h\cdot\mathbf u$ มากสุดเมื่อ $\mathbf u$ ชี้ตาม $\nabla h$

<details><summary><b>เฉลย E4</b></summary>

**(d)** $\nabla h\cdot\mathbf u$ มากสุดเมื่อ $\mathbf u$ ชี้ตาม $\nabla h$ $D_{\mathbf u}h=\nabla h\cdot\mathbf u=|\nabla h|\cos\phi$ มากสุดตามทิศ $\nabla h$ และเป็นศูนย์ตามเส้นชั้น น้ำใต้ดินไหลตาม $-\nabla h$

</details>

**E5** (ตัวอย่างที่ 32) การไหลตามกฎของดาร์ซีคือ $Q=\dfrac{KA\,\Delta h}{L}$ ความคลาดเคลื่อนสัมพัทธ์คือ 10% ใน $K$, 2% ใน $A$, 5% ใน $\Delta h$ และ 1% ใน $L$ จงใช้ผลต่างเชิงอนุพันธ์รวมหาความคลาดเคลื่อนสัมพัทธ์**กรณีแย่ที่สุด**ของ $Q$

**คำตอบของคุณ:** ________ %

<details><summary><b>เฉลย E5</b></summary>

**18 %** $\ln Q=\ln K+\ln A+\ln\Delta h-\ln L$ ดังนั้น $\dfrac{dQ}{Q}=\dfrac{dK}{K}+\dfrac{dA}{A}+\dfrac{d\Delta h}{\Delta h}-\dfrac{dL}{L}$ กรณีแย่ที่สุดรวมขนาด: $10+2+5+1=18\%$ โดย $K$ มีผลมากที่สุด

</details>

**E6** (ตัวอย่างที่ 32) อุณหภูมิของหินคือ $T(z,t)$ ขณะที่หินถูกยกตัวขึ้น $z=z(t)$ จงหา $\dfrac{dT}{dt}$ ตามเส้นทางของหิน

- (a) $\dfrac{\partial T}{\partial z}\dfrac{dz}{dt}+\dfrac{\partial T}{\partial t}$
- (b) $\dfrac{\partial T}{\partial z}\dfrac{dz}{dt}$
- (c) $\dfrac{\partial T}{\partial t}$
- (d) $\dfrac{\partial T}{\partial z}+\dfrac{\partial T}{\partial t}$

<details><summary><b>เฉลย E6</b></summary>

**(a)** $\dfrac{\partial T}{\partial z}\dfrac{dz}{dt}+\dfrac{\partial T}{\partial t}$ อนุพันธ์รวม (หลักเกณฑ์ลูกโซ่สองตัวแปร): หินเย็นลงทั้งเพราะเส้นอุณหภูมิตามความลึกเปลี่ยน **และ**เพราะหินเคลื่อนขึ้นมาที่ตื้นกว่า

</details>

**E7** (ตัวอย่างที่ 33) ฟองแก๊สในแมกมาเป็นไปตาม $V=nRT/P$ ที่ $T$ คงที่ ความดันลดลง 10% จงใช้ผลต่างเชิงอนุพันธ์ประมาณร้อยละการเปลี่ยนแปลงของ $V$

**คำตอบของคุณ:** ________ %

<details><summary><b>เฉลย E7</b></summary>

**10 %** $\dfrac{dV}{V}=\dfrac{dT}{T}-\dfrac{dP}{P}=0-(-10\%)=+10\%$ ใกล้ผิวโลก $P$ มีค่าน้อย ความดันที่ลดลงเท่ากันจึงทำให้ขยายตัว**เชิงสัดส่วน**มาก

</details>

# ส่วน F: เทคนิคการหาปริพันธ์ (Integration Techniques)

⭐⭐⭐ ยาก · ตัวอย่างที่ 34–43 · สัปดาห์ที่ 12–16 ในประมวลรายวิชา

## ตัวอย่างที่ 34 ⭐⭐⭐: ความหนาของตะกอนจากอัตราการสะสมแบบเป็นวัฏจักร (ปฏิยานุพันธ์พร้อมเงื่อนไขเริ่มต้น)

**บริบท** วัฏจักรวงโคจรของโลก (วัฏจักรมิลานโควิช, Milankovitch cycles) ทำให้อัตราการตกตะกอนแกว่งขึ้นลง สมมติว่า

$$r(t) = r_0 + r_1\sin\frac{2\pi t}{P},\qquad r_0 = 2\ \text{cm/kyr},\ r_1 = 0.8\ \text{cm/kyr},\ P = 100\ \text{kyr}.$$

**แคลคูลัส** ความหนา $H(t)$ สอดคล้องกับ $H'(t) = r(t)$, $H(0) = 0$ ปฏิยานุพันธ์ (antiderivative) คือ
$$H(t) = r_0t - \frac{r_1P}{2\pi}\cos\frac{2\pi t}{P} + C,\quad H(0)=0\Rightarrow C = \frac{r_1P}{2\pi}.$$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ถ้ารู้ว่าตะกอนสะสมเร็วเพียงใด $r(t)=H'(t)$ ความหนา $H(t)$ คือปฏิยานุพันธ์ (antiderivative) ของอัตรานั้น ค่าคงตัวของการหาปริพันธ์กำหนดได้จากเงื่อนไขเริ่มต้น $H(0)=0$ อัตราที่เปลี่ยนเป็นจังหวะตามวัฏจักรวงโคจรของโลกทำให้เกิดชั้นตะกอนเป็นจังหวะที่นักธรณีวิทยานับได้เหมือนวงปีของต้นไม้

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ตั้งโจทย์** $H'(t)=r(t)=r_0+r_1\sin\dfrac{2\pi t}{P}$ โดย $r_0=2$ cm/kyr, $r_1=0.8$ cm/kyr, $P=100$ kyr และ $H(0)=0$
2. **ปฏิยานุพันธ์** $\int r_0\,dt=r_0t$ สำหรับไซน์ใช้ $\int\sin(ct)\,dt=-\dfrac1c\cos(ct)$ โดย $c=\dfrac{2\pi}{P}$: $H(t)=r_0t-\dfrac{r_1P}{2\pi}\cos\dfrac{2\pi t}{P}+C$
3. **เงื่อนไขเริ่มต้น** $H(0)=0-\dfrac{r_1P}{2\pi}\cdot1+C=0\Rightarrow C=\dfrac{r_1P}{2\pi}=\dfrac{0.8\times100}{2\pi}\approx12.732$ cm ดังนั้น $H(t)=r_0t+\dfrac{r_1P}{2\pi}\left(1-\cos\dfrac{2\pi t}{P}\right)$
4. **ตรวจคำตอบโดยหาอนุพันธ์** $H'(t)=r_0+\dfrac{r_1P}{2\pi}\cdot\sin\dfrac{2\pi t}{P}\cdot\dfrac{2\pi}{P}=r(t)$ ✓
5. **ตัวเลข** หลังครึ่งวัฏจักร $t=50$ kyr: $H=100+12.732(1-\cos\pi)=100+25.46=125.46$ cm หลังครบหนึ่งวัฏจักร $t=100$: $H=200$ cm หลัง $t=1000$ kyr (10 วัฏจักร): $H=2000$ cm $=20$ m เพราะ $\cos(20\pi)=1$

```python
t = sp.symbols('t', nonnegative=True); C = sp.symbols('C')
r0, r1, Pc = 2, sp.Rational(8, 10), 100
r = r0 + r1*sp.sin(2*sp.pi*t/Pc)
Hgen = sp.integrate(r, t) + C
Cval = sp.solve(sp.Eq(Hgen.subs(t, 0), 0), C)[0]
Ht = sp.simplify(Hgen.subs(C, Cval))
show("H(t)", Ht)
print(f"constant of integration C = {float(Cval):.3f} cm")
for tk in (50, 100, 1000):
    print(f"after {tk:>5} kyr: H = {float(Ht.subs(t, tk)):8.2f} cm = {float(Ht.subs(t, tk))/100:.2f} m")
```

```python
T = np.linspace(0, 500, 1000)
fig, (a1, a2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)
a1.plot(T, sp.lambdify(t, r)(T), "C1", lw=2); a1.set(ylabel="rate r(t) (cm/kyr)", title="Cyclic deposition rate and accumulated thickness")
a2.plot(T, sp.lambdify(t, Ht)(T)/100, lw=2); a2.plot(T, r0*T/100, "k--", lw=1, label="constant rate r₀")
a2.set(xlabel="time (kyr)", ylabel="thickness H (m)"); a2.legend(); plt.tight_layout(); plt.show()
```

**การตีความ** ในหนึ่งล้านปี (10 วัฏจักรเต็ม) พจน์โคไซน์กลับมาที่ค่าเริ่มต้น ความหนาจึงเท่ากับ $r_0\times1000$ kyr = **20 m** พอดี วัฏจักรปรากฏเป็นชั้นหนาและชั้นบางสลับกัน
นักธรณีวิทยาอ่านย้อนกลับ (*การลำดับชั้นหินเชิงวัฏจักร, cyclostratigraphy*): การนับวัฏจักรเหล่านี้ในแท่งตัวอย่างหินเจาะช่วยหาอายุตะกอนได้ละเอียดถึงระดับหลายหมื่นปี

**ความหมายของคำตอบ** เมื่อครบวัฏจักรเต็ม ส่วนไซน์จะเฉลี่ยหักล้างกัน ความหนาจึงเท่ากับอัตราเฉลี่ยคูณเวลา คือ 20 m ต่อล้านปี ภายในแต่ละวัฏจักรชั้นตะกอนจะโตเร็วสลับช้า ปรากฏเป็นชั้นหินสลับกัน การนับจังหวะ 100 kyr (ความรีของวงโคจร) 41 kyr (ความเอียงของแกน) หรือ 21 kyr (การส่ายของแกน) ให้นาฬิกาดาราศาสตร์สำหรับหาอายุหินตะกอน

**ข้อผิดพลาดที่พบบ่อย**

- ลืมค่าคงตัว $C$ หรือให้ $C=0$ ซึ่งทำให้ $H(0)=-12.7$ cm เป็นไปไม่ได้
- หาปริพันธ์ $\sin\dfrac{2\pi t}{P}$ เป็น $-\cos\dfrac{2\pi t}{P}$ โดยลืมตัวคูณ $\dfrac{P}{2\pi}$ จากฟังก์ชันภายใน
- หน่วยปนกัน อัตราอยู่ในหน่วย cm ต่อ kyr เวลาจึงต้องเป็น kyr และคำตอบจะออกมาเป็น cm

**ลองทำดู** ความหนาหลัง 125 kyr (หนึ่งวัฏจักรกับอีกหนึ่งในสี่) เป็นเท่าใด ทำไมจึงมากกว่า $r_0\times125$

## ตัวอย่างที่ 35 ⭐⭐⭐: ความร้อนจากหินแกรนิตที่มีกัมมันตภาพรังสี (การหาปริพันธ์โดยการแทนค่า (integration by substitution))

**บริบท** หินแกรนิตมี U, Th และ K ซึ่งผลิตความร้อน อัตราการผลิตความร้อนจากกัมมันตภาพรังสี (radiogenic heat production) ลดลงตามความลึก:
$A(z) = A_0 e^{-z/h_r}$ โดย $A_0 = 2.5\ \mu\text{W/m}^3$ และ $h_r = 10$ km สมดุลความร้อนให้ $\dfrac{dq}{dz} = -A(z)$ ดังนั้นฟลักซ์ความร้อนที่ผิวโลก (surface heat flow) คือ

$$q_s = q_m + \int_0^{\infty} A_0 e^{-z/h_r}\,dz,$$

โดยที่ $q_m = 30$ mW/m² คือความร้อนที่ไหลขึ้นมาจากเนื้อโลก (mantle)

**แคลคูลัส** แทนค่า $u = -z/h_r$, $du = -dz/h_r$: $\displaystyle\int A_0e^{-z/h_r}dz = -A_0h_r\int e^{u}\,du = -A_0h_r e^{-z/h_r}+C$
เมื่อหาปริพันธ์**อีกครั้ง**ร่วมกับกฎของฟูรีเย (Fourier's law) $k\,T'(z) = q(z)$ จะได้โปรไฟล์อุณหภูมิทั้งหมด

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** การผลิตความร้อนในหินแกรนิตลดลงแบบเลขชี้กำลังตามความลึก และความร้อนที่ไหลออกที่ผิวโลกคือ **ปริพันธ์** ของความร้อนทั้งหมดที่ผลิตอยู่ข้างใต้ **การเปลี่ยนตัวแปร (substitution)** อย่างง่าย $u=-z/h_r$ เปลี่ยนปริพันธ์ของฟังก์ชันเลขชี้กำลังให้เป็น $\int e^u\,du$ การหาปริพันธ์ครั้งที่สองโดยใช้กฎของฟูรีเย (Fourier's law) ให้อุณหภูมิที่ทุกความลึก

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **เปลี่ยนตัวแปร** ให้ $u=-z/h_r$ ดังนั้น $du=-dz/h_r$ และ $dz=-h_r\,du$ จะได้ $$\int A_0e^{-z/h_r}dz=-A_0h_r\int e^u\,du=-A_0h_re^{-z/h_r}+C.$$ ตรวจสอบ: หาอนุพันธ์ของ $-A_0h_re^{-z/h_r}$ จะได้ $A_0e^{-z/h_r}$ กลับคืนมา
2. **ส่วนที่มาจากเปลือกโลกทั้งหมด (ปริพันธ์ไม่ตรงแบบ)** $$\int_0^\infty A_0e^{-z/h_r}dz=\lim_{b\to\infty}\left[-A_0h_re^{-z/h_r}\right]_0^b=0-(-A_0h_r)=A_0h_r.$$ เมื่อ $A_0=2.5\times10^{-6}$ W/m³ และ $h_r=10^4$ m: $A_0h_r=0.025$ W/m² $=25$ mW/m²
3. **ฟลักซ์ความร้อนที่ผิว** บวกความร้อนที่มาจากเนื้อโลก: $q_s=q_m+A_0h_r=30+25=55$ mW/m² ที่ความลึก $z$ มีเพียงความร้อนที่ผลิตใต้ $z$ ลงไปที่ยังไหลอยู่ ดังนั้น $q(z)=q_m+A_0h_re^{-z/h_r}$
4. **หาปริพันธ์ครั้งที่สอง (กฎของฟูรีเย $kT'(z)=q(z)$)** $$T(z)=T_s+\frac1k\int_0^z q(\zeta)\,d\zeta=T_s+\frac{q_mz}{k}+\frac{A_0h_r^2}{k}\left(1-e^{-z/h_r}\right),$$ โดยใช้การเปลี่ยนตัวแปรแบบเดิมกับพจน์เลขชี้กำลัง
5. **ค่าที่ความลึก 30 km** ($T_s=15$ °C, $k=2.5$ W/m/K, $z=3\times10^4$ m): $\dfrac{q_mz}{k}=\dfrac{0.03\times30000}{2.5}=360$ °C และ $\dfrac{A_0h_r^2}{k}(1-e^{-3})=100\times0.950=95.0$ °C ดังนั้น $T=15+360+95=470$ °C ที่ความลึก 10 km สูตรเดียวกันให้ประมาณ 198 °C

```python
z, zp = sp.symbols('z zeta', nonnegative=True)
A0, hr, qm, k, Ts = sp.symbols('A_0 h_r q_m k T_s', positive=True)

u = sp.symbols('u')
indef = sp.integrate(A0*sp.exp(u)*(-hr), u).subs(u, -z/hr)          # substitution done by hand
show(r"\int A_0e^{-z/h_r}\,dz", indef)
crust = sp.integrate(A0*sp.exp(-z/hr), (z, 0, sp.oo))
show(r"\int_0^\infty A_0e^{-z/h_r}dz", crust)

# q(z) = heat flow at depth z, then T(z) = Ts + ∫0^z q/k
F = sp.integrate(A0*sp.exp(-zp/hr), zp)                            # antiderivative
qz = qm + (sp.limit(F, zp, sp.oo) - F.subs(zp, z))                   # q_m + ∫_z^∞ A dζ
Tz = Ts + sp.integrate(qz.subs(z, zp)/k, (zp, 0, z))
show("q(z)", sp.simplify(qz)); show("T(z)", sp.simplify(Tz))

num = {A0: 2.5e-6, hr: 10e3, qm: 30e-3, k: 2.5, Ts: 15}
print(f"surface heat flow q_s = {float(qz.subs(num).subs(z, 0))*1e3:.1f} mW/m² "
      f"(of which {float(crust.subs(num))*1e3:.1f} mW/m² is from the granite)")
print(f"temperature at 30 km = {float(Tz.subs(num).subs(z, 30e3)):.0f} °C")
```

```python
zz = np.linspace(0, 40e3, 300)
Tf = sp.lambdify(z, Tz.subs(num)); Tlin = 15 + 30e-3/2.5*zz
fig, ax = plt.subplots(figsize=(5.5, 5))
ax.plot(Tf(zz), zz/1e3, lw=2, label="with radiogenic heat")
ax.plot(Tlin, zz/1e3, "--", label="mantle heat only")
ax.invert_yaxis(); ax.set(xlabel="T (°C)", ylabel="depth (km)", title="Geotherm from two integrations"); ax.legend(); plt.show()
```

**การตีความ** กัมมันตภาพรังสีในเปลือกโลกเพิ่มฟลักซ์ความร้อน $A_0h_r = 25$ mW/m² ทำให้ฟลักซ์ความร้อนที่ผิวเพิ่มขึ้นเกือบเท่าตัว นี่คือเหตุผลที่พื้นที่หินแกรนิต เช่น แนวหินแกรนิตภาคเหนือของไทย มีฟลักซ์ความร้อนสูงและมีน้ำพุร้อน (เช่น ฝาง สันกำแพง) ซึ่งเป็นแหล่งพลังงานความร้อนใต้พิภพ (geothermal energy)

**ความหมายของคำตอบ** กัมมันตภาพรังสีในเปลือกโลกตอนบนให้ความร้อน 25 จาก 55 mW/m² ที่มาถึงผิวโลก ทำให้ฟลักซ์ความร้อนเพิ่มเกือบเท่าตัว และทำให้เปลือกโลกที่ความลึก 30 km ร้อนกว่ากรณีมีแค่ความร้อนจากเนื้อโลก (375 °C) ประมาณ 95 °C นี่คือเหตุที่แนวหินแกรนิต เช่น ในภาคเหนือของไทย มีบ่อน้ำพุร้อน (ฝาง สันกำแพง) และมีศักยภาพพลังงานความร้อนใต้พิภพ ตรวจสอบ: ฟลักซ์ความร้อนของทวีปโดยทั่วไปอยู่ที่ 50–70 mW/m² ค่า 55 mW/m² จึงสมจริง

**ข้อผิดพลาดที่พบบ่อย**

- ลืมตัวประกอบ $-h_r$ จาก $dz=-h_r\,du$ ทำให้ได้ $A_0$ แทน $A_0h_r$
- ใช้หน่วยปนกัน $h_r$ ต้องเป็นเมตร (ไม่ใช่ km) เมื่อ $A_0$ มีหน่วย W/m³ มิฉะนั้นฟลักซ์ความร้อนจะผิดไป 1000 เท่า
- เขียนว่า $\lim_{b\to\infty}e^{-b/h_r}=1$ ซึ่งผิด ค่าที่ถูกคือ $0$ เพราะเลขชี้กำลังเข้าใกล้ $-\infty$

**ลองทำดู** เปลี่ยน $h_r$ เป็น 15 km หินที่ความลึก 30 km จะร้อนขึ้นเท่าใด

## ตัวอย่างที่ 36 ⭐⭐⭐: เวลาเดินทางของคลื่นไหวสะเทือนเมื่อความเร็วเพิ่มตามความลึก (การแทนค่า)

**บริบท** ในแอ่งตะกอน การอัดตัวทำให้ความเร็วคลื่น P เพิ่มขึ้นตามความลึก: $v(z) = v_0 + kz$ โดย $v_0 = 1.8$ km/s และ $k = 0.6$ s⁻¹
คลื่นที่เคลื่อนที่ในแนวดิ่งใช้เวลา $dt = dz/v$ สำหรับแต่ละชั้นบาง $dz$

**แคลคูลัส** ใช้การแทนค่า (substitution) $u = v_0 + kz$, $du = k\,dz$:

$$T(z) = \int_0^z\frac{d\zeta}{v_0+k\zeta} = \frac1k\int_{v_0}^{v_0+kz}\frac{du}{u} = \frac1k\ln\!\left(1+\frac{kz}{v_0}\right).$$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** คลื่นไหวสะเทือนที่ผ่านชั้นบาง $d\zeta$ ด้วยความเร็ว $v$ ใช้เวลา $d\zeta/v$ เวลาเดินทางทั้งหมดจึงเป็นปริพันธ์ของความช้า (slowness) $1/v$ ตามความลึก เมื่อความเร็วเพิ่มเชิงเส้นตามความลึก การเปลี่ยนตัวแปร (substitution) $u=v_0+k\zeta$ จะเปลี่ยนปริพันธ์เป็น $\int du/u$ ซึ่งได้ลอการิทึม

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ตั้งปริพันธ์** $t(z)=\displaystyle\int_0^z\frac{d\zeta}{v_0+k\zeta}$ โดย $v_0=1.8$ km/s, $k=0.6$ s⁻¹, $z=3$ km
2. **เปลี่ยนตัวแปร** $u=v_0+k\zeta\Rightarrow du=k\,d\zeta\Rightarrow d\zeta=\dfrac{du}{k}$ เปลี่ยนขอบเขตด้วย: $\zeta=0\to u=v_0$ และ $\zeta=z\to u=v_0+kz$
3. **หาปริพันธ์** $t(z)=\dfrac1k\displaystyle\int_{v_0}^{v_0+kz}\frac{du}{u}=\frac1k\Big[\ln u\Big]_{v_0}^{v_0+kz}=\frac1k\ln\!\left(1+\frac{kz}{v_0}\right)$
4. **แทนค่า** $\dfrac{kz}{v_0}=\dfrac{0.6\times3}{1.8}=1$ ดังนั้น $t=\dfrac{\ln2}{0.6}\approx1.155$ s ขาเดียว และเวลาไป-กลับ $\text{TWT}=2t\approx2.310$ s ถ้าความเร็วคงที่ $1.8$ km/s จะได้ $2\times3/1.8\approx3.333$ s
5. **หาความลึกย้อนกลับ** จาก $kt=\ln(1+kz/v_0)$ ได้ $z=\dfrac{v_0}{k}\left(e^{kt}-1\right)$ ตัวสะท้อนที่ TWT $2.0$ s มี $t=1$ s ดังนั้น $z=3(e^{0.6}-1)\approx2.466$ km
6. **ตรวจลิมิต $k\to0$** ใช้ $\ln(1+x)\approx x$ เมื่อ $x$ เล็ก (หรือโลปีตาล) ได้ $\dfrac1k\ln(1+kz/v_0)\to\dfrac{z}{v_0}$ ตรงกับสูตรความเร็วคงที่ ✓

```python
z, zeta = sp.symbols('z zeta', positive=True); v0, k = sp.Rational(18, 10), sp.Rational(6, 10)
Tz = sp.integrate(1/(v0 + k*zeta), (zeta, 0, z))
show("T(z)", sp.simplify(Tz))
one_way = float(Tz.subs(z, 3))
print(f"one-way time to 3 km = {one_way:.4f} s, two-way time (reflection) = {2*one_way:.3f} s")
print(f"if the velocity stayed at v0 = 1.8 km/s: two-way time = {2*3/1.8:.3f} s")
# invert: depth of a reflector seen at two-way time 2.0 s
zz = sp.solve(sp.Eq(2*Tz, 2.0), z)[0]
print(f"a reflector at two-way time 2.0 s is at depth {float(zz):.3f} km")
```

```python
Z = np.linspace(0, 5, 200)
fig, ax = plt.subplots(figsize=(5.5, 5))
ax.plot(2*sp.lambdify(z, Tz)(Z), Z, lw=2, label="v = v₀ + kz")
ax.plot(2*Z/1.8, Z, "--", label="constant v₀ = 1.8 km/s")
ax.invert_yaxis(); ax.set(xlabel="two-way travel time (s)", ylabel="depth (km)", title="Time–depth conversion"); ax.legend(); plt.show()
```

**การตีความ** เนื่องจากความเร็วเพิ่มตามความลึก เวลาเดินทางไป-กลับถึงความลึก 3 km จึงเป็น **2.31 s** ซึ่งน้อยกว่า 3.33 s ในกรณีความเร็วคงที่มาก
ภาพการสะท้อนคลื่นไหวสะเทือนถูกบันทึกในรูปของ *เวลา* นักธรณีวิทยาปิโตรเลียมในอ่าวไทยใช้การแปลงนี้เปลี่ยนเป็น *ความลึก* ก่อนการขุดเจาะ หากใช้แบบจำลองความเร็วผิด ตำแหน่งแหล่งกักเก็บอาจคลาดไปหลายร้อยเมตร

**ความหมายของคำตอบ** เพราะหินมีความเร็วสูงขึ้นตามความลึก คลื่นจึงไปถึง 3 km ด้วยเวลาไป-กลับ 2.31 s แทนที่จะเป็น 3.33 s ความเร็วเฉลี่ยคือ $3/1.155\approx2.60$ km/s อยู่ระหว่าง 1.8 km/s ที่ผิวกับ 3.6 km/s ที่ตัวสะท้อน การแปลงเวลาสะท้อนเป็นความลึกด้วยความเร็วคงที่จะวางเป้าหมายลึกผิดตำแหน่ง เป็นความผิดพลาดที่มีราคาแพงเมื่อเจาะหาน้ำมัน แก๊ส หรือน้ำร้อนใต้พิภพ

**ข้อผิดพลาดที่พบบ่อย**

- ลืมตัวคูณ $\dfrac1k$ ที่มาจาก $d\zeta=du/k$
- เปลี่ยนตัวแปรแต่ยังใช้ขอบเขตเดิม $0$ และ $z$ ต้องเปลี่ยนขอบเขตเป็นค่าของ $u$ หรือแทนกลับก่อนหาค่า
- เทียบเวลาขาเดียวกับเวลาไป-กลับบนภาพตัดขวางไหวสะเทือน ข้อมูลการสะท้อนบันทึกเป็นเวลาไป-กลับ

**ลองทำดู** แก้สมการ $T(z)$ หา $z$ เพื่อให้ได้ความลึกเป็นฟังก์ชันของเวลาเที่ยวเดียว $z(T) = \dfrac{v_0}{k}\left(e^{kT}-1\right)$ และตรวจสอบว่าได้ 3 km สำหรับเวลาข้างต้น

## ตัวอย่างที่ 37 ⭐⭐⭐: จุดสูงสุดของฮับเบิร์ต (Hubbert's peak): การผลิตน้ำมันจากแหล่งที่มีจำกัด (เศษส่วนย่อย (partial fractions))

**บริบท** M. King Hubbert จำลองปริมาณการผลิตสะสม $Q(t)$ ของแหล่งน้ำมันที่มีปริมาณสำรองที่ผลิตได้ทั้งหมด $Q_\infty$ ด้วยสมการลอจิสติก (logistic equation)

$$\frac{dQ}{dt} = rQ\left(1-\frac{Q}{Q_\infty}\right).$$

**แคลคูลัส** แยกตัวแปร: $\displaystyle\int\frac{dQ}{Q\,(1-Q/Q_\infty)} = \int r\,dt$
ด้านซ้ายต้องใช้**เศษส่วนย่อย**: $\dfrac{1}{Q(1-Q/Q_\infty)} = \dfrac1Q + \dfrac{1}{Q_\infty - Q}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** สมการลอจิสติก $dQ/dt=rQ(1-Q/Q_\infty)$ บอกว่าการผลิตเพิ่มขึ้นตามปริมาณที่ค้นพบ แต่จะช้าลงเมื่อทรัพยากรใกล้หมด ในการแก้สมการเราต้อง **แยกตัวแปร** และปริพันธ์ที่ได้ $\int\frac{dQ}{Q(1-Q/Q_\infty)}$ จะหาได้ก็ต่อเมื่อใช้ **การแยกเศษส่วนย่อย (partial fractions)** แยกออกเป็นลอการิทึมอย่างง่ายสองพจน์

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **แยกตัวแปร** ย้าย $Q$ ทั้งหมดไปทางซ้าย และ $t$ ไปทางขวา: $$\int\frac{dQ}{Q(1-Q/Q_\infty)}=\int r\,dt.$$
2. **แยกเศษส่วนย่อย** เขียนใหม่เป็น $\dfrac{1}{Q(1-Q/Q_\infty)}=\dfrac{Q_\infty}{Q(Q_\infty-Q)}=\dfrac{A}{Q}+\dfrac{B}{Q_\infty-Q}$ คูณตลอดได้ $Q_\infty=A(Q_\infty-Q)+BQ$ แทน $Q=0$: $A=1$ แทน $Q=Q_\infty$: $B=1$ ดังนั้นตัวถูกหาปริพันธ์คือ $\dfrac1Q+\dfrac{1}{Q_\infty-Q}$
3. **หาปริพันธ์แล้วแก้หา $Q$** $\ln Q-\ln(Q_\infty-Q)=rt+C$ คือ $\ln\dfrac{Q}{Q_\infty-Q}=rt+C$ ยกกำลัง $e$ ทั้งสองข้างแล้วจัดรูป: $$Q(t)=\frac{Q_\infty}{1+e^{-(rt+C)}}.$$
4. **ใช้ค่าเริ่มต้น** วัด $t$ เป็นจำนวนปีหลัง ค.ศ. 1900 ซึ่งมี $Q=0.5$ Gbbl โดย $Q_\infty=230$ Gbbl และ $r=0.065$ ต่อปี: $C=\ln\dfrac{0.5}{229.5}=-6.129$
5. **หาจุดสูงสุด** อัตราการผลิต $P(Q)=rQ(1-Q/Q_\infty)$ เป็นพาราโบลาคว่ำในตัวแปร $Q$ จาก $P'(Q)=r(1-2Q/Q_\infty)=0$ ได้ $Q=Q_\infty/2$ ซึ่ง $P=rQ_\infty/4=0.065\times230/4=3.74$ Gbbl ต่อปี และ $Q=Q_\infty/2$ เกิดเมื่อ $rt+C=0$: $t=6.129/0.065=94.3$ คือประมาณ **ค.ศ. 1994**
6. **อ่านค่าในอนาคต** ใน ค.ศ. 2025 ($t=125$): $Q=230/(1+e^{-(8.125-6.129)})\approx202$ Gbbl ถูกผลิตไปแล้ว จึงเหลือเพียงประมาณ 28 Gbbl

```python
Q, Qi, r, t, C = sp.symbols('Q Q_infty r t C', positive=True)
integrand = 1/(Q*(1 - Q/Qi))
pf = sp.apart(integrand, Q)
show(r"\frac{1}{Q(1-Q/Q_\infty)}", pf)
lhs = sp.integrate(pf, Q)
show(r"\int \frac{dQ}{Q(1-Q/Q_\infty)}", sp.simplify(lhs))

# Solve ln(Q/(Qi-Q)) = r t + C  for Q(t)
Qt = sp.solve(sp.Eq(sp.log(Q/(Qi - Q)), r*t + C), Q)[0]
show("Q(t)", sp.simplify(Qt))

# check with SymPy's ODE solver
Qf = sp.Function('Q')
print("dsolve agrees:", sp.dsolve(sp.Eq(Qf(t).diff(t), r*Qf(t)*(1 - Qf(t)/Qi))).rhs.has(sp.exp))
```

```python
Qinf, rv, Q1900 = 230.0, 0.065, 0.5                  # Gbbl, 1/yr, cumulative in 1900 (illustrative)
Cv = np.log(Q1900/(Qinf - Q1900))
years = np.arange(1900, 2101)
Qcum = Qinf/(1 + np.exp(-(rv*(years - 1900) + Cv)))
P = rv*Qcum*(1 - Qcum/Qinf)                          # annual production dQ/dt
peak = years[np.argmax(P)]
print(f"peak production year ≈ {peak}, when Q = Q∞/2 = {Qinf/2:.0f} Gbbl; peak rate = r·Q∞/4 = {rv*Qinf/4:.2f} Gbbl/yr")
fig, ax1 = plt.subplots(); ax2 = ax1.twinx()
ax1.plot(years, P, "C3", lw=2, label="production dQ/dt"); ax2.plot(years, Qcum, "C0", lw=1.5, label="cumulative Q")
ax1.axvline(peak, c="gray", ls="--"); ax1.set(xlabel="year", ylabel="Gbbl / yr"); ax2.set_ylabel("cumulative Gbbl")
ax1.set_title("Hubbert curve (illustrative oil province)"); ax2.spines["right"].set_visible(True); plt.show()
```

**การตีความ** เศษส่วนย่อยเปลี่ยนปริพันธ์ที่ยากให้เป็นลอการิทึมสองพจน์ ซึ่งให้เส้นโค้งลอจิสติกรูปตัว S อัตราการผลิต (อนุพันธ์ของมัน) คือ**เส้นโค้งฮับเบิร์ต (Hubbert curve)** รูประฆัง ซึ่งมีค่าสูงสุดเมื่อผลิตไปแล้วครึ่งหนึ่งของปริมาณสำรอง
Hubbert ใช้แบบจำลองนี้ในปี 1956 ทำนายจุดสูงสุดการผลิตน้ำมันของสหรัฐฯ ในปี 1970 แบบจำลองเดียวกันนี้ยังใช้กับทรัพยากรแร่และการสูบน้ำบาดาลเกินสมดุล

**ความหมายของคำตอบ** ผลผลิตสะสมมีรูปเส้นโค้ง S และอัตราการผลิตต่อปี (อนุพันธ์ของมัน) เป็น **เส้นโค้งฮับเบิร์ต (Hubbert curve)** รูประฆังคว่ำ ซึ่งสูงสุดเมื่อผลิตทรัพยากรไปแล้วครึ่งหนึ่งพอดี ฮับเบิร์ตใช้แบบจำลองนี้ในปี 1956 ทำนายจุดสูงสุดการผลิตน้ำมันของสหรัฐฯ ในปี 1970 และแบบจำลองเดียวกันใช้กับแหล่งแร่และการสูบน้ำบาดาลเกินสมดุล ตรวจสอบ: อัตราสูงสุด $rQ_\infty/4$ เพิ่มขึ้นตามทั้งขนาดทรัพยากรและอัตราการเติบโต ซึ่งสมเหตุสมผล

**ข้อผิดพลาดที่พบบ่อย**

- หาปริพันธ์ของ $\dfrac{1}{Q_\infty-Q}$ ได้ $+\ln(Q_\infty-Q)$ ซึ่งผิด เนื่องจากมีเครื่องหมายลบหน้า $Q$ คำตอบที่ถูกคือ $-\ln(Q_\infty-Q)$
- ลืมค่าคงที่ $C$ ทำให้ไม่สามารถปรับเส้นโค้งให้ตรงกับค่าเริ่มต้นใน ค.ศ. 1900
- สับสนระหว่างผลผลิตสะสม $Q$ (Gbbl) กับอัตราการผลิต $dQ/dt$ (Gbbl ต่อปี)

**ลองทำดู** จงแสดงโดยการหาอนุพันธ์ของ $P(Q) = rQ(1-Q/Q_\infty)$ ว่าอัตราการผลิตสูงสุดเกิดขึ้นที่ $Q = Q_\infty/2$

## ตัวอย่างที่ 38 ⭐⭐⭐: การทำให้น้ำเหมืองที่เป็นกรดเป็นกลางด้วยหินปูน (เศษส่วนย่อย)

**บริบท** น้ำทิ้งที่เป็นกรด (ความเข้มข้นกรด $a = 2.0$ mmol/L) ไหลผ่านหินปูนบดซึ่งให้คาร์บอเนต $b = 3.0$ mmol/L หากทำปฏิกิริยาไปแล้ว $x(t)$ mmol/L กฎอัตราอันดับสองให้

$$\frac{dx}{dt} = k(a-x)(b-x),\qquad k = 0.5\ \text{L/(mmol·h)},\ x(0) = 0.$$

**แคลคูลัส** แยกตัวแปรและใช้เศษส่วนย่อย (partial fractions):
$$\frac{1}{(a-x)(b-x)} = \frac{1}{b-a}\left(\frac{1}{a-x}-\frac{1}{b-x}\right)\ \Rightarrow\ \int\frac{dx}{(a-x)(b-x)} = \frac{1}{b-a}\ln\frac{b-x}{a-x}+C.$$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ในปฏิกิริยาอันดับสอง อัตราขึ้นกับปริมาณสองตัวที่ลดลง $\dfrac{dx}{dt}=k(a-x)(b-x)$ เมื่อแยกตัวแปรจะเหลือ $\displaystyle\int\frac{dx}{(a-x)(b-x)}$ ซึ่งเป็นฟังก์ชันตรรกยะที่การแยกเศษส่วนย่อย (partial fractions) แยกออกเป็นลอการิทึมง่าย ๆ สองตัว ผลลัพธ์บอกว่าการบำบัดใช้เวลานานเท่าใด

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **แยกตัวแปร** $\dfrac{dx}{(a-x)(b-x)}=k\,dt$ โดยกรด $a=2$ mmol/L คาร์บอเนต $b=3$ mmol/L $k=0.5$ L/(mmol·h) และ $x(0)=0$
2. **แยกเศษส่วนย่อย** เขียน $\dfrac{1}{(a-x)(b-x)}=\dfrac{A}{a-x}+\dfrac{B}{b-x}$ จะได้ $1=A(b-x)+B(a-x)$ แทน $x=a$: $A=\dfrac1{b-a}$ แทน $x=b$: $B=-\dfrac1{b-a}$
3. **หาปริพันธ์ทีละส่วน** $\displaystyle\int\frac{dx}{a-x}=-\ln(a-x)$ ดังนั้น $\dfrac1{b-a}\left[-\ln(a-x)+\ln(b-x)\right]=\dfrac1{b-a}\ln\dfrac{b-x}{a-x}=kt+C$
4. **ค่าคงตัวจาก $x(0)=0$:** $C=\dfrac1{b-a}\ln\dfrac ba$ นำมาลบกันแล้วแก้หา $t$: $t=\dfrac{1}{(b-a)k}\ln\dfrac{a(b-x)}{b(a-x)}$
5. **เวลาที่กรดถูกสะเทินไป 90%** $x=0.9a=1.8$: $t=\dfrac{1}{1\times0.5}\ln\dfrac{2\times1.2}{3\times0.2}=2\ln4=\ln16\approx2.77$ ชั่วโมง
6. **แก้หา $x(t)$ และตรวจสอบ** ให้ $E=e^{(b-a)kt}$ จะได้ $x(t)=\dfrac{ab(E-1)}{bE-a}$ หลัง 2 ชั่วโมง $E=e$ และ $x\approx1.675$ เหลือกรดประมาณ $0.325$ mmol/L เมื่อ $t\to\infty$ จะได้ $x\to a=2$ กรดถูกใช้หมดเพราะมีคาร์บอเนตมากกว่ากรด

```python
x, t = sp.symbols('x t', positive=True); a_, b_, k_ = sp.Rational(2), sp.Rational(3), sp.Rational(1, 2)
show(r"\text{partial fractions}", sp.apart(1/((a_ - x)*(b_ - x)), x))
lhs = sp.log((b_ - x)/(a_ - x))/(b_ - a_)
C = lhs.subs(x, 0)                                   # x(0) = 0
xt = sp.solve(sp.Eq(lhs, k_*t + C), x)[0]
show("x(t)", sp.simplify(xt))
check = sp.simplify(sp.diff(xt, t) - k_*(a_ - xt)*(b_ - xt))
print("x(t) satisfies the ODE:", check == 0)
t90 = sp.solve(sp.Eq(lhs.subs(x, sp.Rational(9, 5)), k_*t + C), t)[0]
print(f"time to neutralise 90% of the acid (x = 1.8): t = {sp.nsimplify(t90)} = {float(t90):.3f} h")
```

```python
T = np.linspace(0, 8, 300); xf = sp.lambdify(t, xt)
fig, ax = plt.subplots()
ax.plot(T, 2.0 - xf(T), lw=2, label="acid remaining a − x")
ax.plot(T, 3.0 - xf(T), lw=2, label="carbonate remaining b − x")
ax.axvline(float(t90), c="gray", ls=":"); ax.text(float(t90) + 0.1, 2.5, f"90% neutralised\nat {float(t90):.2f} h")
ax.set(xlabel="time (h)", ylabel="concentration (mmol/L)", title="Second-order neutralisation"); ax.legend(); plt.show()
```

**การตีความ** เศษส่วนย่อยให้ $t = \dfrac{1}{k(b-a)}\ln\dfrac{a(b-x)}{b(a-x)}$ การทำให้กรดเป็นกลาง 90 % ใช้เวลา $\ln 4 / 0.5 \approx$ **2.77 h**
วิศวกรใช้เวลาพักน้ำ (residence time) นี้ในการออกแบบขนาดรางระบายน้ำหินปูนที่เหมือง รางต้องเก็บน้ำไว้ได้อย่างน้อยนานเท่านั้น

**ความหมายของคำตอบ** บ่อบำบัดต้องกักน้ำไว้อย่างน้อยประมาณ 2.8 ชั่วโมงเพื่อกำจัดกรด 90% ที่อัตรานี้ ปฏิกิริยาช้าลงเมื่อสารตั้งต้นทั้งสองถูกใช้ไป ไม่กี่เปอร์เซ็นต์สุดท้ายจึงใช้เวลานานกว่าช่วงแรกมาก ถ้าปริมาณหินปูน $b$ น้อยกว่า $0.9a$ เป้าหมาย 90% จะไม่มีวันถึง ซึ่งเป็นการตรวจสอบการออกแบบที่ได้จากสูตรโดยตรง

**ข้อผิดพลาดที่พบบ่อย**

- เครื่องหมายผิดใน $\displaystyle\int\frac{dx}{a-x}$ ค่าที่ถูกคือ $-\ln|a-x|$ ไม่ใช่ $+\ln|a-x|$ (อนุพันธ์ภายในคือ $-1$)
- ลืมค่าคงตัวของการหาปริพันธ์ ทำให้เวลาผิดเพราะไม่สอดคล้องกับ $x(0)=0$
- ใช้สูตรนี้เมื่อ $a=b$ ซึ่งทำให้ $b-a=0$ และใช้การแยกเศษส่วนย่อยไม่ได้ ปริพันธ์จะกลายเป็น $\int(a-x)^{-2}dx$ แทน

**ลองทำดู** หากใช้หินปูนมากขึ้นเป็นสองเท่า ($b = 6$ mmol/L) การทำให้เป็นกลาง 90 % จะใช้เวลาเท่าใด

## ตัวอย่างที่ 39 ⭐⭐⭐: ค่าผิดปกติของความโน้มถ่วงจากสายแร่ที่ฝังอยู่ (การแทนค่าด้วยฟังก์ชันตรีโกณมิติ (trigonometric substitution))

**บริบท** สายแร่ที่มีความหนาแน่นสูงอาจจำลองเป็นทรงกลมที่ฝังอยู่ซึ่งมีมวลส่วนเกิน $\Delta M$ ที่ความลึก $z$ ตามแนวสำรวจจะเกิดค่าผิดปกติของความโน้มถ่วงในแนวดิ่ง (gravity anomaly)

$$\Delta g(x) = \frac{G\,\Delta M\,z}{(x^2+z^2)^{3/2}}.$$

**คำถาม**
1. หาปริพันธ์ของ $\Delta g$ ตามแนวสำรวจ ใช้**การแทนค่าด้วยฟังก์ชันตรีโกณมิติ** $x = z\tan\theta$, $dx = z\sec^2\theta\,d\theta$, $x^2 + z^2 = z^2\sec^2\theta$:

$$\int\frac{dx}{(x^2+z^2)^{3/2}} = \int\frac{z\sec^2\theta}{z^3\sec^3\theta}d\theta = \frac{1}{z^2}\int\cos\theta\,d\theta = \frac{\sin\theta}{z^2}+C = \frac{x}{z^2\sqrt{x^2+z^2}}+C.$$

2. ใช้ความกว้างครึ่งหนึ่ง (half-width) $x_{1/2}$ (ตำแหน่งที่ $\Delta g$ ลดลงเหลือครึ่งหนึ่งของค่าสูงสุด) ประมาณความลึก ตามกฎของนักธรณีฟิสิกส์ $z \approx 1.305\,x_{1/2}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ทรงกลมหนาแน่นที่ฝังอยู่ใต้ดินทำให้ค่าแรงโน้มถ่วงเพิ่มขึ้นตาม $(x^2+z^2)^{-3/2}$ ปริพันธ์ที่มี $x^2+z^2$ แก้ได้สวยงามด้วย **การแทนค่าด้วยฟังก์ชันตรีโกณมิติ (trigonometric substitution)** $x=z\tan\theta$ เพราะ $1+\tan^2\theta=\sec^2\theta$ ช่วยกำจัดรากที่สอง พื้นที่ใต้กราฟความผิดปกติจึงบอกมวลที่ฝังอยู่ และความกว้างของกราฟบอกความลึก

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **มวลและค่าสูงสุด** $\Delta M=\tfrac43\pi R^3\Delta\rho=\tfrac43\pi(150)^3(1500)=2.12\times10^{10}$ kg (ประมาณ 21.2 ล้านตัน) ตรงเหนือทรงกลม ($x=0$): $\Delta g=\dfrac{G\Delta M}{z^2}=\dfrac{6.674\times10^{-11}\times2.12\times10^{10}}{400^2}=8.85\times10^{-6}$ m/s² $=0.885$ mGal (1 mGal $=10^{-5}$ m/s²)
2. **แทน $x=z\tan\theta$** จะได้ $dx=z\sec^2\theta\,d\theta$ และ $x^2+z^2=z^2\sec^2\theta$ ดังนั้น $(x^2+z^2)^{3/2}=z^3\sec^3\theta$: $$\int\frac{dx}{(x^2+z^2)^{3/2}}=\int\frac{z\sec^2\theta}{z^3\sec^3\theta}d\theta=\frac{1}{z^2}\int\cos\theta\,d\theta=\frac{\sin\theta}{z^2}+C.$$
3. **กลับไปเป็นตัวแปร $x$** วาดสามเหลี่ยมมุมฉากที่มีด้านตรงข้ามมุมเป็น $x$ และด้านประชิดเป็น $z$ ด้านตรงข้ามมุมฉากคือ $\sqrt{x^2+z^2}$ ดังนั้น $\sin\theta=\dfrac{x}{\sqrt{x^2+z^2}}$ และปฏิยานุพันธ์คือ $\dfrac{x}{z^2\sqrt{x^2+z^2}}$
4. **หาปริพันธ์ตลอดแนวสำรวจ** เมื่อ $x\to\pm\infty$ ปฏิยานุพันธ์เข้าใกล้ $\pm\dfrac{1}{z^2}$ (คือ $\theta\to\pm\tfrac{\pi}{2}$) ดังนั้น $$\int_{-\infty}^{\infty}\Delta g\,dx=G\Delta Mz\cdot\frac{2}{z^2}=\frac{2G\Delta M}{z}\approx708\ \text{mGal·m}.$$
5. **ความลึกจากครึ่งความกว้าง** $\Delta g$ ลดลงเหลือครึ่งหนึ่งของค่าสูงสุดเมื่อ $(x^2+z^2)^{3/2}=2z^3$ คือ $x_{1/2}=z\sqrt{2^{2/3}-1}=0.766z=307$ m เมื่อกลับสูตรจะได้ $z=x_{1/2}/0.766\approx1.305\,x_{1/2}$ ซึ่งให้ความลึกจริง 400 m กลับคืนมา

```python
x, X = sp.symbols('x X', real=True); zz, G, dM = sp.symbols('z G Delta_M', positive=True)
dg = G*dM*zz/(x**2 + zz**2)**sp.Rational(3, 2)
anti = sp.integrate(1/(x**2 + zz**2)**sp.Rational(3, 2), x)
show(r"\int\frac{dx}{(x^2+z^2)^{3/2}}", anti)
line = sp.integrate(dg, (x, -sp.oo, sp.oo))
show(r"\int_{-\infty}^{\infty}\Delta g\,dx", line)

# half-width depth rule
xh = sp.solve(sp.Eq(dg, dg.subs(x, 0)/2), x)
xh = [s for s in xh if s.is_positive][0]
show("x_{1/2}", sp.simplify(xh)); print("z / x_1/2 =", float(zz/xh))

Gv, R, drho, zv = 6.674e-11, 150.0, 1500.0, 400.0     # sphere radius 150 m, Δρ = 1500 kg/m³, depth 400 m
dMv = 4/3*np.pi*R**3*drho
dgf = sp.lambdify(x, dg.subs({G: Gv, dM: dMv, zz: zv}))
print(f"ΔM = {dMv:.2e} kg -> peak anomaly {dgf(0)*1e5:.3f} mGal, half-width {float(xh.subs(zz, zv)):.0f} m")
```

```python
rng = np.random.default_rng(19)
xs = np.arange(-2000, 2001, 50.0)
obs = dgf(xs)*1e5 + rng.normal(0, 0.01, xs.size)    # survey with 0.01 mGal noise
half = obs.max()/2; above = xs[obs >= half]; xhalf = (above.max() - above.min())/2
print(f"from the 'field data': x_1/2 ≈ {xhalf:.0f} m -> estimated depth z ≈ 1.305·x_1/2 = {1.305*xhalf:.0f} m (true 400 m)")
fig, ax = plt.subplots()
ax.plot(xs, obs, "o", ms=3, label="survey readings (synthetic)"); ax.plot(xs, dgf(xs)*1e5, lw=2, label="buried-sphere model")
ax.axhline(half, ls=":", c="gray"); ax.set(xlabel="x along survey line (m)", ylabel="Δg (mGal)", title="Gravity anomaly over an ore body")
ax.legend(); plt.show()
```

**การตีความ** ปริพันธ์ตามแนวสำรวจมีค่าเท่ากับ $\dfrac{2G\Delta M}{z}$ ดังนั้นเมื่อทราบความลึก (จากความกว้างครึ่งหนึ่ง) ก็สามารถประมาณ**มวลส่วนเกิน** $\Delta M$ (ปริมาณตันของสายแร่) ได้จากพื้นที่ใต้กราฟค่าผิดปกติ
นี่คือทฤษฎีบทของเกาส์ (Gauss's theorem) ในรูปหนึ่งมิติที่ใช้ในการสำรวจแร่

**ความหมายของคำตอบ** สายแร่ขนาด 150 m ที่ความลึก 400 m ทำให้เกิดความผิดปกติไม่ถึง 1 mGal ซึ่งน้อยแต่เครื่องวัดแรงโน้มถ่วงสมัยใหม่วัดได้ง่าย (ความละเอียดประมาณ 0.01 mGal) ครึ่งความกว้างบอกนักธรณีฟิสิกส์ว่าต้องเจาะลึกเท่าใด และพื้นที่ใต้กราฟบอกมวลส่วนเกิน คือปริมาณแร่ ตรวจสอบ: วัตถุที่ลึกกว่าให้กราฟที่กว้างและแบนกว่า ดังนั้นความผิดปกติที่กว้างและอ่อนแสดงว่าแหล่งอยู่ลึก

**ข้อผิดพลาดที่พบบ่อย**

- ลืมแปลงหน่วย m/s² เป็น mGal (คูณด้วย $10^5$)
- ตอบในรูปของ $\theta$ โดยไม่แปลงกลับเป็น $x$ ด้วยสามเหลี่ยมอ้างอิง
- ใช้ความกว้างเต็มที่ครึ่งความสูง ($2x_{1/2}$) ในสูตร $z\approx1.305\,x_{1/2}$ ทำให้ความลึกเพิ่มเป็นสองเท่า

**ลองทำดู** เพิ่มความลึกเป็นสองเท่าเป็น 800 m ค่าผิดปกติสูงสุดและความกว้างครึ่งหนึ่งจะเปลี่ยนไปอย่างไร ยังตรวจวัดได้หรือไม่ถ้ามีสัญญาณรบกวน 0.01 mGal

## ตัวอย่างที่ 40 ⭐⭐⭐: อายุเฉลี่ยของอะตอมกัมมันตรังสีและเวลาเดินทางเฉลี่ยของน้ำบาดาล (การหาปริพันธ์ทีละส่วน (integration by parts))

**บริบท** ถ้าอะตอมสลายตัวด้วยอัตรา $\lambda$ ความน่าจะเป็นที่อะตอมหนึ่งอยู่รอดถึงเวลา $t$ แล้วสลายตัวในช่วง $[t, t+dt]$ คือ $\lambda e^{-\lambda t}dt$
**อายุเฉลี่ย (mean life)** คือ $\bar t = \int_0^\infty t\,\lambda e^{-\lambda t}\,dt$ นักอุทกวิทยาใช้ปริพันธ์เดียวกันนี้ (โดย $\lambda = 1/\tau$) สำหรับ
*แบบจำลองเอกซ์โพเนนเชียล (exponential model)* ของอายุน้ำบาดาลในชั้นหินอุ้มน้ำที่ผสมกันดี

**แคลคูลัส** การหาปริพันธ์ทีละส่วน $\int u\,dv = uv - \int v\,du$ โดย $u = t$, $dv = \lambda e^{-\lambda t}dt$:

$$\int t\lambda e^{-\lambda t}dt = -te^{-\lambda t} + \int e^{-\lambda t}dt = -te^{-\lambda t} - \frac{1}{\lambda}e^{-\lambda t} + C.$$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ค่าเฉลี่ยของการแจกแจงความน่าจะเป็นคือ $\int t\,f(t)\,dt$ สำหรับการสลายตัวของสารกัมมันตรังสี $f(t)=\lambda e^{-\lambda t}$ เราจึงต้องหาปริพันธ์ของ **ผลคูณ** ระหว่าง $t$ กับฟังก์ชันเลขชี้กำลัง ซึ่งเป็นงานของ **การหาปริพันธ์โดยการแยกส่วน (integration by parts)** โดยเลือก $u=t$ เพื่อให้อนุพันธ์ของมันง่ายขึ้น

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **เลือก $u$ และ $dv$** ให้ $u=t$ (หาอนุพันธ์แล้วง่ายขึ้น: $du=dt$) และ $dv=\lambda e^{-\lambda t}dt$ (หาปริพันธ์ได้ง่าย: $v=-e^{-\lambda t}$)
2. **ใช้สูตร $\int u\,dv=uv-\int v\,du$** $$\int t\lambda e^{-\lambda t}dt=-te^{-\lambda t}+\int e^{-\lambda t}dt=-te^{-\lambda t}-\frac1\lambda e^{-\lambda t}+C.$$ ตรวจโดยหาอนุพันธ์: กฎผลคูณให้ $t\lambda e^{-\lambda t}$ กลับคืนมา
3. **ปริพันธ์จำกัดเขตถึง $T$** $$\int_0^T t\lambda e^{-\lambda t}dt=\frac1\lambda-\left(T+\frac1\lambda\right)e^{-\lambda T}.$$
4. **ให้ $T\to\infty$** $Te^{-\lambda T}\to0$ (ฟังก์ชันเลขชี้กำลังลดลงเร็วกว่าตัวประกอบเชิงเส้นที่เพิ่มขึ้น ยืนยันได้ด้วยโลปีตาล) ดังนั้นช่วงชีวิตเฉลี่ยคือ $\bar t=\dfrac1\lambda$ และเนื่องจาก $\lambda=\dfrac{\ln2}{t_{1/2}}$ จึงได้ $\bar t=\dfrac{t_{1/2}}{\ln2}\approx1.44\,t_{1/2}$
5. **ตัวเลขของทริเทียม** $t_{1/2}=12.3$ ปี ให้ $\lambda=0.6931/12.3=0.05635$ ต่อปี และ $\bar t=17.75$ ปี เมื่อขีดจำกัดบน $T=40$ ปี: $e^{-\lambda T}=0.105$ ดังนั้น $\int_0^{40}=17.75-(40+17.75)(0.105)=11.68$ ปี คิดเป็น $65.8\%$ ของลิมิต ส่วนที่เหลือมาจากอะตอมจำนวนน้อยที่สลายตัวช้ามาก

```python
t, T = sp.symbols('t T', positive=True); lam = sp.symbols('lambda', positive=True)
u, dv = t, lam*sp.exp(-lam*t)
v = sp.integrate(dv, t)
by_parts = u*v - sp.integrate(v*sp.diff(u, t), t)
show(r"\int t\lambda e^{-\lambda t}dt", sp.simplify(by_parts))
print("check (derivative of result = integrand):", sp.simplify(sp.diff(by_parts, t) - t*lam*sp.exp(-lam*t)) == 0)

finite = sp.simplify(by_parts.subs(t, T) - by_parts.subs(t, 0))
show(r"\int_0^T t\lambda e^{-\lambda t}dt", finite)
mean = sp.limit(finite, T, sp.oo)
show(r"\bar t=\lim_{T\to\infty}\int_0^T t\lambda e^{-\lambda t}dt", mean)

# repeated integration by parts for the second moment -> standard deviation
m2 = sp.integrate(t**2*lam*sp.exp(-lam*t), (t, 0, sp.oo))
show(r"\sigma=\sqrt{\overline{t^2}-\bar t^2}", sp.simplify(sp.sqrt(m2 - mean**2)))

for name, half in [("¹⁴C", 5730), ("²²²Rn (in groundwater, days)", 3.82), ("³H tritium", 12.32)]:
    print(f"{name:30s} half-life {half:>8} -> mean life 1/λ = half-life/ln2 = {half/np.log(2):.2f}")
```

**การตีความ** อายุเฉลี่ยคือ $1/\lambda = t_{1/2}/\ln 2 \approx 1.44\,t_{1/2}$ ซึ่งยาวกว่าครึ่งชีวิต (half-life) เพราะมีอะตอมจำนวนน้อยที่อยู่รอดได้นานมาก
ในทางอุทกวิทยา ถ้าน้ำในบ่อมีการกระจายอายุแบบเอกซ์โพเนนเชียลที่มีเวลาเดินทางเฉลี่ย $\tau$ แล้ว $\tau$ จะบอกว่าโดยเฉลี่ยสารมลพิษที่รั่วไหลจะใช้เวลานานเท่าใดกว่าจะถึงบ่อ

**ความหมายของคำตอบ** ช่วงชีวิตเฉลี่ยยาวกว่าครึ่งชีวิตประมาณ 44% เพราะอะตอมจำนวนน้อยอยู่รอดได้นานมาก จึงดึงค่าเฉลี่ยให้สูงขึ้น สำหรับคาร์บอน-14 ได้ 8267 ปี และสำหรับทริเทียมประมาณ 17.7 ปี ในอุทกธรณีวิทยา ปริพันธ์เดียวกันให้เวลาเดินทางเฉลี่ยของน้ำในชั้นหินอุ้มน้ำที่ผสมกันดี คือเวลาเฉลี่ยที่สารปนเปื้อนใช้เดินทางถึงบ่อ ตรวจสอบ: ผลลัพธ์ $1/\lambda$ มีหน่วยเป็นเวลา ซึ่งถูกต้องสำหรับอายุเฉลี่ย

**ข้อผิดพลาดที่พบบ่อย**

- เลือก $u=e^{-\lambda t}$ และ $dv=t\,dt$ ทำให้ปริพันธ์ใหม่ $\int\tfrac{t^2}{2}\lambda e^{-\lambda t}dt$ ยากขึ้น ไม่ใช่ง่ายขึ้น
- ทำเครื่องหมายลบใน $v=-e^{-\lambda t}$ หาย ทำให้ค่าเฉลี่ยติดลบ
- สับสนระหว่างครึ่งชีวิต $t_{1/2}$ กับช่วงชีวิตเฉลี่ย $1/\lambda$

**ลองทำดู** ใช้การหาปริพันธ์ทีละส่วนสองครั้ง คำนวณ $\int_0^{\infty}t^2\lambda e^{-\lambda t}\,dt$ ด้วยมือ แล้วตรวจสอบว่าตรงกับผลจาก SymPy คือ $2/\lambda^2$

## ตัวอย่างที่ 41 ⭐⭐⭐: พื้นดินเก็บความร้อนได้เท่าใดในแต่ละปี (การหาปริพันธ์ทีละส่วน สองครั้ง)

**บริบท** คลื่นอุณหภูมิผิวดินตามฤดูกาลแพร่ลงสู่พื้นดินและถูกหน่วงลงตามความลึก:

$$T(z,t) = \bar T + A\,e^{-z/d}\cos\!\left(\omega t - \frac zd\right),\qquad d = \sqrt{\frac{2\kappa}{\omega}}.$$

สภาพแพร่ความร้อนของดิน $\kappa = 5\times10^{-7}$ m²/s แอมพลิจูดที่ผิว $A = 10$ °C ความจุความร้อนต่อปริมาตร $\rho c = 2.0\times10^6$ J/(m³·K) ความร้อนส่วนเกินที่เก็บไว้ต่อ m² คือ $Q(t) = \rho c\displaystyle\int_0^\infty (T - \bar T)\,dz$

**แคลคูลัส** ให้ $u = z/d$ และ $\phi = \omega t$ เราต้องหา $\int_0^\infty e^{-u}\cos(\phi-u)\,du$ เมื่อกระจายโคไซน์จะได้ $\int e^{-u}\cos u\,du$ และ $\int e^{-u}\sin u\,du$ ซึ่งแต่ละตัวต้องใช้ **การหาปริพันธ์ทีละส่วน (integration by parts) สองครั้ง** (ปริพันธ์เดิมปรากฏซ้ำ เป็นกรณี "วนกลับ"):
$$\int_0^\infty e^{-u}\cos u\,du = \int_0^\infty e^{-u}\sin u\,du = \tfrac12\ \Rightarrow\ \int_0^\infty e^{-u}\cos(\phi-u)\,du = \frac{\cos\phi+\sin\phi}{2}.$$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** ความร้อนที่สะสมในดินคือปริพันธ์ของอุณหภูมิส่วนเกินตลอดทุกความลึก ตัวถูกหาปริพันธ์เป็นเอกซ์โพเนนเชียลคูณโคไซน์ และการหาปริพันธ์ผลคูณแบบนี้โดยการแยกส่วน (integration by parts) สองครั้งจะได้ปริพันธ์ตัวเดิมกลับมา การย้ายข้างจึงแก้สมการได้ เทคนิคนี้เรียกว่าการแยกส่วนแบบวนกลับ (cyclic)

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ตั้งปริพันธ์** $Q(t)=\rho c\displaystyle\int_0^\infty Ae^{-z/d}\cos\!\left(\omega t-\frac zd\right)dz$ ให้ $s=z/d$, $dz=d\,ds$, $\phi=\omega t$ จะได้ $Q=\rho cAd\displaystyle\int_0^\infty e^{-s}\cos(\phi-s)\,ds$
2. **กระจายโคไซน์** $\cos(\phi-s)=\cos\phi\cos s+\sin\phi\sin s$ จึงต้องหา $I_c=\displaystyle\int_0^\infty e^{-s}\cos s\,ds$ และ $I_s=\displaystyle\int_0^\infty e^{-s}\sin s\,ds$
3. **แยกส่วนครั้งแรก** ให้ $f=\cos s$, $dg=e^{-s}ds$ จะได้ $g=-e^{-s}$: $I_c=\Big[-e^{-s}\cos s\Big]_0^\infty-\displaystyle\int_0^\infty e^{-s}\sin s\,ds=1-I_s$
4. **แยกส่วนครั้งที่สอง** ให้ $f=\sin s$: $I_s=\Big[-e^{-s}\sin s\Big]_0^\infty+\displaystyle\int_0^\infty e^{-s}\cos s\,ds=0+I_c$ ปริพันธ์ตัวเดิมกลับมาอีกครั้ง
5. **แก้สมการ** $I_c=1-I_c\Rightarrow I_c=I_s=\tfrac12$ ดังนั้น $Q(t)=\rho cAd\,\dfrac{\cos\omega t+\sin\omega t}{2}=\dfrac{\rho cAd}{\sqrt2}\sin\!\left(\omega t+\dfrac\pi4\right)$ มีแอมพลิจูด $\dfrac{\rho cAd}{\sqrt2}$
6. **ตัวเลขของคลื่นรายปี** $\omega=\dfrac{2\pi}{365\times86400}\approx1.99\times10^{-7}$ s⁻¹ และ $d=\sqrt{2\kappa/\omega}=\sqrt{2\times5\times10^{-7}/1.99\times10^{-7}}\approx2.24$ m แอมพลิจูด $=\dfrac{2\times10^6\times10\times2.24}{\sqrt2}\approx3.17\times10^7$ J/m² $\approx31.7$ MJ/m² สำหรับคลื่นรายวัน $d$ เล็กลง $\sqrt{365}$ เท่า เหลือประมาณ $0.117$ m

```python
u, ph = sp.symbols('u phi', real=True)
I1 = sp.integrate(sp.exp(-u)*sp.cos(u), (u, 0, sp.oo)); I2 = sp.integrate(sp.exp(-u)*sp.sin(u), (u, 0, sp.oo))
print("∫ e^(-u) cos u du =", I1, "  ∫ e^(-u) sin u du =", I2)
full = sp.integrate(sp.exp(-u)*sp.expand_trig(sp.cos(ph - u)), (u, 0, sp.oo))
show(r"\int_0^\infty e^{-u}\cos(\phi-u)\,du", sp.simplify(full))

kap, A, rc = 5e-7, 10.0, 2.0e6
for name, period in [("annual", 365.25*86400), ("daily", 86400)]:
    w = 2*np.pi/period; d = np.sqrt(2*kap/w)
    print(f"{name:7s} wave: damping depth d = {d:.3f} m, stored-heat amplitude ρcAd/√2 = {rc*A*d/np.sqrt(2):.3e} J/m²")
```

```python
w = 2*np.pi/(365.25*86400); d = np.sqrt(2*kap/w)
Z = np.linspace(0, 10, 300)
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 4.5))
for mth in range(0, 12, 2):
    phi_ = 2*np.pi*mth/12
    a1.plot(A*np.exp(-Z/d)*np.cos(phi_ - Z/d), Z, lw=1.5, label=f"month {mth}")
a1.invert_yaxis(); a1.set(xlabel="T − T̄ (°C)", ylabel="depth (m)", title="annual temperature wave in the ground"); a1.legend(fontsize=8)
months = np.linspace(0, 12, 200)
a2.plot(months, rc*A*d*(np.cos(2*np.pi*months/12) + np.sin(2*np.pi*months/12))/2/1e6, "C3", lw=2)
a2.axhline(0, c="k", lw=.6); a2.set(xlabel="month", ylabel="stored heat Q (MJ/m²)", title="heat stored, peaks 1.5 months after surface max")
plt.tight_layout(); plt.show()
```

**การตีความ** คลื่นรายปีแทรกลงไปด้วยความลึกหน่วง **d ≈ 2.24 m** (คลื่นรายวันเพียง 0.12 m) และพื้นดินเก็บและคายความร้อนประมาณ **±3.17 × 10⁷ J/m²** (31.7 MJ/m²) ในแต่ละปี
เนื่องจาก $(\cos\phi+\sin\phi)/2 = \tfrac{1}{\sqrt2}\cos(\phi-\pi/4)$ การเก็บความร้อนจึงสูงสุดหลังอุณหภูมิผิวดินสูงสุด 1/8 ปี (≈ 1.5 เดือน) แหล่งเก็บความร้อนตามฤดูกาลนี้คือสิ่งที่ปั๊มความร้อนใต้พิภพ (ground-source heat pump) นำมาใช้

**ความหมายของคำตอบ** ดินแต่ละตารางเมตรดูดซับและคายความร้อนประมาณ 32 MJ ต่อปี ราวพลังงานในน้ำมันเบนซิน 1 ลิตร แอมพลิจูดลดเหลือ 5% ที่ $3d\approx6.7$ m และคลื่นไปถึงความลึก 1 m ช้ากว่าผิวดินประมาณ 26 วัน ห้องใต้ดินจึงเย็นในฤดูร้อน และปั๊มความร้อนใต้ดินจึงใช้งานได้ ส่วนคลื่นรายวันหายไปภายในราว 30 cm

**ข้อผิดพลาดที่พบบ่อย**

- สลับบทบาทของ $f$ กับ $dg$ ในขั้นที่สอง ต้องให้ $e^{-s}$ เป็นส่วนที่หาปริพันธ์ทั้งสองครั้ง มิฉะนั้นจะย้อนขั้นแรกกลับไปเฉย ๆ
- คิดว่าเมื่อปริพันธ์ตัวเดิมกลับมาแปลว่าทำต่อไม่ได้ ให้ถือเป็นตัวไม่ทราบค่าแล้วแก้ $I_c=1-I_c$
- ลืม $dz=d\,ds$ เมื่อแทน $s=z/d$ ทำให้ตัวคูณ $d$ ในความร้อนสะสมหายไป

**ลองทำดู** ที่ความลึกเท่าใดคลื่นอุณหภูมิรายปีมีเฟสตรงข้ามกับผิวดินพอดี (ฤดูร้อนที่ผิว = ฤดูหนาวด้านล่าง) (คำใบ้: $z/d = \pi$)

## ตัวอย่างที่ 42 ⭐⭐⭐: จะเกิดอาฟเตอร์ช็อกกี่ครั้ง (ปริพันธ์ไม่ตรงแบบ (improper integrals))

**บริบท** อัตราการเกิดอาฟเตอร์ช็อก (aftershock) เป็นไปตาม**กฎของโอโมริ–อุตสึ (Omori–Utsu law)**

$$n(t) = \frac{K}{(t+c)^p}\quad\text{(events/day)},$$

โดย $K = 200$, $c = 0.05$ วัน และโดยทั่วไป $p \approx 1.1$ จำนวนอาฟเตอร์ช็อก*ทั้งหมดที่จะเกิดขึ้น*คือปริพันธ์ไม่ตรงแบบ $N_\infty = \int_0^\infty n(t)\,dt$

**แคลคูลัส** $\displaystyle\int_0^\infty\frac{K\,dt}{(t+c)^p} = \lim_{T\to\infty}\left[\frac{K(t+c)^{1-p}}{1-p}\right]_0^T$ ซึ่ง**ลู่เข้าก็ต่อเมื่อ $p>1$ เท่านั้น**:
$N_\infty = \dfrac{Kc^{1-p}}{p-1}$ สำหรับ $p=1$ จะลู่ออกในลักษณะ $K\ln T$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** กฎของโอโมริ–อุตสึ (Omori–Utsu law) ให้ **อัตรา** การเกิดแผ่นดินไหวตาม ดังนั้นจำนวนทั้งหมดคือ **ปริพันธ์** ของอัตรา เมื่อถามจำนวนรวมตลอดเวลาในอนาคต จึงเป็น **ปริพันธ์ไม่ตรงแบบ (improper integral)** $\int_0^\infty$ ซึ่งนิยามด้วยลิมิต การที่ค่าเป็นจำนวนจำกัดหรือไม่ขึ้นกับเลขชี้กำลัง $p$ เหมือนกับการทดสอบ $p$ ของ $\int_1^\infty x^{-p}dx$

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **หาปริพันธ์ถึงเวลาจำกัด $T$** (กฎยกกำลังโดยให้ $u=t+c$ ใช้ได้เมื่อ $p\neq1$): $$N(T)=\int_0^T\frac{K\,dt}{(t+c)^p}=\frac{K}{1-p}\left[(T+c)^{1-p}-c^{1-p}\right].$$
2. **หาลิมิตเมื่อ $T\to\infty$** ถ้า $p>1$ จะได้ $1-p<0$ และ $(T+c)^{1-p}\to0$ ดังนั้น $$N_\infty=\frac{K}{1-p}\left(0-c^{1-p}\right)=\frac{Kc^{1-p}}{p-1}.$$ ถ้า $p<1$ พจน์ยกกำลังโตไม่จำกัด และถ้า $p=1$ ปริพันธ์คือ $K\ln\frac{T+c}{c}\to\infty$ ทั้งสองกรณี **ลู่ออก (diverge)**
3. **แทนค่า ($K=200$, $c=0.05$ วัน, $p=1.1$)** $c^{1-p}=0.05^{-0.1}=1.349$ ดังนั้น $N_\infty=\dfrac{200\times1.349}{0.1}\approx2699$ ครั้ง
4. **ลู่เข้าเร็วเพียงใด** หลัง 1 วัน $N\approx708$ (26%) หลัง $T\approx363$ วัน (ค่าเริ่มต้นในแอป) $N\approx1589$ (58.9%) และอัตราลดลงเหลือ $n=200/(363.05)^{1.1}\approx0.31$ ครั้งต่อวัน หลัง 10 ปีได้เพียง 67% การแก้สมการ $N(T)=0.9N_\infty$ ให้ $T\approx1.4$ ล้านปี

```python
t, T = sp.symbols('t T', positive=True); Kc, cc, p = sp.symbols('K c p', positive=True)
NT = sp.integrate(Kc/(t + cc)**p, (t, 0, T), conds='none')
show("N(T)", sp.simplify(NT))
for pv in [sp.Rational(11, 10), 1, sp.Rational(9, 10)]:
    lim = sp.limit(sp.integrate(200/(t + sp.Rational(1, 20))**pv, (t, 0, T)), T, sp.oo)
    print(f"p = {str(pv):>5}: N∞ = {lim if lim == sp.oo else f'{float(lim):.0f} aftershocks'}")

Kv, cv, pv = 200, 0.05, 1.1
Ninf = Kv*cv**(1-pv)/(pv-1)
Nf = lambda T: Kv/(1-pv)*((T + cv)**(1-pv) - cv**(1-pv))
for T_ in [1, 30, 365, 3650]:
    print(f"first {T_:>5} days: {Nf(T_):7.0f} aftershocks ({Nf(T_)/Ninf*100:4.1f}% of total {Ninf:.0f})")
T90 = optimize.brentq(lambda T: Nf(T) - 0.9*Ninf, 1, 1e12)
print(f"time until 90% of all aftershocks have happened: {T90:.2e} days = {T90/365.25:.1e} years !")
```

```python
TT = np.logspace(-2, 5, 300)
fig, ax = plt.subplots()
for pv_ in [0.9, 1.0, 1.1, 1.3]:
    NN = Kv*np.log((TT + cv)/cv) if pv_ == 1 else Kv/(1-pv_)*((TT + cv)**(1-pv_) - cv**(1-pv_))
    ax.semilogx(TT, NN, lw=2, label=f"p = {pv_}" + ("  (converges)" if pv_ > 1 else "  (diverges)"))
ax.set(xlabel="days after mainshock T", ylabel="cumulative aftershocks N(T)", title="Improper integral: does N(T) level off?")
ax.legend(); plt.show()
```

**การตีความ** เมื่อ $p = 1.1$ ปริพันธ์ลู่เข้า แต่**ช้ามาก**: หลังผ่านไป 10 ปี เกิดขึ้นเพียงประมาณสองในสามของจำนวนรวม "ในที่สุด" และต้องใช้เวลามากกว่าหนึ่งล้านปีจึงจะถึง 90 % ในความเป็นจริง อาฟเตอร์ช็อกจะกลืนไปกับแผ่นดินไหวพื้นหลังก่อนหน้านั้นนานมาก
กรณี $p \le 1$ คือปริพันธ์ $p$ ที่ลู่ออกแบบคลาสสิก ซึ่งเป็นการทดสอบเดียวกับที่เรียนในตำราสำหรับ $\int_1^\infty x^{-p}dx$

**ความหมายของคำตอบ** เมื่อ $p=1.1$ จำนวนรวมตามทฤษฎีมีค่าจำกัด (ประมาณ 2700 ครั้ง) แต่ส่วนใหญ่เกิดในช่วงแรก หนึ่งในสี่เกิดในวันแรก และหางของการแจกแจงยาวมากจนในทางปฏิบัติแผ่นดินไหวตามจะกลืนไปกับแผ่นดินไหวพื้นหลังนานก่อนที่ปริพันธ์จะ \"ครบ\" สำหรับการวางแผนรับมือภัย ค่าที่มีประโยชน์คือ $N(T)$ ในอีกไม่กี่วันหรือสัปดาห์ ซึ่งเป็นวิธีออกคำพยากรณ์แผ่นดินไหวตาม ตรวจสอบ: ถ้า $p\le1$ สูตรจะทำนายว่ามีแผ่นดินไหวตามไม่จำกัด จึงเป็นเหตุผลที่ค่า $p$ ที่วัดได้มักมากกว่า 1 เล็กน้อย

**ข้อผิดพลาดที่พบบ่อย**

- ใช้ปฏิยานุพันธ์จากกฎยกกำลังเมื่อ $p=1$ กรณีนี้ต้องใช้ $\ln(t+c)$
- แทน $T=\infty$ ตรง ๆ แทนที่จะเขียนเป็นลิมิต ปริพันธ์ไม่ตรงแบบนิยามเป็น $\lim_{T\to\infty}\int_0^T$ เสมอ
- คิดว่าอัตราที่ลดลง ($n(t)\to0$) ทำให้จำนวนรวมมีค่าจำกัดเสมอ กรณี $p=1$ อัตราก็ลดลงแต่ยังลู่ออก

**ลองทำดู** ใช้สูตรของ $N(T)$ ประมาณจำนวนอาฟเตอร์ช็อกที่เกิดเฉพาะในวันที่ 30 นั่นคือ $N(31) - N(30)$

## ตัวอย่างที่ 43 ⭐⭐⭐: ความเร็วหลุดพ้นและพวยแก๊สภูเขาไฟบนดวงจันทร์ไอโอ (ปริพันธ์ไม่ตรงแบบ)

**บริบท** งานที่ต้องใช้ยกมวล $m$ จากผิวดาวเคราะห์ ($r = R$) ไปถึงระยะอนันต์คือ

$$W = \int_R^\infty\frac{GMm}{r^2}\,dr = \lim_{b\to\infty}\left[-\frac{GMm}{r}\right]_R^b = \frac{GMm}{R}.$$

ให้ $\tfrac12mv^2 = W$ จะได้ **ความเร็วหลุดพ้น (escape velocity)** $v_{esc} = \sqrt{2GM/R}$ หากอัตราเร็วเริ่มต้น $v < v_{esc}$ การอนุรักษ์พลังงานให้ระยะไกลสุด: $\dfrac1R - \dfrac1{r_{\max}} = \dfrac{v^2}{2GM}$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** แรงโน้มถ่วงลดลงตาม $1/r^2$ แต่ไม่เคยเป็นศูนย์ การยกมวลไปถึงระยะอนันต์จึงต้องหาปริพันธ์ของแรงตลอดระยะทางที่ไม่สิ้นสุด ปริพันธ์ไม่ตรงแบบ (improper integral) นี้ยังมีค่าจำกัด วัตถุที่เร็วพอจึงหลุดพ้นจากดาวเคราะห์ได้ การตั้งพลังงานจลน์ให้เท่ากับงานนี้ให้ความเร็วหลุดพ้น (escape velocity)

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **เขียนปริพันธ์ไม่ตรงแบบเป็นลิมิต** $W=\displaystyle\int_R^\infty\frac{GMm}{r^2}dr=\lim_{b\to\infty}\int_R^b\frac{GMm}{r^2}dr$
2. **หาปริพันธ์** ปฏิยานุพันธ์ของ $r^{-2}$ คือ $-r^{-1}$: $\displaystyle\int_R^b\frac{GMm}{r^2}dr=GMm\left(\frac1R-\frac1b\right)$ เมื่อ $b\to\infty$ จะได้ $\dfrac1b\to0$ ดังนั้น $W=\dfrac{GMm}{R}$ (มีค่าจำกัด ปริพันธ์ลู่เข้า)
3. **ความเร็วหลุดพ้น** $\tfrac12mv^2=\dfrac{GMm}{R}\Rightarrow v_{\text{esc}}=\sqrt{\dfrac{2GM}{R}}$ มวล $m$ ตัดกันหมด ค่านี้จึงเท่ากันทั้งเม็ดฝุ่นและก้อนหินใหญ่
4. **ไอโอ (Io):** $GM=5.96\times10^{12}$ m³/s², $R=1.8216\times10^6$ m จะได้ $v_{\text{esc}}=\sqrt{2\times5.96\times10^{12}/1.8216\times10^6}\approx2558$ m/s $\approx2.56$ km/s (โลก 11.19 km/s ดวงจันทร์ 2.38 km/s)
5. **ความสูงของพลูมที่ช้ากว่า** เมื่อ $v<v_{\text{esc}}$ ปริพันธ์เดียวกันจาก $R$ ถึง $r_{\max}$ ให้ $\dfrac1R-\dfrac1{r_{\max}}=\dfrac{v^2}{2GM}$ เมื่อ $v=1$ km/s: $\dfrac1{r_{\max}}=5.490\times10^{-7}-8.39\times10^{-8}\approx4.651\times10^{-7}$ m⁻¹ ดังนั้น $r_{\max}\approx2150$ km และความสูงคือ $r_{\max}-R\approx329$ km

```python
r, R, b = sp.symbols('r R b', positive=True); GMm = sp.symbols('GMm', positive=True)
W = sp.limit(sp.integrate(GMm/r**2, (r, R, b)), b, sp.oo)
show(r"\int_R^\infty\frac{GMm}{r^2}dr", W)
bodies = {"Earth": (3.986004e14, 6.371e6), "Moon": (4.9048e12, 1.7374e6), "Io (Jupiter's moon)": (5.959916e12, 1.8216e6)}
for name, (GM, Rb) in bodies.items():
    print(f"{name:20s} v_esc = {np.sqrt(2*GM/Rb)/1000:6.2f} km/s")
GM, Rb = bodies["Io (Jupiter's moon)"]; vv = 1000.0
rmax = 1/(1/Rb - vv**2/(2*GM))
g_io = GM/Rb**2
print(f"Io plume, vent speed {vv:.0f} m/s: height = {(rmax - Rb)/1000:.0f} km "
      f"(constant-gravity estimate v²/2g = {vv**2/(2*g_io)/1000:.0f} km, g = {g_io:.3f} m/s²)")
```

```python
V = np.linspace(0, 2500, 300)
GM, Rb = bodies["Io (Jupiter's moon)"]
hexact = 1/(1/Rb - V**2/(2*GM)) - Rb
fig, ax = plt.subplots()
ax.plot(V, hexact/1000, lw=2, label="exact (gravity weakens with height)")
ax.plot(V, V**2/(2*GM/Rb**2)/1000, "--", label="constant gravity v²/2g")
ax.axvline(np.sqrt(2*GM/Rb), c="C3", ls=":", label="escape velocity")
ax.set(xlabel="vent speed (m/s)", ylabel="plume height (km)", ylim=(0, 3000), title="Volcanic plume height on Io"); ax.legend(); plt.show()
```

**การตีความ** ปริพันธ์ไม่ตรงแบบ (improper integral) นี้ลู่เข้า ดังนั้นพลังงานจำกัดก็เพียงพอที่จะหลุดพ้นได้: **11.19 km/s** จากโลก 2.38 km/s จากดวงจันทร์ และ **2.56 km/s** จากไอโอ
ไอโอซึ่งเป็นวัตถุที่มีภูเขาไฟมากที่สุดในระบบสุริยะ มีพวยแก๊สที่พุ่งสูงหลายร้อยกิโลเมตร อัตราเร็วที่ปล่อง 1 km/s ไปได้ไกลประมาณ 329 km มากกว่า 278 km ที่ทำนายด้วยแรงโน้มถ่วงคงที่ เพราะแรงโน้มถ่วงลดลงตามความสูง

**ความหมายของคำตอบ** แรงโน้มถ่วงของไอโอน้อย ($g\approx1.80$ m/s²) แก๊สภูเขาไฟที่พุ่งออกด้วยความเร็ว 1 km/s จึงขึ้นไปได้ประมาณ 330 km สอดคล้องกับพลูมขนาดยักษ์ของ Pele และ Loki ที่ยานอวกาศถ่ายไว้ สูตรแรงโน้มถ่วงคงที่ $v^2/2g\approx278$ km ให้ค่าต่ำเกินไปเพราะแรงโน้มถ่วงอ่อนลงเมื่อพลูมสูงขึ้น บนโลกความเร็วพ่นเท่ากันนี้ต่ำกว่า 11.2 km/s มาก วัตถุจากภูเขาไฟจึงตกกลับลงมาเสมอ

**ข้อผิดพลาดที่พบบ่อย**

- แทน $\infty$ ลงใน $-1/r$ ตรง ๆ ควรเขียนลิมิต $b\to\infty$ ให้ชัดเจน ซึ่งทำให้ปริพันธ์ไม่ตรงแบบมีความหมาย
- ปน km กับ m $GM$ มีหน่วย m³/s² ดังนั้น $R$ ต้องเป็นเมตรและ $v$ เป็น m/s
- รายงาน $r_{\max}$ เป็นความสูงของพลูม $r_{\max}$ วัดจากจุดศูนย์กลาง ต้องลบ $R$ ออก

**ลองทำดู** ทำไมภูเขาไฟบนโลกจึงพ่นหินออกสู่อวกาศไม่ได้ เปรียบเทียบวัตถุที่พ่นออกมาเร็วที่สุด (~600 m/s) กับความเร็วหลุดพ้นของโลก

# ส่วน G: การประยุกต์ปริพันธ์ (Applications of Integration)

⭐⭐⭐ ยาก · ตัวอย่างที่ 44–46 · สัปดาห์ที่ 17 ในประมวลรายวิชา

## ตัวอย่างที่ 44 ⭐⭐⭐: ปริมาตรน้ำหลากจากไฮโดรกราฟของแม่น้ำ (พื้นที่ใต้เส้นโค้ง (area under a curve))

**บริบท** สถานีวัดน้ำบันทึกอัตราการไหล $Q(t)$ (m³/s) ทุก 3 ชั่วโมงระหว่างพายุฝนมรสุม **พื้นที่ใต้ไฮโดรกราฟ (hydrograph)** คือปริมาตรน้ำทั้งหมด
$V = \int Q\,dt$ เมื่อหักการไหลพื้นฐาน (baseflow ซึ่งเป็นการไหลที่มาจากน้ำบาดาล) ออก จะได้**น้ำท่าโดยตรง (direct runoff)** ซึ่งเป็นพื้นที่*ระหว่างเส้นโค้งสองเส้น*
เมื่อหารด้วยพื้นที่รับน้ำ จะได้ความลึกน้ำท่า (mm) ซึ่งนักอุทกวิทยานำไปเปรียบเทียบกับปริมาณฝน

ไฮโดรกราฟพายุสังเคราะห์ (รูปแบบแกมมา):
$Q(t) = Q_b(t) + Q_p\left(\dfrac{t}{t_p}\right)^m e^{\,m(1-t/t_p)}$ โดย $Q_p = 420$ m³/s, $t_p = 18$ h, $m = 3$ และการไหลพื้นฐาน $Q_b(t) = 35 + 0.15t$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** อัตราการไหล $Q(t)$ เป็นอัตรา (ลูกบาศก์เมตรต่อวินาที) ดังนั้น **พื้นที่ใต้กราฟชลภาพ (hydrograph)** $\int Q\,dt$ คือปริมาตรน้ำ เมื่อหักการไหลพื้นฐาน (baseflow) ออก จะได้ **พื้นที่ระหว่างเส้นโค้งสองเส้น** ซึ่งคือน้ำท่าจากพายุ สถานีวัดจริงบันทึกค่าทุก ๆ ไม่กี่ชั่วโมง เราจึงเปรียบเทียบ **การหาปริพันธ์เชิงตัวเลข** (กฎสี่เหลี่ยมคางหมูและกฎของซิมป์สัน) กับปริพันธ์ที่แม่นตรง

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ปริมาตรการไหลพื้นฐาน (พหุนามอย่างง่าย)** $$\int_0^{120}(35+0.15t)\,dt=\left[35t+0.075t^2\right]_0^{120}=4200+1080=5280\ \text{m}^3\text{/s·h}.$$ คูณด้วย 3600 s/h: $1.90\times10^7$ m³ $=19.01$ ล้าน m³
2. **น้ำท่าโดยตรง (เปลี่ยนตัวแปร $s=t/t_p$)** $$\int_0^\infty Q_p\left(\tfrac{t}{t_p}\right)^3e^{3(1-t/t_p)}dt=Q_pt_pe^3\int_0^\infty s^3e^{-3s}ds=Q_pt_pe^3\cdot\frac{3!}{3^4}.$$ ปริพันธ์ตัวสุดท้ายใช้การหาปริพันธ์โดยการแยกส่วนสามครั้ง แทนค่า: $420\times18\times20.09\times\tfrac{6}{81}=11\,248$ m³/s·h $=40.49$ ล้าน m³ การหยุดที่ 120 ชั่วโมงแทบไม่เปลี่ยนค่า เพราะพายุจบไปแล้ว
3. **ปริมาตรรวม** $19.01+40.49=59.50$ ล้าน m³ คือพื้นที่แม่นตรงใต้ชลภาพทั้งหมด
4. **ตรวจสอบเชิงตัวเลขด้วยข้อมูลสถานีวัด** เมื่ออ่านค่าทุก $\Delta t=3$ ชั่วโมง กฎสี่เหลี่ยมคางหมู $T_n=\tfrac{\Delta t}{2}\left[Q_0+2Q_1+\cdots+2Q_{n-1}+Q_n\right]$ ให้ 59.50 ล้าน m³ และกฎของซิมป์สันให้ 59.49 ทั้งสองต่างจากค่าแม่นตรงประมาณ 0.01% เมื่อเว้นช่วง 12 หรือ 24 ชั่วโมง ความคลาดเคลื่อนจะเพิ่มขึ้น เพราะจุดข้อมูลพลาดยอดที่แหลม
5. **ความลึกน้ำท่า** กระจายน้ำท่าโดยตรงบนพื้นที่ลุ่มน้ำ 1250 km²: $\dfrac{40.49\times10^6\ \text{m}^3}{1250\times10^6\ \text{m}^2}=0.0324$ m $=32.4$ mm

```python
Qp, tp, m = 420.0, 18.0, 3.0
Qb = lambda t: 35 + 0.15*t
Q = lambda t: Qb(t) + Qp*(t/tp)**m*np.exp(m*(1 - t/tp))

t_obs = np.arange(0, 121, 3.0)                       # every 3 h for 5 days
Q_obs = Q(t_obs)
sec = 3600.0
V_trap = integrate.trapezoid(Q_obs, t_obs)*sec
V_simp = integrate.simpson(Q_obs, x=t_obs)*sec
V_exact = integrate.quad(Q, 0, 120)[0]*sec
print(f"total volume  : trapezoid {V_trap/1e6:7.2f} | Simpson {V_simp/1e6:7.2f} | exact {V_exact/1e6:7.2f}  million m³")

# exact direct runoff with the antiderivative (substitute s = t/tp, gamma-type integral)
ts = sp.symbols('t', positive=True)
direct_expr = Qp*(ts/tp)**3*sp.exp(3*(1 - ts/tp))
V_direct = float(sp.integrate(direct_expr, (ts, 0, 120)))*sec
V_direct_inf = float(sp.integrate(direct_expr, (ts, 0, sp.oo)))*sec
area_km2 = 1250
print(f"direct runoff : {V_direct/1e6:.2f} million m³ (to t = 120 h), {V_direct_inf/1e6:.2f} million m³ (improper, t → ∞)")
print(f"runoff depth over a {area_km2} km² catchment = {V_direct/(area_km2*1e6)*1000:.1f} mm of water")
```

```python
tt = np.linspace(0, 120, 500)
fig, ax = plt.subplots()
ax.fill_between(tt, Qb(tt), Q(tt), color="C0", alpha=.3, label="direct runoff = ∫(Q − Q_b) dt")
ax.fill_between(tt, 0, Qb(tt), color="C2", alpha=.25, label="baseflow = ∫ Q_b dt")
ax.plot(tt, Q(tt), "C0", lw=2); ax.plot(t_obs, Q_obs, "ko", ms=3, label="gauge readings (every 3 h)")
ax.set(xlabel="time since rain started (h)", ylabel="discharge Q (m³/s)", title="Storm hydrograph"); ax.legend(); plt.show()
```

**การตีความ** แม้มีค่าอ่านเพียงทุก 3 ชั่วโมง กฎสี่เหลี่ยมคางหมู (trapezoid rule) และกฎของซิมป์สัน (Simpson's rule) ให้ผลตรงกับปริพันธ์แม่นตรงภายในประมาณ 0.1 % ดังนั้นการหาปริพันธ์เชิงตัวเลขจากข้อมูลภาคสนามจึงเชื่อถือได้
ในที่นี้น้ำท่าโดยตรงมีประมาณ 40 ล้าน m³ เทียบเท่าชั้นน้ำลึกประมาณ 3 cm ทั่วทั้งพื้นที่รับน้ำ เมื่อเปรียบเทียบกับปริมาณฝนจะได้*สัมประสิทธิ์น้ำท่า (runoff coefficient)* ซึ่งใช้ในการออกแบบเขื่อนและระบบป้องกันน้ำท่วม (เช่น ในลุ่มน้ำเจ้าพระยา)

**ความหมายของคำตอบ** พายุครั้งนี้ทำให้เกิดน้ำท่าประมาณ 40 ล้าน m³ ไหลลงแม่น้ำ เทียบเท่ากับชั้นน้ำหนา 32 mm ทั่วทั้งลุ่มน้ำ เมื่อหารด้วยความลึกฝนจะได้ **สัมประสิทธิ์น้ำท่า (runoff coefficient)** ซึ่งวิศวกรใช้ออกแบบเขื่อน ทางระบายน้ำล้น และระบบป้องกันน้ำท่วม เช่นในลุ่มน้ำเจ้าพระยา ตรวจสอบ: หน่วยคือ (m³/s) × s = m³ และกฎเชิงตัวเลขให้ค่าตรงกับปริพันธ์แม่นตรงเมื่อเก็บข้อมูลถี่พอ

**ข้อผิดพลาดที่พบบ่อย**

- ลืมแปลงชั่วโมงเป็นวินาที (×3600) เมื่อ $Q$ มีหน่วย m³/s และ $t$ มีหน่วยชั่วโมง
- นับการไหลพื้นฐานรวมใน \"น้ำท่าจากพายุ\" ส่วนที่มาจากพายุคือพื้นที่ **ระหว่าง** $Q$ และ $Q_b$
- ในกฎสี่เหลี่ยมคางหมู ให้น้ำหนักจุดปลายเป็น 2 แทนที่จะเป็น 1

**ลองทำดู** เปลี่ยนการเก็บข้อมูลเป็นทุก 12 ชั่วโมง (`np.arange(0, 121, 12.0)`) ค่าประมาณจากกฎสี่เหลี่ยมคางหมูและกฎของซิมป์สันแย่ลงเท่าใด

## ตัวอย่างที่ 45 ⭐⭐⭐: การกัดเซาะและการทับถมของดินถล่ม (พื้นที่ระหว่างเส้นโค้ง (area between curves))

**บริบท** การสำรวจภูมิประเทศสองครั้งของลาดเขา ก่อน ($z_1$) และหลัง ($z_2$) เกิดดินถล่ม ให้โปรไฟล์สองเส้นตามแนวลาด ($x$ หน่วย m):

$$z_1(x) = 120 - 0.4x,\qquad z_2(x) = z_1(x) - 12\sin\!\left(\frac{\pi x}{150}\right)e^{-x/400}\quad (0\le x\le 300).$$

บริเวณที่ $z_1 > z_2$ มีวัสดุ**ถูกพาออกไป** (รอยแผลดินถล่ม) บริเวณที่ $z_2 > z_1$ มีวัสดุ**ทับถม** (ปลายดินถล่ม)
พื้นที่หน้าตัดคือพื้นที่**ระหว่างเส้นโค้ง**: $\displaystyle\int|z_1 - z_2|\,dx$ โดยแบ่งช่วงที่จุดตัดของเส้นโค้ง

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** เมื่อเปรียบเทียบการสำรวจลาดเขาเดียวกันสองครั้ง วัสดุที่ถูกกัดเซาะหรือทับถมที่แต่ละจุดคือระยะห่างแนวดิ่ง $z_1-z_2$ **พื้นที่ระหว่างเส้นโค้ง** (ต่อความกว้าง 1 เมตร) บอกปริมาณวัสดุที่เคลื่อนที่ แต่ระยะห่างนี้เปลี่ยนเครื่องหมาย จึงต้อง **แบ่งปริพันธ์ที่จุดที่เส้นโค้งตัดกัน** ส่วนหนึ่งสำหรับการกัดเซาะ อีกส่วนสำหรับการทับถม

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **หาระยะห่าง** $z_1-z_2=12\sin\!\left(\dfrac{\pi x}{150}\right)e^{-x/400}$ เส้นตรง $120-0.4x$ หักล้างกันหมด
2. **หาจุดที่เส้นโค้งตัดกัน** $e^{-x/400}>0$ เสมอ ระยะห่างจึงเป็นศูนย์เฉพาะเมื่อ $\sin(\pi x/150)=0$: $x=0,150,300$ m บนช่วง $(0,150)$ ไซน์เป็นบวก ($z_1>z_2$ คือการกัดเซาะ) บนช่วง $(150,300)$ เป็นลบ (การทับถม)
3. **ปฏิยานุพันธ์ (หาปริพันธ์โดยการแยกส่วนสองครั้งแบบวนกลับ)** ให้ $a=1/400$ และ $k=\pi/150$: $$\int e^{-ax}\sin kx\,dx=-\frac{e^{-ax}\left(a\sin kx+k\cos kx\right)}{a^2+k^2}+C.$$ ตรวจสอบโดยหาอนุพันธ์
4. **คำนวณแต่ละส่วน** $$A_{\text{erosion}}=12\int_0^{150}e^{-ax}\sin kx\,dx=\frac{12k\left(1+e^{-150a}\right)}{a^2+k^2}=953.2\ \text{m}^2,$$ $$A_{\text{deposit}}=-12\int_{150}^{300}e^{-ax}\sin kx\,dx=\frac{12k\left(e^{-150a}+e^{-300a}\right)}{a^2+k^2}=655.1\ \text{m}^2.$$
5. **ปริมาตร** สำหรับดินถล่มกว้าง 80 m: ถูกกัดเซาะ $953.2\times80\approx76\,253$ m³ ทับถม $655.1\times80\approx52\,408$ m³ การขนดินที่ทับถมออกต้องใช้รถบรรทุกขนาด 10 m³ จำนวน $\lceil52\,408/10\rceil=5241$ เที่ยว สูญเสียสุทธิ $953.2-655.1=298.1$ m² ต่อความกว้าง 1 เมตร

```python
x = sp.symbols('x', real=True)
z1 = 120 - sp.Rational(2, 5)*x
z2 = z1 - 12*sp.sin(sp.pi*x/150)*sp.exp(-x/400)
diff_ = sp.simplify(z1 - z2)
show("z_1 - z_2", diff_)
roots = sorted(set([0.0] + [float(r) for r in sp.solve(sp.sin(sp.pi*x/150), x) if 0 <= r <= 300] + [150.0, 300.0]))
print("intersection points (m):", roots)

eroded = sp.integrate(diff_, (x, 0, 150))             # z1 > z2 on (0,150)
deposited = -sp.integrate(diff_, (x, 150, 300))       # z2 > z1 on (150,300)
print(f"eroded area    = {float(eroded):7.1f} m² (per metre width)")
print(f"deposited area = {float(deposited):7.1f} m²")
print(f"net change     = {float(eroded - deposited):7.1f} m²  (positive -> material left the profile, e.g. washed into a river)")
show(r"\int_0^{150}(z_1-z_2)\,dx", sp.simplify(eroded))
```

```python
xx = np.linspace(0, 300, 600)
Z1, Z2 = sp.lambdify(x, z1)(xx), sp.lambdify(x, z2)(xx)
fig, ax = plt.subplots(figsize=(9, 4.2))
ax.plot(xx, Z1, "k", lw=1.5, label="before (z₁)"); ax.plot(xx, Z2, "C3", lw=2, label="after (z₂)")
ax.fill_between(xx, Z1, Z2, where=Z1 > Z2, color="C1", alpha=.5, label=f"erosion {float(eroded):.0f} m²")
ax.fill_between(xx, Z1, Z2, where=Z2 >= Z1, color="C0", alpha=.5, label=f"deposition {float(deposited):.0f} m²")
ax.set(xlabel="distance down slope x (m)", ylabel="elevation (m)", title="Landslide cut-and-fill from repeat surveys"); ax.legend(); plt.show()
```

**การตีความ** มีวัสดุถูกพาออกไปประมาณ 950 m² ต่อความกว้าง 1 เมตร และทับถม 655 m² ดังนั้นโดยรวมดินถล่มสูญเสียวัสดุ (บางส่วนไหลลงลำธาร หรือเศษดินหินมีการขยายตัวต่างกัน)
เมื่อคูณด้วยความกว้างของดินถล่มจะได้ปริมาตร การคำนวณ "DEM of difference" แบบนี้ใช้จริงกับข้อมูล LiDAR หรือโดรนที่สำรวจซ้ำหลังเกิดดินถล่มในภาคเหนือของไทย

**ความหมายของคำตอบ** วัสดุที่ออกจากรอยแผลถล่ม (953 m²) มากกว่าที่ทับถมที่ตีนลาด (655 m²) ดังนั้นประมาณ 300 m² ต่อความกว้าง 1 เมตรออกไปจากแนวสำรวจ เช่นถูกลำธารพาไปหรือถูกอัดตัวต่างกัน การคำนวณ \"ผลต่างของ DEM\" แบบนี้ใช้กับการสำรวจด้วยโดรนหรือไลดาร์ซ้ำหลังเกิดดินถล่มในภาคเหนือของไทย เพื่อประมาณปริมาตรและค่าใช้จ่ายในการฟื้นฟู ตรวจสอบ: ถ้าหาปริพันธ์ $z_1-z_2$ ตลอดช่วง 0–300 m โดยไม่แบ่ง จะได้แค่ค่าสุทธิ 298 m² และพลาดทั้งการกัดเซาะและการทับถม

**ข้อผิดพลาดที่พบบ่อย**

- หาปริพันธ์ $z_1-z_2$ ตลอดช่วงโดยไม่แบ่ง ทำให้การกัดเซาะและการทับถมหักล้างกัน
- รายงานพื้นที่การทับถมเป็นค่าลบ พื้นที่ระหว่างเส้นโค้งเป็นบวกเสมอ (หาปริพันธ์ของเส้นบนลบเส้นล่าง)
- เรียกผลลัพธ์ว่าปริมาตร ซึ่งยังเป็นพื้นที่ต่อความกว้าง 1 เมตร (m²) จนกว่าจะคูณด้วยความกว้างของดินถล่ม

**ลองทำดู** ถ้าดินถล่มกว้าง 80 m (และโปรไฟล์เหมือนกันตลอดความกว้าง) หินถูกกัดเซาะไปกี่ m³ ต้องใช้รถบรรทุกขนาด 10 m³ กี่คันเพื่อขนวัสดุที่ทับถมออก

## ตัวอย่างที่ 46 ⭐⭐⭐: โลกหนาแน่นกว่าที่ใจกลางหรือไม่ (ปริพันธ์จำกัดเขตบนทรงกลม)

**บริบท** มีตัวเลขสองค่าที่ทราบอย่างแม่นยำจากวงโคจรและการหมุนของโลก คือความหนาแน่นเฉลี่ยของโลก 5514 kg/m³ และ **ตัวประกอบโมเมนต์ความเฉื่อย (moment of inertia factor)** $I/(MR^2) = 0.3307$ ทรงกลมที่เนื้อเดียวกันจะมีค่า 0.4
ลองใช้แบบจำลองความหนาแน่นที่ลดลงจากใจกลางออกไป: $\rho(r) = \rho_0\left(1-\alpha\dfrac{r^2}{R^2}\right)$

**แคลคูลัส** หาปริพันธ์ผ่านเปลือกทรงกลม:
$$M = \int_0^R 4\pi r^2\rho(r)\,dr = 4\pi\rho_0R^3\left(\frac13-\frac\alpha5\right),\qquad I = \int_0^R\frac{8\pi}{3}r^4\rho(r)\,dr = \frac{8\pi}{3}\rho_0R^5\left(\frac15-\frac\alpha7\right).$$

### คำอธิบายโดยละเอียด

**แนวคิดหลัก** มวลและโมเมนต์ความเฉื่อยต่างเป็นผลรวมของเปลือกทรงกลมบาง ๆ จึงเป็นปริพันธ์จำกัดเขต (definite integral) ของความหนาแน่นคูณกำลังของ $r$ เนื่องจากโมเมนต์ความเฉื่อยให้น้ำหนักกับเปลือกชั้นนอกมากกว่ามาก ($r^4$ แทน $r^2$) การเปรียบเทียบสองค่านี้จึงบอกการกระจายมวลภายในโลก ทั้งที่เราเก็บตัวอย่างได้ลึกเพียงไม่กี่กิโลเมตร

**วิธีทำทีละขั้น** (ใช้ค่าเริ่มต้นเดียวกับในโค้ดด้านล่าง):

1. **ชิ้นเปลือก** เปลือกรัศมี $r$ หนา $dr$ มีปริมาตร $4\pi r^2dr$ และมวล $dm=4\pi r^2\rho(r)\,dr$ โมเมนต์ความเฉื่อยรอบแกนคือ $\tfrac23r^2\,dm$
2. **มวล** เมื่อ $\rho=\rho_0(1-\alpha r^2/R^2)$: $M=4\pi\rho_0\displaystyle\int_0^R\left(r^2-\frac{\alpha r^4}{R^2}\right)dr=4\pi\rho_0\left(\frac{R^3}3-\frac{\alpha R^3}5\right)=4\pi\rho_0R^3\left(\frac13-\frac\alpha5\right)$
3. **โมเมนต์ความเฉื่อย** $I=\dfrac{8\pi}3\rho_0\displaystyle\int_0^R\left(r^4-\frac{\alpha r^6}{R^2}\right)dr=\frac{8\pi}3\rho_0R^5\left(\frac15-\frac\alpha7\right)$
4. **อัตราส่วน** ($\rho_0$ และ $R$ ตัดกัน): $\dfrac{I}{MR^2}=\dfrac23\cdot\dfrac{\frac15-\frac\alpha7}{\frac13-\frac\alpha5}$ สำหรับดาวที่เนื้อเดียวกัน ($\alpha=0$) ได้ $\dfrac23\cdot\dfrac35=0.4$
5. **จับคู่กับค่าวัดได้ 0.3307** ลอง $\alpha=0.794$: $\dfrac15-\dfrac{0.794}7\approx0.08657$, $\dfrac13-\dfrac{0.794}5\approx0.17453$ อัตราส่วน $\approx\dfrac23\times0.4960\approx0.3307$ ✓
6. **ความหนาแน่น** ความหนาแน่นเฉลี่ยคือ $\dfrac{M}{\frac43\pi R^3}=3\rho_0\left(\frac13-\frac\alpha5\right)=\rho_0(1-0.6\alpha)=5514$ kg/m³ ดังนั้น $\rho_0=\dfrac{5514}{1-0.4764}\approx10{,}530$ kg/m³ ที่ใจกลาง และ $\rho_0(1-\alpha)\approx2170$ kg/m³ ที่ผิว มวลรวมคือ $\tfrac43\pi R^3\times5514\approx5.97\times10^{24}$ kg

```python
r, R, rho0, al = sp.symbols('r R rho_0 alpha', positive=True)
rho = rho0*(1 - al*r**2/R**2)
M = sp.integrate(4*sp.pi*r**2*rho, (r, 0, R)); I = sp.integrate(sp.Rational(8, 3)*sp.pi*r**4*rho, (r, 0, R))
show("M", sp.factor(M)); show("I", sp.factor(I))
ratio = sp.simplify(I/(M*R**2))
show(r"\frac{I}{MR^2}", ratio)
print("uniform sphere (α = 0):", ratio.subs(al, 0))
alpha_v = float(sp.nsolve(ratio - 0.3307, al, 0.8))
rho0_v = 5514/(1 - 3*alpha_v/5)
print(f"α = {alpha_v:.4f}  ->  central density ρ0 = {rho0_v:.0f} kg/m³, surface density ρ(R) = {rho0_v*(1 - alpha_v):.0f} kg/m³")
```

```python
rr = np.linspace(0, 1, 200)
fig, ax = plt.subplots()
ax.plot(rr*6371, rho0_v*(1 - alpha_v*rr**2), lw=2, label=f"ρ(r) model, α = {alpha_v:.2f}")
ax.axhline(5514, c="gray", ls="--", label="uniform 5514 kg/m³")
ax.axvspan(0, 3480, color="C1", alpha=.1, label="real core (r < 3480 km)")
ax.set(xlabel="radius r (km)", ylabel="density (kg/m³)", title="A density model that matches M and I"); ax.legend(); plt.show()
```

**การตีความ** การทำให้ $I/(MR^2) = 0.3307$ ต้องใช้ $\alpha\approx0.79$ ซึ่งให้ความหนาแน่นที่ใจกลางประมาณ **10,500 kg/m³** ราวสองเท่าของหินที่ผิว (≈ 2,200 kg/m³ ในแบบจำลองเรียบนี้)
ดังนั้นมวลต้องรวมตัวอยู่ทางใจกลาง นี่เป็นหลักฐานยุคแรกก่อนวิชาวิทยาแผ่นดินไหวว่าโลกมีแก่นเหล็กที่หนาแน่น (ความหนาแน่นจริงที่ใจกลาง ≈ 13,000 kg/m³)

**ความหมายของคำตอบ** ค่าที่ต่ำกว่า 0.4 เกิดได้ก็ต่อเมื่อมวลกระจุกอยู่ใกล้ใจกลาง แบบจำลองง่าย ๆ นี้ต้องการใจกลางที่หนาแน่นราวสองเท่าของค่าเฉลี่ย และความหนาแน่นที่ผิวใกล้เคียงหินเปลือกโลกจริง (ประมาณ 2700 kg/m³) โลกจึงต้องมีแก่นที่หนาแน่น ซึ่งเรารู้ว่าเป็นเหล็ก แบบจำลองคลื่นไหวสะเทือน (PREM) ปรับค่านี้เป็นประมาณ 13,000 kg/m³ ในแก่นชั้นใน

**ข้อผิดพลาดที่พบบ่อย**

- ใช้สูตรทรงกลมตัน $\tfrac25MR^2$ กับแต่ละเปลือก เปลือกบางมีโมเมนต์ความเฉื่อย $\tfrac23r^2\,dm$
- หาปริพันธ์ $\rho(r)\,dr$ แทน $4\pi r^2\rho(r)\,dr$ ต้องคูณด้วยพื้นที่ผิวของแต่ละเปลือก
- ถือว่า $R$ ใน $\alpha r^2/R^2$ เป็นตัวแปร $R$ เป็นค่าคงตัว ตัวแปรที่หาปริพันธ์คือ $r$ เท่านั้น

**ลองทำดู** ดาวอังคารมี $I/(MR^2) = 0.3639$ เมื่อใช้แบบจำลองเดียวกัน มวลของดาวอังคารรวมตัวที่ใจกลางมากกว่าหรือน้อยกว่าโลก
